# ۱. معرفی پروژه و پرسش‌های راهبردی

## ۱.۱. مسئله کسب‌وکار

در بازار کار ایران، آگهی استخدامی هم‌زمان سه کار انجام می‌دهد: تقاضای کارفرما را اعلام می‌کند، سطح انتظار او از مهارت و سابقه را نشان می‌دهد و در صورت اعلام حقوق، قیمت پیشنهادی نیروی کار را آشکار می‌کند. مجموع این آگهی‌ها در طول زمان، تصویری کمّی از «چه کسی، با چه مهارتی، در کجا و با چه قیمتی» استخدام می‌کند می‌سازد.

جاب‌ویژن به‌عنوان یکی از بزرگ‌ترین پلتفرم‌های کاریابی ایران، داده آگهی‌های خود را به‌صورت عمومی منتشر کرده است. این داده فرصتی کمیاب برای تحلیل بازار کار فراهم می‌کند، اما خام است: ۴۳ ستون با انواع مختلف، متن آزاد فارسی، ساختارهای JSON، مقادیر گمشده و یک محدودیت بنیادی — تنها حدود یک‌سوم آگهی‌ها حقوق را اعلام کرده‌اند.

هدف این پروژه تبدیل این داده خام به **بینش قابل تصمیم‌گیری** برای سه مخاطب است:

- **مدیر منابع انسانی یا صاحب کسب‌وکار:** با چه حقوق و مزایایی نیرو جذب کنیم؟ آگهی را چه زمانی و با چه شرطی منتشر کنیم؟ کدام مهارت واقعاً ارزش افزوده دارد؟
- **کارجو:** کدام مهارت، مدرک، سابقه و جغرافیا بیشترین بازده مالی را دارد؟ شکاف جنسیتی و استانی چقدر است؟
- **سیاست‌گذار و تحلیلگر بازار کار:** تقاضا در چه فصولی اوج می‌گیرد؟ کدام بخش‌ها در حال رشد یا کوچک‌شدن‌اند؟ تمرکز جغرافیایی چگونه تغییر کرده است؟

## ۱.۲. نقش ما در این پروژه

ما در نقش تیم تحلیل داده جاب‌ویژن عمل می‌کنیم. مدیرعامل و مدیر منابع انسانی یک شرکت فرضی «روبه‌رشد» از ما خواسته‌اند پیش از تدوین برنامه استخدام سال آینده، تصویری مستند از بازار ارائه دهیم. بنابراین هر تحلیل باید به یکی از این تصمیم‌ها گره بخورد: **چه کسی را با چه قیمتی، در چه زمانی و با چه شرطی جذب کنیم؟**

## ۱.۳. چهار پرسش مدیریتی

صورت پروژه چهار پرسش کلان را مطرح می‌کند. این پرسش‌ها به شاخص‌های قابل اندازه‌گیری شکسته شده‌اند تا پاسخ‌ها قابل بررسی و رد یا تأیید باشند.

**جدول ۱.۱ | نقشه پرسش‌های مدیریتی، شاخص‌ها و بخش‌های پاسخ**

| پرسش مدیریتی | شاخص و مبنای سنجش | بخش‌های پاسخ |
| --- | --- | --- |
| بسته جبران خدمات رقابتی برای هر رده شغلی و استان چیست؟ | میانه و چارک‌های حقوق ماهانه و درآمد ساعتی به تفکیک دسته شغلی، استان، سابقه و مدرک | ۷، ۱۰، ۱۲، ۱۳، ۱۴، ۲۲ |
| تقاضا در چه فصلی اوج می‌گیرد و کدام مهارت‌ها ارزش افزوده مالی دارند؟ | سری زمانی ماهانه و شمسی، شاخص فصلی، صرفه قیمتی مهارت‌ها، رشد تقاضای مهارت | ۱۰، ۱۹، ۲۲ |
| ساختار رقابت کارفرمایان چگونه است و چگونه آگهی متمایز می‌شود؟ | نرخ افشای حقوق، تمرکز استانی و صنعتی، تیپ‌شناسی کارفرما، مزایای متمایزکننده | ۹، ۱۴، ۲۰، ۲۲ |
| داده چه ظرفیت‌هایی برای خلق ارزش در ذی‌نفعان دارد؟ | ترکیب یافته‌ها در یک بسته پیشنهادی و داشبورد تصمیم | ۲۲، ۲۳ |

**توضیح جدول ۱.۱:** این جدول خودش نتیجه تحلیل نیست؛ نقشه راه است. هر شاخص واحد، نمونه و دوره مشخصی دارد و در بخش مربوط به‌صورت جداگانه تعریف می‌شود. برای نمونه، «میانه حقوق» فقط روی آگهی‌های دارای حقوق اعلام‌شده محاسبه می‌شود، در حالی که «نرخ افشای حقوق» روی کل آگهی‌ها اندازه‌گیری می‌شود؛ این دو را نمی‌توان به‌جای هم خواند.

## ۱.۴. رویکرد تحلیل

مسیر پروژه از داده خام به تصمیم، پنج گام دارد:

1. **شناخت و کنترل کیفیت:** اندازه، انواع، گمشدگی، تکرار، بازه و کاردینالیتی. بدون این گام، هر عدد بعدی ممکن است بی‌معنا باشد.
2. **ساخت لایه تحلیل:** یک جدول واحد در سطح آگهی می‌سازیم که همه ویژگی‌های عددی، دسته‌ای و متنی در آن استخراج شده‌اند. جدول‌های کمکی مهارت، تحصیلات و مزایا نیز یک‌بار ساخته و در همه بخش‌ها بازاستفاده می‌شوند تا محاسبات تکراری نشود.
3. **تحلیل اکتشافی و آماری:** توزیع، همبستگی، آزمون فرض و تحلیل سازوکار گمشدگی.
4. **مدل‌سازی و یادگیری بدون نظارت:** رگرسیون حقوق برای شناسایی عوامل تعیین‌کننده، خوشه‌بندی عناوین شغلی و شبکه مهارت‌ها.
5. **ترجمه به تصمیم:** جدول بسته جبران خدمات، تقویم انتشار آگهی و چک‌لیست متن آگهی.

## ۱.۵. چند اصطلاح پایه

- **آگهی:** یک ردیف داده. شناسه آن `ad_id` است. یک شرکت می‌تواند چند آگهی داشته باشد؛ بنابراین شمارش آگهی با شمارش شرکت برابر نیست.
- **افشای حقوق:** `Jobpost_SalaryCanBeShown == True` یعنی آگهی بازه حقوق را اعلام کرده است. در غیر این صورت مقدار حقوق خالی است، نه صفر.
- **حقوق ماهانه:** نقطه میانی بازه حقوق اعلام‌شده، به واحد **میلیون تومان در ماه**. این عدد مبلغ قرارداد نیست؛ قیمت پیشنهادی آگهی است.
- **درآمد ساعتی:** حقوق ماهانه تقسیم بر ساعات کار ماهانه برآوردشده از متن «ساعات کاری». این شاخص برای مقایسه مشاغل تمام‌وقت، پاره‌وقت و پروژه‌ای لازم است، چون حقوق ماهانه به‌تنهایی ساعت کار را نشان نمی‌دهد.
- **دسته شغلی:** `Jobpost_MainJobCategory`، تخصص اصلی آگهی (مثلاً «توسعه نرم‌افزار و برنامه‌نویسی»). **صنعت:** `Jobpost_IndustryFa`، حوزه فعالیت کارفرما (مثلاً «فناوری اطلاعات / نرم‌افزار و سخت‌افزار»). این دو یکی نیستند: یک شرکت صنعتی می‌تواند برنامه‌نویس استخدام کند.
- **ترجیح جنسیت:** متن آگهی، نه جنسیت نیروی استخدام‌شده. «فقط آقا» یعنی کارفرما استخدام مرد را شرط کرده است.
- **بازه سنی:** حداقل و حداکثر سن درخواستی کارفرما، نه سن واقعی کارجویان.
- **پرت:** مقدار دور از بدنه توزیع. پرت را حذف نمی‌کنیم؛ آن را علامت می‌زنیم، در توصیف نگه می‌داریم و فقط در مدل‌سازی با قاعده شفاف محدود می‌کنیم.

## ۱.۶. از عدد تا بینش

در این نوت‌بوک هر ادعا سه جزء دارد: **عدد**، **مقایسه** و **محدودیت**. مثلاً «میانه حقوق دسته فناوری اطلاعات ۱۴٫۵ میلیون تومان و ۳۸٪ بالاتر از میانه کل بازار است» یک بینش است؛ عدد ۱۴٫۵ به‌تنهایی اطلاعات نیست. در پایان هر بخش، جمله‌های «پاسخ» و «جمع‌بندی کسب‌وکاری» همین ساختار را دارند.

> **قاعده طلایی:** هیچ شاخصی را بدون مخرج و نمونه گزارش نمی‌کنیم. هرجا نمونه کوچک باشد، هشدار می‌دهیم؛ هرجا تفاوت معنادار نباشد، آن را می‌گوییم.

## ۲.۲. دیکشنری متغیرهای آگهی

**جدول ۲.۲ | متغیرهای آگهی، شرایط احراز و مزایا**

| گروه | ستون | مفهوم | نوع | نقش |
| --- | --- | --- | --- | --- |
| شناسه | `RowNumber` → `ad_id` | شماره یکتای ردیف | عدد | شناسه |
| زمان | `Jobpost_ActivationTime_YEAR_MONTH` | ماه انتشار آگهی | تاریخ ماهانه | محور زمان |
| عنوان | `Jobpost_RawTitle` | عنوان خام شغلی | متن آزاد | تحلیل عنوان |
| جنسیت | `Jobpost_PreferredGender` | ترجیح یا شرط جنسیت | دسته‌ای ۵ سطح | تحلیل شکاف |
| مکان | `Jobpost_ProvinceFa/En` | استان محل آگهی | دسته‌ای | تحلیل جغرافیایی |
| همکاری | `Jobpost_WorkTypeFa/En` | نوع همکاری | دسته‌ای ۴ سطح | تحلیل ساعت و درآمد |
| افشا | `Jobpost_SalaryCanBeShown` | آیا حقوق اعلام شده است | بولین | کلید زیرمجموعه حقوقی |
| حقوق | `Jobpost_MinSalary` | حداقل بازه پیشنهادی (میلیون تومان/ماه) | عدد | هدف و شاخص |
| حقوق | `Jobpost_MaxSalary` | حداکثر بازه پیشنهادی (میلیون تومان/ماه) | عدد | هدف و شاخص |
| شغل | `Jobpost_MainJobCategory` | دسته شغلی اصلی | دسته‌ای ۷۷ سطح | محور اصلی تحلیل |
| شغل | `Jobpost_SecondaryJobCategories` | دسته‌های فرعی، جداشده با خط تیره | متن چندمقداری | تحلیل ترکیبی |
| صنعت | `Jobpost_IndustryFa/En` | صنعت آگهی | دسته‌ای ۴۴ سطح | مقایسه بخشی |
| سابقه | `Jobpost_RequiredExperienceYears` | سال سابقه لازم | عدد صحیح | تعیین‌کننده حقوق |
| سابقه | `Jobpost_RequiredRelatedExperienceInThisIndustry` | الزام سابقه مرتبط | بولین | تعیین‌کننده حقوق |
| ساعت | `Jobpost_CompanyWorkTimesDescription` | شرح ساعات کاری | متن آزاد | استخراج ساعت هفتگی |

## ۲.۴. آنچه در داده نیست

دقت در «نبودها» به‌اندازه دقت در «بودها» مهم است:

- **متن کامل آگهی و شرح شغل** وجود ندارد؛ فقط عنوان، مهارت‌ها، مزایا و شرایط احراز.
- **نام شرکت** ناشناس‌سازی شده است؛ تحلیل رقابت در سطح ویژگی شرکت انجام می‌شود، نه برند.
- **تاریخ دقیق انتشار** وجود ندارد؛ فقط ماه.
- **اطلاعات کارجو، استخدام موفق، حقوق قرارداد واقعی و نرخ ماندگاری** وجود ندارد.
- **تعدیل تورم** در داده نیست؛ مقادیر اسمی‌اند و مقایسه زمانی باید با شاخص تورم تفسیر شود.
- **دستمزد ساعتی** مستقیماً ثبت نشده است؛ باید از حقوق ماهانه و ساعت کاری استخراج شود.

## ۲.۵. واحد حقوق: چرا میلیون تومان؟

کارت داده واحد حقوق را «Toman» نوشته است، اما بازه مقادیر (۲ تا ۱۰۰ با میانه ۱۰) با «تومان» خام سازگار نیست: حقوق ماهانه ۱۰ تومان بی‌معناست. سه شاهد نشان می‌دهد مقیاس واقعی **میلیون تومان در ماه** است:

1. میانه حداقل حقوق اعلام‌شده ۱۰ و میانه حداکثر ۱۲ است؛ با سطح دستمزد بازار در ۱۴۰۱–۱۴۰۲ هم‌خوان است.
2. صدک ۹۹ حداقل حقوق ۳۵ و بیشینه ۷۰ است؛ بازه‌ای که برای مدیران و مشاغل تخصصی در همان سال منطقی است.
3. اگر مقیاس «هزار تومان» بود، میانه ۱۰ هزار تومان در ماه با هیچ استانداردی سازگار نبود.

این فرض در سراسر نوت‌بوک یکسان اعمال می‌شود. اگر مقیاس داده تغییر کند، همه اعداد حقوقی باید بازمحاسبه شوند.

# ۳. ابزارها، توابع کمکی و بارگذاری داده

این بخش زیرساخت تحلیل را می‌سازد و یک‌بار برای همیشه اجرا می‌شود. همه جدول‌ها با یک تابع (`show_table`)، همه اعداد کلیدی با یک قالب (`show_metrics`) و همه نمودارها با یک سبک رنگی نمایش داده می‌شوند تا خواننده بین بخش‌های مختلف سردرگم نشود.

- `fa` متن فارسی را برای نمایش در نمودار آماده می‌کند (شکل‌دهی حروف و ترتیب راست‌به‌چپ).
- `RTLTable` و `show_table` جدول را با شماره، عنوان، نمونه و دوره نمایش می‌دهند و واحد هر ستون را در عنوان آن نگه می‌دارند.
- `MetricBlock` و `show_metrics` عددهای کلیدی هر بخش را جدا می‌کنند تا در متن قابل ارجاع باشند.
- `AnswerText` و `show_answer` پاسخ نهایی هر پرسش را با شواهد عددی نشان می‌دهند.
- `complete` نمودار را با شماره یکتا در `outputs/figures/` ذخیره و در نوت‌بوک نمایش می‌دهد؛ همین فایل‌ها در گزارش PDF استفاده می‌شوند.

در پایان سلول، داده خام خوانده می‌شود و در صورت نبود فایل، مسیر جایگزین بررسی و راهنمای دانلود نمایش داده می‌شود.

```python
%matplotlib inline
from pathlib import Path
import json
import math
import re
import time
import unicodedata
from html import escape

import numpy as np
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt
from matplotlib import ft2font
from matplotlib.ticker import FuncFormatter, PercentFormatter
import seaborn as sns
from packaging.version import Version
import arabic_reshaper
from bidi.algorithm import get_display
from IPython.display import display, HTML

import scipy
import sklearn
import statsmodels
import networkx as nx

pd.set_option('display.max_rows', 250)
pd.set_option('display.max_columns', 50)
pd.set_option('display.width', 220)

# پالت یکدست پروژه
TEAL, DARK, GOLD = '#2f7f7a', '#263746', '#c9974b'
RED, GRAY, LIGHT = '#b5534b', '#7b8c8c', '#e8f2f1'
BLUE, PURPLE, GREEN = '#5b8db8', '#9c7bb5', '#7aa974'
PALETTE = [TEAL, GOLD, BLUE, PURPLE, GREEN, RED, '#c9a227', GRAY]

sns.set_theme(style='whitegrid', font='DejaVu Sans', rc={
    'axes.spines.top': False, 'axes.spines.right': False,
    'grid.color': '#e3eceb', 'axes.edgecolor': '#c3d4d2',
    'axes.labelcolor': DARK, 'text.color': DARK,
    'xtick.color': GRAY, 'ytick.color': GRAY,
    'axes.unicode_minus': False, 'figure.dpi': 110,
    'figure.facecolor': 'white', 'savefig.facecolor': 'white',
})


def fa(value):
    """آماده‌سازی متن فارسی برای matplotlib."""
    value = str(value)
    value = re.sub(r'(?<![0-9])\d{4}-\d{2}(?:-\d{2})?(?![0-9])',
                   lambda m: '\u200e' + m.group() + '\u200e', value)
    if Version(matplotlib.__version__) >= Version('3.11') and getattr(ft2font, '__libraqm_version__', ''):
        return value
    return '\n'.join(get_display(arabic_reshaper.reshape(line)) for line in value.split('\n'))


_FA_DIGITS = str.maketrans('0123456789,', '۰۱۲۳۴۵۶۷۸۹٬')
_FA_NUM = str.maketrans('0123456789', '۰۱۲۳۴۵۶۷۸۹')


def fa_num(value):
    """رقم فارسی بدون جداکننده هزارگان (برای سال و شماره)."""
    return str(value).translate(_FA_NUM)


def nf(value, decimals=0):
    """عدد فارسی با جداکننده هزارگان."""
    if value is None or (isinstance(value, float) and math.isnan(value)):
        return '—'
    return f'{value:,.{decimals}f}'.translate(_FA_DIGITS)


def pf(value, decimals=1):
    return nf(value, decimals) + '٪'


def _cell(value, kind='auto'):
    if value is None:
        return '—'
    try:
        if pd.isna(value):
            return '—'
    except (TypeError, ValueError):
        pass
    if kind == 'int':
        return f'{int(round(float(value))):,}'
    if kind == 'money':
        return f'{float(value):,.1f}'
    if kind == 'money2':
        return f'{float(value):,.2f}'
    if kind == 'pct':
        return f'{float(value):,.1f}٪'
    if kind == 'pp':
        return f'{float(value):+,.1f}'
    if kind == 'score':
        return f'{float(value):.2f}'
    if kind == 'hours':
        return f'{float(value):,.1f}'
    if kind == 'hourly':
        return f'{float(value):,.0f}'
    if kind == 'ratio':
        return f'{float(value):.3f}'
    if kind == 'pvalue':
        return f'{float(value):.2e}'
    if kind == 'signed':
        return f'{int(round(float(value))):+d}'
    if kind == 'bool':
        return 'بله' if bool(value) else 'خیر'
    if kind == 'auto':
        if isinstance(value, (bool, np.bool_)):
            return 'بله' if value else 'خیر'
        if isinstance(value, (int, np.integer)):
            return f'{int(value):,}'
        if isinstance(value, (float, np.floating)):
            return f'{float(value):,.2f}'
        return str(value)
    return f'{value:,.1f}'


class RTLTable:
    def __init__(self, caption, frame, note=None):
        self.caption, self.frame, self.note = caption, frame, note

    def __repr__(self):
        return self.caption + '\n' + self.frame.to_string()

    def _repr_html_(self):
        table = self.frame.to_html(border=0, classes='jv')
        note = f'<div class="jv-note">{escape(self.note)}</div>' if self.note else ''
        return (f'<div class="jv-wrap"><div class="jv-caption">{escape(self.caption)}</div>'
                f'{table}{note}</div>')


def show_table(number, title, frame, formats=None, labels=None, sample=None, period=None,
               index_name=None, note=None):
    """نمایش یکدست جدول با شماره، نمونه و دوره."""
    out = frame.to_frame() if isinstance(frame, pd.Series) else frame.copy()
    assert out.shape[1] + out.index.nlevels <= 9, (number, out.shape)
    formats, labels = formats or {}, labels or {}
    for col in out.columns:
        kind = formats.get(col, 'auto')
        out[col] = out[col].map(lambda v: _cell(v, kind))
    if index_name:
        out.index = out.index.rename(index_name)
    out = out.rename(columns={key: labels.get(value, value) for key, value in labels.items()})
    assert out.columns.is_unique, number
    caption = f'جدول {number} | {title}'
    if sample:
        caption += f' | {sample}'
    if period is not False:
        caption += f' | {period or PERIOD_LABEL}'
    display(RTLTable(caption, out, note))


class MetricBlock:
    def __init__(self, number, rows):
        self.number, self.rows = number, rows

    def __repr__(self):
        return f'عددهای کلیدی {self.number}\n' + '\n'.join(f'{k}: {v}' for k, v in self.rows)

    def _repr_html_(self):
        items = ''.join(
            f'<li>{escape(str(label))}: <bdi dir="ltr"><strong>{escape(str(value))}</strong></bdi></li>'
            for label, value in self.rows)
        return (f'<div class="jv-metrics"><div class="jv-metrics-title">عددهای کلیدی {escape(str(self.number))}</div>'
                f'<ul>{items}</ul></div>')


def show_metrics(number, rows):
    display(MetricBlock(number, rows))


class AnswerText:
    def __init__(self, title, paragraphs):
        self.title, self.paragraphs = title, paragraphs

    def __repr__(self):
        return self.title + '\n\n' + '\n\n'.join(self.paragraphs)

    def _repr_html_(self):
        body = ''.join(f'<p>{escape(str(p))}</p>' for p in self.paragraphs)
        return (f'<div class="jv-answer"><div class="jv-answer-title">{escape(self.title)}</div>'
                f'{body}</div>')


def show_answer(title, paragraphs):
    display(AnswerText(title, paragraphs))


def heading(ax, number, title, period=None):
    ax.set_title(fa(f'نمودار {number} | {title} | {period or PERIOD_LABEL}'), fontsize=10.5, pad=12)


def axes_labels(ax, x, y):
    ax.set_xlabel(fa(x))
    ax.set_ylabel(fa(y))


def sync_snapshots():
    """بازسازی جدول‌های مشتق از ads پس از افزودن ستون‌های تازه.

    paid: آگهی‌های دارای حقوق اعلام‌شده؛ hourly_ready: آگهی‌های دارای حقوق و ساعت معتبر.
    این تابع تضمین می‌کند بخش‌های بعدی روی جدیدترین نسخه ads کار کنند.
    """
    globals()['paid'] = ads.loc[ads.salary_disclosed].copy()
    globals()['hourly_ready'] = ads.loc[ads.salary_disclosed & ads.hourly_income.notna()].copy()


def complete(number=None):
    """مرتب‌سازی، ذخیره و نمایش نمودار جاری."""
    fig = plt.gcf()
    fig.tight_layout()
    if number:
        for attempt in range(3):
            try:
                FIG_DIR.mkdir(parents=True, exist_ok=True)
                fig.savefig(FIG_DIR / f'fig_{str(number).replace(".", "_")}.png',
                            dpi=160, bbox_inches='tight', facecolor='white')
                break
            except OSError as error:
                if attempt == 2:
                    print(f'هشدار: ذخیره نمودار {number} ناموفق بود ({error})')
                time.sleep(0.6)
    plt.show()


def bar_labels(ax, fmt='{:,.0f}', horizontal=False, padding=2, fontsize=8.5):
    for patch in ax.patches:
        if horizontal:
            value = patch.get_width()
            ax.annotate(fmt.format(value), (value, patch.get_y() + patch.get_height() / 2),
                        xytext=(padding, 0), textcoords='offset points', va='center', fontsize=fontsize)
        else:
            value = patch.get_height()
            ax.annotate(fmt.format(value), (patch.get_x() + patch.get_width() / 2, value),
                        xytext=(0, padding), textcoords='offset points', ha='center', fontsize=fontsize)


def pct_axis(ax, decimals=0):
    ax.yaxis.set_major_formatter(PercentFormatter(xmax=100, decimals=decimals))


display(HTML("""
<style>
.jp-RenderedMarkdown, .text_cell_render, .rendered_html { direction: rtl; text-align: right; }
.jp-RenderedMarkdown ul, .jp-RenderedMarkdown ol { padding-right: 1.5em; padding-left: 0; }
.jp-RenderedMarkdown pre, .jp-RenderedMarkdown code { direction: ltr; text-align: left; }
.jp-RenderedMarkdown table { direction: rtl; }
.jv-wrap { direction: rtl; text-align: right; margin: 6px 0 16px; overflow-x: auto; }
.jv-caption { font-size: 13px; font-weight: 700; color: #1f4b48; margin: 2px 0 7px; }
.jv-note { font-size: 12px; color: #5c6f6d; margin-top: 5px; }
.jv-wrap table.jv { border-collapse: collapse; width: 100%; font-size: 12.5px; background: #fff; }
.jv-wrap table.jv thead th { background: #e8f2f1; color: #173b37; font-weight: 700;
  border-bottom: 2px solid #b9d6d3; padding: 7px 10px; white-space: nowrap; }
.jv-wrap table.jv tbody td { border-bottom: 1px solid #e6eeed; padding: 6px 10px; vertical-align: top; }
.jv-wrap table.jv tbody tr:nth-child(even) td { background: #f8fbfa; }
.jv-wrap table.jv tbody tr:hover td { background: #eef7f5; }
.jv-metrics { direction: rtl; text-align: right; border-right: 4px solid #2f7f7a; background: #f1f8f6;
  padding: 8px 14px; border-radius: 6px; margin: 10px 0 14px; font-size: 13px; }
.jv-metrics-title { font-weight: 700; color: #173b37; margin-bottom: 3px; }
.jv-metrics ul { margin: 3px 0 0; padding-right: 18px; }
.jv-answer { direction: rtl; text-align: right; border-right: 4px solid #c9974b; background: #fdf8ef;
  padding: 8px 14px; border-radius: 6px; margin: 10px 0 14px; font-size: 13px; }
.jv-answer-title { font-weight: 700; color: #6b4a13; margin-bottom: 3px; }
.jv-answer p { margin: 4px 0; }
</style>
"""))


# ---------------------------------------------------------------------------
# پیکربندی پروژه
# ---------------------------------------------------------------------------
PERIOD_START = '2022-04'
PERIOD_END = '2024-02'
PERIOD_LABEL = 'فروردین ۱۴۰۱ تا بهمن ۱۴۰۲'
MIN_N = 30           # حداقل نمونه برای گزارش درصد و میانه
MIN_CAT = 100        # آستانه دسته شغلی پرتکرار
ALPHA = 0.05         # سطح معناداری
RANDOM_STATE = 42
SALARY_UNIT = 'میلیون تومان در ماه'

DATA_DIR = Path('JobVision-data')
FIG_DIR = Path('outputs/figures')

J_MONTHS = ['فروردین', 'اردیبهشت', 'خرداد', 'تیر', 'مرداد', 'شهریور',
            'مهر', 'آبان', 'آذر', 'دی', 'بهمن', 'اسفند']
SEASONS = ['بهار', 'تابستان', 'پاییز', 'زمستان']


def jmonth(year_month):
    """برچسب ماه شمسی متناظر با ماه میلادی داده.

    داده فقط دقت ماه دارد؛ هر ماه میلادی به ماه شمسی که بیشتر روزهایش در آن است
    نسبت داده می‌شود (مثلاً 2022-04 → فروردین ۱۴۰۱).
    """
    year, month = (int(part) for part in str(year_month).split('-'))
    jm = (month + 8) % 12 + 1
    jy = year - 621 if month >= 4 else year - 622
    return jy, jm, f'{J_MONTHS[jm - 1]} {fa_num(jy)}'


def jseason(year_month):
    _, jm, _ = jmonth(year_month)
    return SEASONS[(jm - 1) // 3]


# ---------------------------------------------------------------------------
# بارگذاری داده
# ---------------------------------------------------------------------------
CANDIDATES = [DATA_DIR / 'JobVision_Jobposts_Dataset.csv',
              Path('JobVision_Jobposts_Dataset.csv'),
              Path('/content/JobVision_Jobposts_Dataset.csv')]
DATA_FILE = next((path for path in CANDIDATES if path.is_file()), None)
if DATA_FILE is None:
    raise FileNotFoundError(
        'فایل JobVision_Jobposts_Dataset.csv پیدا نشد. آن را در پوشه JobVision-data/ '
        'قرار دهید یا از https://huggingface.co/datasets/JobVision/JobVision_Jobposts_Dataset دانلود کنید.')

raw = pd.read_csv(DATA_FILE, low_memory=False)
raw = raw.rename(columns={
    'RowNumber': 'ad_id',
    'Comany_CompanyOwnershipTypesFa': 'Company_OwnershipFa',
    'Company_CompanyOwnershipTypesEn': 'Company_OwnershipEn',
})
raw['release_month'] = raw['Jobpost_ActivationTime_YEAR_MONTH'].astype(str)

assert raw.ad_id.is_unique, 'شناسه آگهی یکتا نیست'
assert len(raw) == 364_838, f'تعداد ردیف غیرمنتظره: {len(raw):,}'

versions = pd.DataFrame({
    'کتابخانه': ['Python', 'pandas', 'numpy', 'matplotlib', 'seaborn', 'scipy', 'scikit-learn', 'statsmodels', 'networkx'],
    'نسخه': [f'{__import__("sys").version_info.major}.{__import__("sys").version_info.minor}.{__import__("sys").version_info.micro}',
             pd.__version__, np.__version__, matplotlib.__version__, sns.__version__,
             scipy.__version__, sklearn.__version__, statsmodels.__version__, nx.__version__],
}).set_index('کتابخانه')

show_table('3.1', 'نسخه کتابخانه‌های اجرا', versions,
           labels={'نسخه': 'نسخه'}, sample='محیط اجرا', period=False,
           index_name='کتابخانه')

load_summary = pd.DataFrame({
    'مقدار': [f'{len(raw):,}', '۴۳', f'{raw.memory_usage(deep=True).sum() / 1024**2:,.0f} مگابایت',
              f'{DATA_FILE.stat().st_size / 1024**2:,.0f} مگابایت', f'{raw.release_month.nunique():,}'],
}, index=['تعداد آگهی', 'تعداد ستون اصلی', 'حجم در حافظه', 'حجم فایل', 'تعداد ماه'])
show_table('3.2', 'خلاصه بارگذاری داده', load_summary,
           sample='داده خام', period=False, index_name='مشخصه')

show_metrics('بارگذاری', [
    ('فایل داده', str(DATA_FILE).replace('\\', '/')),
    ('بازه ماه‌ها', f'{raw.release_month.min()} تا {raw.release_month.max()}'),
    ('ستون شناسه', 'ad_id (یکتا)'),
])
```

# ۴. شناخت داده و کیفیت

## ۴.۲. گمشدگی: کجا و چقدر؟

**جدول ۴.۳ | تعداد و درصد مقادیر گمشده در همه ستون‌ها**

</div>

```python
def column_group(name):
    if name.startswith('Company_'):
        return 'کارفرما'
    if name in {'ad_id', 'release_month', 'Jobpost_ActivationTime_YEAR_MONTH'}:
        return 'شناسه و زمان'
    if name in {'Jobpost_MinSalary', 'Jobpost_MaxSalary', 'Jobpost_SalaryCanBeShown'}:
        return 'حقوق'
    if name in {'Jobpost_SoftwareSkills', 'Jobpost_LanguageSkills', 'Jobpost_AcademicFields'}:
        return 'مهارت و تحصیلات'
    if name in {'Jobpost_BenefitFa', 'Jobpost_BenefitEn'}:
        return 'مزایا'
    return 'مشخصات آگهی'


cols_for_missing = [c for c in raw.columns if c != 'release_month']
missing = pd.DataFrame({
    'گروه': [column_group(c) for c in cols_for_missing],
    'تعداد گمشده': raw[cols_for_missing].isna().sum().values,
}, index=cols_for_missing)
missing['درصد'] = missing['تعداد گمشده'] / len(raw) * 100
missing = missing.sort_values('درصد', ascending=False)
show_table('4.3', 'مقادیر گمشده در همه ستون‌ها', missing,
           formats={'تعداد گمشده': 'int', 'درصد': 'pct'},
           sample=f'{len(raw):,} آگهی', period=False, index_name='ستون')

missing_rows = raw.isna().sum(axis=1)
row_bins = pd.cut(missing_rows, bins=[-1, 0, 2, 5, 9, 43],
                  labels=['بدون گمشدگی', '۱ تا ۲', '۳ تا ۵', '۶ تا ۹', '۱۰ و بیشتر'])
row_missing = row_bins.value_counts().reindex(['بدون گمشدگی', '۱ تا ۲', '۳ تا ۵', '۶ تا ۹', '۱۰ و بیشتر'])
row_missing = row_missing.rename_axis('تعداد خانه گمشده در ردیف').to_frame('تعداد آگهی')
row_missing['درصد'] = row_missing['تعداد آگهی'] / len(raw) * 100
show_table('4.4', 'توزیع گمشدگی در سطح ردیف', row_missing,
           formats={'تعداد آگهی': 'int', 'درصد': 'pct'}, sample=f'{len(raw):,} آگهی', period=False,
           index_name='تعداد خانه گمشده')

fig, ax = plt.subplots(figsize=(10.5, 10.5))
order = missing.sort_values('درصد')
colors = {'حقوق': GOLD, 'مهارت و تحصیلات': BLUE, 'کارفرما': PURPLE,
          'مزایا': GREEN, 'مشخصات آگهی': TEAL, 'شناسه و زمان': GRAY}
ax.barh(range(len(order)), order['درصد'], color=[colors[g] for g in order['گروه']], height=0.75)
ax.set_yticks(range(len(order)))
ax.set_yticklabels([fa(c) for c in order.index], fontsize=7.5)
ax.set_xlabel(fa('درصد مقادیر گمشده'))
ax.set_xlim(0, 102)
for i, value in enumerate(order['درصد']):
    ax.text(value + 1, i, f'{value:.0f}٪', va='center', fontsize=7)
handles = [plt.Rectangle((0, 0), 1, 1, color=c) for c in colors.values()]
ax.legend(handles, [fa(k) for k in colors], fontsize=8, loc='lower right', title=fa('گروه ستون'))
heading(ax, '۴.۱', 'درصد گمشدگی همه ۴۳ ستون')
complete('4.1')

fig, ax = plt.subplots(figsize=(8.5, 4.4))
bins = row_missing['تعداد آگهی']
bars = ax.bar(range(len(bins)), bins.values, color=[TEAL, TEAL, BLUE, PURPLE, RED], width=0.65)
ax.set_xticks(range(len(bins)))
ax.set_xticklabels([fa(label) for label in bins.index], rotation=20, ha='right', fontsize=9)
ax.set_ylabel(fa('تعداد آگهی'))
for bar, value in zip(bars, bins.values):
    ax.text(bar.get_x() + bar.get_width() / 2, bar.get_height(), f'{value:,}', ha='center',
            va='bottom', fontsize=8.5)
heading(ax, '۴.۲', 'توزیع تعداد خانه‌های گمشده در هر آگهی', period='کل داده')
complete('4.2')
```

**تفسیر جدول ۴.۳ و نمودار ۴.۱:** گمشدگی یک الگوی ساختاری دارد، نه تصادفی. سه ستون «اختیاری» داده تقریباً همیشه خالی‌اند: پشتیبانی از معلولیت (۹۵٫۱٪)، دسته‌های شغلی فرعی (۷۷٫۷٪) و مهارت زبان (۶۴٫۵٪). در مقابل، ستون‌های اصلی آگهی — عنوان، دسته شغلی، استان، نوع همکاری، سابقه و اندازه شرکت — هیچ مقدار گمشده‌ای ندارند. این یعنی هسته داده قابل اتکاست و کمبودها به لایه‌های تکمیلی (مهارت، مزایا، سن) مربوط می‌شود.

حقوق نیز ۶۶٫۹٪ گمشده دارد، اما این گمشدگی از جنس «ثبت نشده» نیست؛ آگهی رسماً حقوق را اعلام نکرده است. تفاوت این دو در بخش ۹ بررسی می‌شود.

**تفسیر جدول ۴.۴ و نمودار ۴.۲:** بیشتر ردیف‌ها بین ۳ تا ۷ خانه خالی دارند و میانه گمشدگی هر ردیف ۵ خانه است. فقط ۲۰۹ آگهی (۰٫۱٪) کاملاً پر هستند و ۱۷٫۵٪ ردیف‌ها ۱۰ خانه یا بیشتر خالی دارند. این توزیع نشان می‌دهد حذف ردیف‌های ناقص، بخش بزرگی از داده را نابود می‌کند؛ بنابراین راهبرد درست، **حذف نکردن** و تحلیل هر شاخص روی نمونه در دسترس آن است.

## ۴.۳. تکرار، کلید و سازگاری منطقی

</div>

```python
checks = pd.DataFrame({
    'مقدار': [f'{len(raw):,}', f'{raw.ad_id.nunique():,}', f'{int(raw.duplicated().sum()):,}',
              f'{int(raw.ad_id.duplicated().sum()):,}',
              f'{int((raw.Jobpost_MinSalary > raw.Jobpost_MaxSalary).sum()):,}',
              f'{int((~raw.Jobpost_SalaryCanBeShown & raw.Jobpost_MinSalary.notna()).sum()):,}',
              f'{int((raw.Jobpost_SalaryCanBeShown & raw.Jobpost_MinSalary.isna()).sum()):,}'],
    'وضعیت': ['—', 'یکتا ✓', 'بدون تکرار ✓', 'بدون تکرار ✓', 'سازگار ✓', 'سازگار ✓', 'سازگار ✓'],
}, index=['تعداد ردیف', 'شناسه یکتا', 'ردیف کاملاً تکراری', 'شناسه تکراری',
          'حداقل حقوق بزرگ‌تر از حداکثر', 'حقوق بدون پرچم افشا', 'پرچم افشا بدون حقوق'])
show_table('4.5', 'کنترل تکرار، کلید و سازگاری حقوق', checks, sample=f'{len(raw):,} آگهی',
           period=False, index_name='بررسی')

salary_flag_crosstab = pd.crosstab(
    raw.Jobpost_SalaryCanBeShown.map({True: 'پرچم افشا = True', False: 'پرچم افشا = False'}),
    raw.Jobpost_MinSalary.notna().map({True: 'حقوق موجود', False: 'حقوق خالی'}))
show_table('4.6', 'تطابق پرچم افشا با وجود مقدار حقوق', salary_flag_crosstab,
           formats={'حقوق خالی': 'int', 'حقوق موجود': 'int'},
           sample=f'{len(raw):,} آگهی', period=False, index_name='پرچم افشا')
```

**تفسیر جدول‌های ۴.۵ و ۴.۶:** داده از نظر ساختاری پاک است: هیچ ردیف کاملاً تکراری، هیچ شناسه تکراری و هیچ بازه حقوق ناسازگاری (حداقل بزرگ‌تر از حداکثر) وجود ندارد. تطابق پرچم افشا با وجود مقدار حقوق نیز کامل است: هر جا `SalaryCanBeShown=True` است، هر دو سر بازه پر است و هر جا `False` است، هر دو خالی‌اند. بنابراین «حقوق گمشده» در این داده یک نقص ثبت نیست؛ یک **انتخاب آگاهانه کارفرما** برای عدم افشاست. همین موضوع، تحلیل بخش ۹ را ضروری می‌کند.

## ۴.۴. بازه زمانی و پوشش ماهانه

</div>

```python
monthly = raw.groupby('release_month').size().rename('تعداد آگهی').to_frame()
monthly['ماه شمسی'] = [jmonth(m)[2] for m in monthly.index]
monthly['سهم'] = monthly['تعداد آگهی'] / len(raw) * 100
monthly['وضعیت'] = np.where(monthly.index.isin([monthly.index.min(), monthly.index.max()]), 'ناقص', 'کامل')
show_table('4.7', 'پوشش ماهانه داده', monthly,
           formats={'تعداد آگهی': 'int', 'سهم': 'pct'},
           sample=f'{len(raw):,} آگهی', period=False, index_name='ماه میلادی')

fig, ax = plt.subplots(figsize=(12.5, 4.6))
x = np.arange(len(monthly))
values = monthly['تعداد آگهی'].values
bar_colors = [RED if s == 'ناقص' else TEAL for s in monthly['وضعیت']]
ax.bar(x, values, color=bar_colors, width=0.72)
ax.plot(x, values, color=DARK, linewidth=1.2, alpha=0.7)
ax.set_xticks(x)
ax.set_xticklabels([fa(m) for m in monthly.index], rotation=65, ha='right', fontsize=8)
ax.set_ylabel(fa('تعداد آگهی'))
peak = int(np.argmax(values))
low = int(np.argmin(values[1:-1])) + 1
ax.annotate(fa(f'اوج: {monthly.index[peak]} — {values[peak]:,}'),
            xy=(peak, values[peak]), xytext=(peak - 4.5, values[peak] * 0.98),
            fontsize=8.5, arrowprops=dict(arrowstyle='->', color=GOLD))
ax.annotate(fa(f'کمترین ماه کامل: {monthly.index[low]} — {values[low]:,}'),
            xy=(low, values[low]), xytext=(low + 1.2, values[low] + 6500),
            fontsize=8.5, arrowprops=dict(arrowstyle='->', color=RED))
handles = [plt.Rectangle((0, 0), 1, 1, color=TEAL), plt.Rectangle((0, 0), 1, 1, color=RED)]
ax.legend(handles, [fa('ماه کامل'), fa('ماه ناقص (ابتد/انتهای داده)')], fontsize=8)
heading(ax, '۴.۳', 'تعداد آگهی منتشرشده در هر ماه', period='کل داده')
complete('4.3')
```

```python
in_period = raw.release_month.between(PERIOD_START, PERIOD_END)
show_metrics('پوشش دوره', [
    ('کل بازه', f'{raw.release_month.min()} تا {raw.release_month.max()}'),
    ('ماه کامل', '۲۳ ماه'),
    ('آگهی داخل دوره تحلیل', f'{int(in_period.sum()):,} ({in_period.mean()*100:.1f}٪)'),
    ('آگهی خارج از دوره', f'{int((~in_period).sum()):,}'),
])
```

**تفسیر جدول ۴.۷ و نمودار ۴.۳:** داده ۲۵ ماه را پوشش می‌دهد، اما مارس ۲۰۲۲ (۹۰۱ آگهی) و مارس ۲۰۲۴ (۳٬۸۵۴ آگهی) آشکارا ناقص‌اند؛ ماه اول و آخر داده به‌صورت برشی ثبت شده‌اند. ۲۳ ماه کامل باقی می‌ماند که ۳۶۰٬۰۸۳ آگهی (۹۸٫۷٪) را در خود دارد. دوره تحلیل از فروردین ۱۴۰۱ تا بهمن ۱۴۰۲ تعریف می‌شود و دو ماه ناقص فقط در توصیف کیفیت داده می‌مانند.

اوج انتشار اکتبر ۲۰۲۳ (مهر ۱۴۰۲) با ۲۴٬۰۱۷ آگهی است و کمترین ماه کامل، مارس ۲۰۲۳ (اسفند ۱۴۰۱) با ۵٬۵۳۱ آگهی. این افت شدید درست پیش از نوروز رخ داده است و در بخش ۱۰ به‌عنوان سرنخ فصلی بررسی می‌شود. توجه کنید که این دو عدد، ماه‌های میلادی‌اند؛ نگاشت به ماه شمسی در جدول ۴.۷ آمده و در تحلیل روند اصل قرار می‌گیرد.

## ۴.۵. کاردینالیتی: کدام ستون‌ها قابل گروه‌بندی‌اند؟

</div>

```python
card_cols = ['Jobpost_RawTitle', 'Jobpost_CompanyWorkTimesDescription', 'Jobpost_SoftwareSkills',
             'Jobpost_BenefitFa', 'Jobpost_AcademicFields', 'Jobpost_LanguageSkills',
             'Jobpost_MainJobCategory', 'Jobpost_SecondaryJobCategories', 'Jobpost_IndustryFa',
             'Company_IndustryFa', 'Company_CityFa', 'Jobpost_ProvinceFa', 'Jobpost_WorkTypeFa',
             'Jobpost_PreferredGender', 'Company_SizeFa', 'Jobpost_RequiredExperienceYears']
cardinality = pd.DataFrame({
    'تعداد مقدار یکتا': [raw[c].nunique(dropna=True) for c in card_cols],
    'درصد گمشده': [raw[c].isna().mean() * 100 for c in card_cols],
}, index=card_cols)
cardinality['نوع'] = ['متن آزاد' if raw[c].nunique() > 500 else 'دسته‌ای' for c in card_cols]
show_table('4.8', 'کاردینالیتی ستون‌های کلیدی', cardinality,
           formats={'تعداد مقدار یکتا': 'int', 'درصد گمشده': 'pct'},
           sample=f'{len(raw):,} آگهی', period=False, index_name='ستون')

show_metrics('کاردینالیتی', [
    ('عنوان شغلی یکتا', f'{raw.Jobpost_RawTitle.nunique():,}'),
    ('شرح ساعات کاری یکتا', f'{raw.Jobpost_CompanyWorkTimesDescription.nunique():,}'),
    ('ترکیب مهارت نرم‌افزاری یکتا', f'{raw.Jobpost_SoftwareSkills.nunique(dropna=True):,}'),
    ('دسته شغلی اصلی / صنعت', f'{raw.Jobpost_MainJobCategory.nunique()} / {raw.Jobpost_IndustryFa.nunique()}'),
])
```

**تفسیر جدول ۴.۸:** دو جهان متفاوت در این داده کنار هم است. ستون‌های مدیریتی (دسته شغلی، صنعت، استان، نوع همکاری، اندازه شرکت) کاردینالیتی پایین و کنترل‌شده دارند و برای گروه‌بندی آماده‌اند. در مقابل، متن‌های آزاد بسیار پرتنوع‌اند: ۶۸٬۷۱۱ عنوان شغلی، ۶۸٬۲۴۷ شرح ساعات کاری و ۴۹٬۱۱۵ ترکیب مهارت نرم‌افزاری یکتا. تقریباً هر آگهی شرح ساعات کاری خودش را دارد؛ پس استخراج ویژگی از این متن، یک گام اجباری است، نه اختیاری.

۳٬۵۶۱ ترکیب صنعتی متفاوت برای شرکت‌ها نیز نشان می‌دهد صنعت شرکت چندمقداری است و باید به‌صورت فهرست پردازش شود، نه یک برچسب واحد.

# ۵. پاک‌سازی و ساخت لایه تحلیل

از این بخش به بعد، همه تحلیل‌ها روی یک جدول واحد در سطح آگهی به نام `ads` انجام می‌شود. این جدول سه ویژگی دارد: فقط ماه‌های کامل را در بر می‌گیرد، متن‌های کلیدی نرمال شده‌اند و ستون‌های پایه حقوق و زمان در آن ساخته شده است. ساخت یک‌باره این لایه، از محاسبات تکراری در بخش‌های بعد جلوگیری می‌کند.

## ۵.۱. تعیین دوره تحلیل و ستون‌های زمانی

</div>

```python
ads = raw.loc[raw.release_month.between(PERIOD_START, PERIOD_END)].copy()

months = sorted(ads.release_month.unique())
month_index = {month: position for position, month in enumerate(months)}
ads['month_index'] = ads.release_month.map(month_index)
jalali = [jmonth(month) for month in ads.release_month]
ads['jyear'] = [item[0] for item in jalali]
ads['jmonth_num'] = [item[1] for item in jalali]
ads['jmonth_label'] = [item[2] for item in jalali]
ads['jseason'] = [jseason(month) for month in ads.release_month]
ads['jseason_num'] = (ads.jmonth_num - 1) // 3
ads['year_label'] = ads.jyear.astype(str).map(lambda value: nf(int(value)))

period_table = pd.DataFrame({
    'مقدار': [f'{len(raw):,}', f'{len(raw) - len(ads):,}', f'{len(ads):,}',
              f'{ads.release_month.min()} تا {ads.release_month.max()}',
              f'{nf(ads.jyear.min())} تا {nf(ads.jyear.max())}', f'{ads.month_index.nunique()}', '۲۳ ماه'],
}, index=['کل ردیف‌های خام', 'حذف‌شده (ماه ناقص)', 'ردیف‌های دوره تحلیل', 'بازه میلادی',
          'بازه شمسی', 'تعداد ماه میلادی', 'تعداد ماه کامل'])
show_table('5.1', 'اثر تعیین دوره تحلیل', period_table, sample=f'{len(ads):,} آگهی', period=False,
           index_name='مشخصه')

show_metrics('دوره تحلیل', [
    ('ردیف‌های واردشده', f'{len(ads):,}'),
    ('حذف‌شده', f'{len(raw) - len(ads):,} ({((len(raw) - len(ads)) / len(raw) * 100):.1f}٪)'),
    ('ماه‌های کامل', f'{ads.release_month.nunique()}'),
    ('سهم دوره از داده', f'{len(ads) / len(raw) * 100:.1f}٪'),
])
```

## ۵.۲. نرمال‌سازی متن‌های کلیدی

</div>

```python
TEXT_COLUMNS = ['Jobpost_RawTitle', 'Jobpost_MainJobCategory', 'Jobpost_SecondaryJobCategories',
                'Jobpost_IndustryFa', 'Jobpost_ProvinceFa', 'Jobpost_PreferredGender',
                'Jobpost_WorkTypeFa', 'Jobpost_CompanyWorkTimesDescription', 'Jobpost_BenefitFa',
                'Company_SizeFa', 'Company_AgeFromEstablishmentYear', 'Company_OwnershipFa',
                'Company_ActivityTypeFa', 'Company_IndustryFa', 'Company_ProvinceFa', 'Company_CityFa',
                'Jobpost_AcademicFields', 'Jobpost_SoftwareSkills', 'Jobpost_LanguageSkills']


def clean_text(value):
    """یکسان‌سازی نویسه‌های عربی، حذف نیم‌فاصله پیوسته و فشرده‌سازی فاصله‌ها."""
    if not isinstance(value, str):
        return value
    text = (value.replace('\u200c', ' ').replace('\u200f', '').replace('\u200e', '')
                 .replace('ي', 'ی').replace('ك', 'ک').replace('ۀ', 'ه').replace('\xa0', ' '))
    text = re.sub(r'\s+', ' ', text)
    return text.strip()


before_after = []
for column in TEXT_COLUMNS:
    before = ads[column].fillna('').astype(str)
    ads[column] = ads[column].map(clean_text)
    after = ads[column].fillna('').astype(str)
    changed = int((before != after).sum())
    if changed:
        before_after.append({'ستون': column, 'ردیف تغییر‌یافته': changed,
                             'سهم': changed / len(ads) * 100})
text_cleanup = pd.DataFrame(before_after).set_index('ستون').sort_values('ردیف تغییر‌یافته', ascending=False)
show_table('5.2', 'اثر نرمال‌سازی متن', text_cleanup,
           formats={'ردیف تغییر‌یافته': 'int', 'سهم': 'pct'},
           sample=f'{len(ads):,} آگهی', index_name='ستون')

show_metrics('نرمال‌سازی', [
    ('ستون پاک‌سازی‌شده', f'{len(TEXT_COLUMNS)}'),
    ('ستونی که تغییر کرد', f'{len(text_cleanup)}'),
    ('بیشترین تغییر', f'{text_cleanup.index[0]} — {text_cleanup["ردیف تغییر‌یافته"].iloc[0]:,} ردیف'),
])
```

## ۵.۳. ساخت ستون‌های پایه حقوق و بررسی سلامت مقادیر

</div>

```python
ads['salary_disclosed'] = ads.Jobpost_SalaryCanBeShown.astype(bool)
ads['salary_min'] = ads.Jobpost_MinSalary.astype(float)
ads['salary_max'] = ads.Jobpost_MaxSalary.astype(float)
ads['salary_mid'] = (ads.salary_min + ads.salary_max) / 2
ads['salary_span'] = ads.salary_max - ads.salary_min
ads['salary_span_pct'] = ads.salary_span / ads.salary_min * 100
ads['log_salary'] = np.log(ads.salary_mid)

paid = ads.loc[ads.salary_disclosed].copy()
salary_profile = paid[['salary_min', 'salary_max', 'salary_mid', 'salary_span', 'salary_span_pct']].describe().T
salary_profile['skew'] = paid[['salary_min', 'salary_max', 'salary_mid', 'salary_span', 'salary_span_pct']].skew()
salary_profile = salary_profile.rename(columns={
    'count': 'تعداد', 'mean': 'میانگین', 'std': 'انحراف معیار', 'min': 'کمینه',
    '25%': 'چارک اول', '50%': 'میانه', '75%': 'چارک سوم', 'max': 'بیشینه', 'skew': 'چولگی'})
salary_profile.index = ['حداقل بازه', 'حداکثر بازه', 'میانه بازه', 'پهنای بازه', 'پهنای نسبی ٪']
salary_profile = salary_profile[['تعداد', 'میانگین', 'کمینه', 'چارک اول', 'میانه',
                                 'چارک سوم', 'بیشینه', 'چولگی']]
show_table('5.3', 'خلاصه آماری متغیرهای حقوق', salary_profile,
           formats={'تعداد': 'int', 'میانگین': 'money2', 'کمینه': 'money2',
                    'چارک اول': 'money2', 'میانه': 'money2', 'چارک سوم': 'money2', 'بیشینه': 'money2',
                    'چولگی': 'score'},
           sample=f'{len(paid):,} آگهی با حقوق اعلام‌شده',
           note='واحد حقوق: میلیون تومان در ماه؛ چولگی پهنای نسبی بر حسب درصد است.')

show_metrics('حقوق اعلام‌شده', [
    ('آگهی با حقوق', f'{len(paid):,} ({len(paid) / len(ads) * 100:.1f}٪)'),
    ('میانه میانه بازه', f'{paid.salary_mid.median():,.1f} {SALARY_UNIT}'),
    ('میانگین میانه بازه', f'{paid.salary_mid.mean():,.1f} {SALARY_UNIT}'),
    ('چارک اول / سوم', f'{paid.salary_mid.quantile(.25):,.1f} / {paid.salary_mid.quantile(.75):,.1f}'),
    ('چولگی میانه بازه', f'{paid.salary_mid.skew():.2f}'),
])
```

## ۵.۴. یکسانی محل آگهی و محل شرکت

</div>

```python
ads['same_province'] = ads.Jobpost_ProvinceFa.eq(ads.Company_ProvinceFa)
location_check = pd.DataFrame({
    'تعداد': [int(ads.same_province.sum()), int((~ads.same_province).sum())],
    'سهم': [ads.same_province.mean() * 100, (1 - ads.same_province.mean()) * 100],
}, index=['استان آگهی = استان شرکت', 'استان آگهی ≠ استان شرکت'])
show_table('5.4', 'یکسانی استان آگهی و استان شرکت', location_check,
           formats={'تعداد': 'int', 'سهم': 'pct'}, sample=f'{len(ads):,} آگهی',
           index_name='وضعیت')
```

## ۵.۵. لاگ تصمیم‌های پاک‌سازی

</div>

```python
decisions = pd.DataFrame([
    ('حذف دو ماه ناقص', 'نگه‌داشتن ۲۰۲۲-۰۴ تا ۲۰۲۴-۰۲', f'{len(raw) - len(ads):,} ردیف حذف شد'),
    ('حذف نکردن ردیف‌های دارای گمشدگی', 'گمشدگی حقوق و مهارت به‌عنوان یافته تحلیل می‌شود',
     f'{int(ads.isna().any(axis=1).sum()):,} ردیف دارای گمشدگی نگه داشته شد'),
    ('نرمال‌سازی نویسه‌های عربی', 'ي→ی، ك→ک، نیم‌فاصله→فاصله',
     f'{len(text_cleanup)} ستون تغییر کرد'),
    ('یکسان‌سازی عنوان غلط ستون', 'Comany_CompanyOwnershipTypesFa → Company_OwnershipFa', 'نام ستون'),
    ('تبدیل پرچم افشا به بولین', 'SalaryCanBeShown → salary_disclosed', f'{len(ads):,} ردیف'),
    ('نگه‌داشتن پرت‌های حقوق', 'وینزور فقط در مدل‌سازی، نه در توصیف', f'{len(paid):,} ردیف با حقوق'),
    ('واحد حقوق', 'میلیون تومان در ماه', 'بدون تبدیل مقیاس'),
    ('استان شرکت برای گمشده‌ها', 'شهر شرکت گمشده → «نامشخص» در بخش ۶', '179 ردیف'),
], columns=['تصمیم', 'قاعده', 'اثر'])
show_table('5.5', 'لاگ تصمیم‌های پاک‌سازی و اثر آن‌ها', decisions.set_index('تصمیم'),
           sample=f'{len(ads):,} آگهی', period=False, index_name='تصمیم')

show_metrics('لایه تحلیل', [
    ('ابعاد جدول ads', f'{len(ads):,} × {ads.shape[1]}'),
    ('ستون‌های ساخته‌شده در بخش ۵', '۱۰'),
    ('نگه‌داشت ردیف', f'{len(ads) / len(raw) * 100:.1f}٪ از داده خام'),
])
```

# ۶. مهندسی ویژگی

ستون‌های پنهان داده را بیرون می‌کشیم: ساعت کاری از متن آزاد، مهارت از JSON، مقطع تحصیلی از فهرست رشته‌ها، مزایا از فهرست کاما‌جداشده و ویژگی‌های کارفرما از مقادیر متنی. خروجی این بخش، جدول `ads` را به یک جدول تحلیل کامل تبدیل می‌کند.

## ۶.۱. ویژگی‌های حقوق

</div>

```python
salary_bins = [0, 8, 12, 20, 40, np.inf]
salary_labels = ['زیر ۸', '۸ تا ۱۲', '۱۲ تا ۲۰', '۲۰ تا ۴۰', 'بالای ۴۰']
ads['salary_band'] = pd.cut(ads.salary_mid, bins=salary_bins, labels=salary_labels)
ads['salary_range_kind'] = np.select(
    [ads.salary_span_pct.lt(15), ads.salary_span_pct.le(25), ads.salary_span_pct.gt(25)],
    ['بازه تنگ (زیر ۱۵٪)', 'بازه متوسط (۱۵ تا ۲۵٪)', 'بازه گسترده (بالای ۲۵٪)'], default='نامعلوم')

band_table = (ads.loc[ads.salary_disclosed]
              .groupby('salary_band', observed=True)
              .agg(**{'تعداد آگهی': ('ad_id', 'size'), 'میانه حقوق': ('salary_mid', 'median'),
                      'میانه پهنای نسبی': ('salary_span_pct', 'median')}))
band_table['سهم'] = band_table['تعداد آگهی'] / band_table['تعداد آگهی'].sum() * 100
show_table('6.1', 'توزیع آگهی‌های دارای حقوق در بازه‌های دستمزد', band_table,
           formats={'تعداد آگهی': 'int', 'میانه حقوق': 'money', 'میانه پهنای نسبی': 'pct', 'سهم': 'pct'},
           sample=f'{int(ads.salary_disclosed.sum()):,} آگهی با حقوق', index_name='بازه حقوق')

range_table = (ads.loc[ads.salary_disclosed].salary_range_kind.value_counts()
               .rename('تعداد آگهی').to_frame())
range_table['سهم'] = range_table['تعداد آگهی'] / range_table['تعداد آگهی'].sum() * 100
show_table('6.2', 'نوع بازه اعلام‌شده در آگهی‌ها', range_table,
           formats={'تعداد آگهی': 'int', 'سهم': 'pct'},
           sample=f'{int(ads.salary_disclosed.sum()):,} آگهی با حقوق', index_name='نوع بازه')
```

**تفسیر جدول‌های ۶.۱ و ۶.۲:** بازار در سه لایه دستمزدی جمع می‌شود: ۱۶٫۸٪ آگهی‌ها زیر ۸ میلیون، ۷۱٫۹٪ بین ۸ تا ۲۰ میلیون و ۱۱٫۳٪ بالای ۲۰ میلیون تومان. لایه میانی، بدنه بازار کار اداری، فروش و خدمات است؛ لایه بالای ۲۰ میلیون که کمتر از یک‌نهم آگهی‌ها را می‌سازد، متعلق به نقش‌های تخصصی، مهندسی و مدیریتی است.

نوع بازه اعلام‌شده هم یک الگوی مدیریتی را آشکار می‌کند: هیچ آگهی‌ای بازه‌ای باریک‌تر از ۱۵٪ اعلام نکرده است؛ یعنی کارفرمایان ایران تقریباً همیشه یک حاشیه مذاکره دست‌کم ۱۵ درصدی در آگهی می‌گنجانند. این خودش یک توصیه عملی است: اگر می‌خواهید آگهی حرفه‌ای‌تر به نظر برسد، بازه‌ای حدود ۲۰ تا ۲۵ درصد اعلام کنید، نه بازه‌ای باریک که تنها سقف انتظار را نشان می‌دهد و نه بازه‌ای بسیار پهن که به بی‌اعتباری آگهی می‌انجامد.

## ۶.۲. استخراج ساعت کاری از متن آزاد

متن «ساعات کاری» تقریباً برای هر آگهی یکتا است (۶۸٬۲۴۷ حالت مختلف). برای تبدیل آن به عدد، یک تحلیل‌گر قاعده‌محور می‌نویسیم که بازه ساعت، بازه روزهای هفته، ساعت روزانه و ساعت هفتگی صریح را تشخیص می‌دهد و نتیجه را با منبع استخراج برچسب می‌زند.

</div>

```python
FA_DIGITS_MAP = str.maketrans('۰۱۲۳۴۵۶۷۸۹٠١٢٣٤٥٦٧٨٩', '01234567890123456789')
DAY_FIXES = [('یک شنبه', 'یکشنبه'), ('دو شنبه', 'دوشنبه'), ('سه شنبه', 'سه‌شنبه'),
             ('چهار شنبه', 'چهارشنبه'), ('پنج شنبه', 'پنجشنبه'), ('پنج‌شنبه', 'پنجشنبه')]
WEEK_DAYS = [('یکشنبه', 1), ('دوشنبه', 2), ('سه‌شنبه', 3), ('چهارشنبه', 4),
             ('پنجشنبه', 5), ('جمعه', 6), ('شنبه', 0)]
DAY_ORDER = dict(WEEK_DAYS)
DAY_RE = '|'.join(re.escape(name) for name, _ in WEEK_DAYS)
TIME_RE = re.compile(r'(\d{1,2})\s*(?:[:.]\s*(\d{2}))?\s*(?:الی|تا|-|–|به)\s*(\d{1,2})\s*(?:[:.]\s*(\d{2}))?')
DAYRANGE_RE = re.compile(rf'({DAY_RE})\s*(?:تا|الی|-|–|لغایت)\s*({DAY_RE})')


def normalize_worktime(value):
    text = str(value).translate(FA_DIGITS_MAP).replace('\u200c', ' ').replace('\xa0', ' ')
    text = re.sub(r'\s+', ' ', text)
    for bad, good in DAY_FIXES:
        text = text.replace(bad, good)
    return text.strip()


def clock(value, minutes):
    hour = value + minutes / 60
    return hour + 12 if hour < 6 else hour


def valid_span(start, end):
    if end <= start:
        end += 12
    span = end - start
    return (start, end, span) if 4 <= span <= 14 else None


def segment_days(segment):
    """تعداد روزهای هفته از یک قطعه متن: بازه روزها، فهرست روزها یا «N روز در هفته»."""
    day_range = DAYRANGE_RE.search(segment)
    if day_range:
        start, end = DAY_ORDER[day_range.group(1)], DAY_ORDER[day_range.group(2)]
        return (end - start) % 7 + 1
    found = {DAY_ORDER[m.group(0)] for m in re.finditer(DAY_RE, segment)}
    if found:
        return len(found)
    count = re.search(r'(\d{1,2})\s*روز\s*(?:کاری|در هفته|از هفته|هفته)', segment)
    if count and 1 <= int(count.group(1)) <= 7:
        return int(count.group(1))
    return None


def parse_worktime(value, work_type):
    """تجزیه متن ساعات کاری به ساعت هفتگی، روزهای هفته و پرچم‌های شیفت."""
    text = normalize_worktime(value)
    weekly = np.nan
    source = 'نامعلوم'
    daily_default = 5.5 if work_type == 'تمام وقت' else 3.0

    matches = list(TIME_RE.finditer(text))
    pieces, ranges = [], []
    for order, match in enumerate(matches):
        start = clock(int(match.group(1)), int(match.group(2) or 0))
        end = clock(int(match.group(3)), int(match.group(4) or 0))
        cleaned = valid_span(start, end)
        if cleaned is None:
            continue
        start, end, span = cleaned
        boundary = matches[order - 1].end() if order else 0
        pieces.append((span, segment_days(text[boundary:match.start()])))
        ranges.append((start, end))

    week_match = re.search(r'(\d{1,3})\s*ساعت\s*(?:در|از)\s*هفته', text)
    month_match = re.search(r'(\d{1,3})\s*ساعت\s*(?:در|از)\s*ماه', text)
    daily_match = re.search(r'(?:روزی|هر روز|در روز)\s*(\d{1,2})(?:[.,](\d))?\s*ساعت', text)

    if pieces and any(days for _, days in pieces):
        full = [span * days for span, days in pieces if days]
        partial = [span * daily_default for span, days in pieces if not days]
        weekly = sum(full) + sum(partial)
        source = 'بازه و روز از متن' if not partial else 'بازه و روز از متن (روز پیش‌فرض برای بخشی)'
    elif week_match:
        weekly, source = float(week_match.group(1)), 'صریح هفتگی'
    elif month_match:
        weekly, source = float(month_match.group(1)) / (52 / 12), 'صریح ماهانه'
    elif daily_match:
        daily = float(daily_match.group(1)) + (float('0.' + daily_match.group(2)) if daily_match.group(2) else 0)
        weekly, source = daily * (segment_days(text) or daily_default), 'ساعت روزانه × روز هفته'
    elif pieces:
        weekly, source = sum(span for span, _ in pieces) * daily_default, 'بازه با روز پیش‌فرض'
    else:
        days = segment_days(text)
        if days:
            weekly, source = 8 * days, 'روز بدون ساعت'
        else:
            defaults = {'تمام وقت': 44.0, 'تمام وقت یا پاره وقت': 44.0, 'پاره وقت': 22.0}
            weekly, source = defaults.get(work_type, np.nan), 'پیش‌فرض نوع همکاری'

    if not np.isnan(weekly) and not 8 <= weekly <= 84:
        weekly, source = np.nan, 'نامعتبر (خارج از بازه معقول)'

    return {
        'weekly_hours': weekly,
        'days_week': segment_days(text),
        'start_hour': ranges[0][0] if ranges else np.nan,
        'end_hour': ranges[0][1] if ranges else np.nan,
        'night_shift': bool(re.search(r'شبانه|شیفت شب|شب کاری|شب‌کاری', text)),
        'flexible_hours': bool(re.search(r'منعطف|شناور', text)),
        'has_thursday': 'پنجشنبه' in text,
        'has_friday': 'جمعه' in text,
        'hours_source': source,
    }


worktime_features = pd.DataFrame(
    [parse_worktime(text, work_type) for text, work_type in
     zip(ads.Jobpost_CompanyWorkTimesDescription, ads.Jobpost_WorkTypeFa)], index=ads.index)
ads = pd.concat([ads, worktime_features], axis=1)

source_table = (ads.hours_source.value_counts().rename('تعداد آگهی').to_frame())
source_table['سهم'] = source_table['تعداد آگهی'] / len(ads) * 100
show_table('6.3', 'منبع استخراج ساعت هفتگی', source_table,
           formats={'تعداد آگهی': 'int', 'سهم': 'pct'}, sample=f'{len(ads):,} آگهی',
           index_name='منبع استخراج',
           note='ردیف‌های «پیش‌فرض نوع همکاری» و «نامعتبر» ساعت هفتگی ندارند و از تحلیل درآمد ساعتی کنار می‌روند.')

hours_by_type = (ads.groupby('Jobpost_WorkTypeFa')
                 .agg(**{'تعداد آگهی': ('ad_id', 'size'),
                         'میانه ساعات هفتگی': ('weekly_hours', 'median'),
                         'میانه روزهای هفته': ('days_week', 'median'),
                         'ساعت هفتگی معلوم': ('weekly_hours', lambda s: s.notna().mean() * 100)}))
show_table('6.4', 'ساعات هفتگی برآوردشده بر پایه نوع همکاری', hours_by_type,
           formats={'تعداد آگهی': 'int', 'میانه ساعات هفتگی': 'hours',
                    'میانه روزهای هفته': 'hours', 'ساعت هفتگی معلوم': 'pct'},
           sample=f'{len(ads):,} آگهی', index_name='نوع همکاری')

fig, axes = plt.subplots(1, 2, figsize=(13.5, 4.4), gridspec_kw={'width_ratios': [1.25, 1]})
ax = axes[0]
hours_values = ads.weekly_hours.dropna()
ax.hist(hours_values, bins=np.arange(0, 90, 2), color=TEAL, edgecolor='white')
ax.axvline(hours_values.median(), color=GOLD, linewidth=1.6,
           label=fa(f'میانه {hours_values.median():.1f} ساعت'))
ax.set_xlabel(fa('ساعت کار در هفته'))
ax.set_ylabel(fa('تعداد آگهی'))
ax.legend(fontsize=8.5)
heading(ax, '۶.۱', 'توزیع ساعات کار هفتگی برآوردشده')
ax = axes[1]
order = ['تمام وقت', 'تمام وقت یا پاره وقت', 'پاره وقت', 'قراردادی / پروژه ای']
data = [ads.loc[ads.Jobpost_WorkTypeFa.eq(k), 'weekly_hours'].dropna() for k in order]
box = ax.boxplot(data, orientation='horizontal', patch_artist=True, widths=0.6,
                 medianprops=dict(color=DARK, linewidth=1.4))
for patch, color in zip(box['boxes'], [TEAL, BLUE, GOLD, PURPLE]):
    patch.set_facecolor(color)
    patch.set_alpha(0.75)
ax.set_yticklabels([fa(k) for k in order], fontsize=8.5)
ax.set_xlabel(fa('ساعت کار در هفته'))
heading(ax, '۶.۲', 'ساعات هفتگی به تفکیک نوع همکاری')
complete('6.1-6.2')
```

**تفسیر جدول‌های ۶.۳ و ۶.۴ و نمودارهای ۶.۱ و ۶.۲:** برای ۹۷٫۶٪ آگهی‌ها ساعت هفتگی برآورد شد. منبع استخراج مهم است: ۵۹٫۲٪ از متن آزاد (بازه ساعت × روزهای هفته) به‌دست آمده که دقیق‌ترین حالت است؛ ۱۷٫۶٪ فقط روزهای هفته را داشتند و با فرض ۸ ساعت در روز برآورد شدند؛ ۱۳٫۷٪ هیچ اطلاعات ساعتی نداشتند و بر پایه نوع همکاری (۴۴ ساعت برای تمام‌وقت و ۲۲ ساعت برای پاره‌وقت) پر شدند. ۱٫۳٪ هم مقدار نامعتبر داشتند و خالی ماندند.

میانه ساعت هفتگی ۴۵ ساعت و دامنه میان‌چارکی ۴۳ تا ۴۸ ساعت است؛ کمی بالاتر از استاندارد قانونی ۴۴ ساعت، چون بخشی از آگهی‌ها شش‌روزه‌اند و بخشی چند بازه زمانی در یک روز دارند. الگوی هفته کاری هم کاملاً ایرانی است: ۴۳٫۵٪ آگهی‌ها پنجشنبه را جزو روزهای کاری ذکر کرده‌اند، در حالی که جمعه فقط در ۱٫۵٪ آگهی‌ها روز کار است.

توجه کنید که این برآورد، ماهیت «ترکیبی» دارد. در بخش‌های بعدی، هر تحلیل درآمد ساعتی روی همان زیرمجموعه‌ای انجام می‌شود که ساعت معتبر دارد و محدودیت این برآورد صریح گزارش می‌شود.

## ۶.۳. استخراج مهارت‌های نرم‌افزاری

</div>

```python
def load_json_list(value):
    if not isinstance(value, str) or not value.strip():
        return []
    try:
        data = json.loads(value)
    except (ValueError, TypeError):
        return []
    return data if isinstance(data, list) else []


SKILL_LEVELS = {'مقدماتی': 1, 'متوسط': 2, 'پیشرفته': 3}
skill_rows = []
for ad_id, value in zip(ads.ad_id, ads.Jobpost_SoftwareSkills):
    seen = set()
    for item in load_json_list(value):
        if not isinstance(item, dict):
            continue
        name_en = str(item.get('TitleEn') or '').strip()
        name_fa = str(item.get('TitleFa') or '').split(',')[0].strip()
        name = name_en if name_en else name_fa
        if not name or name in seen:
            continue
        seen.add(name)
        skill_rows.append((ad_id, name, name_en, name_fa, item.get('Level'), SKILL_LEVELS.get(item.get('Level'), np.nan)))

skills_long = pd.DataFrame(skill_rows, columns=['ad_id', 'skill', 'skill_en', 'skill_fa', 'level', 'level_rank'])
ads['skill_count'] = ads.ad_id.map(skills_long.groupby('ad_id').size()).fillna(0).astype(int)
ads['skill_level_max'] = ads.ad_id.map(skills_long.groupby('ad_id').level_rank.max())

top_skills = (skills_long.groupby('skill')
              .agg(**{'تعداد آگهی': ('ad_id', 'nunique')})
              .sort_values('تعداد آگهی', ascending=False))
top_skills['سهم آگهی‌ها'] = top_skills['تعداد آگهی'] / len(ads) * 100
show_table('6.5', 'بیست مهارت نرم‌افزاری پرتکرار', top_skills.head(20),
           formats={'تعداد آگهی': 'int', 'سهم آگهی‌ها': 'pct'},
           sample=f'{int(ads.skill_count.gt(0).sum()):,} آگهی دارای مهارت', index_name='مهارت')

skill_stats = pd.DataFrame({
    'مقدار': [f'{skills_long.skill.nunique():,}', f'{len(skills_long):,}',
              f'{ads.skill_count.mean():.2f}', f'{int(ads.skill_count.max()):,}',
              f'{ads.skill_count.gt(0).mean() * 100:.1f}٪'],
}, index=['مهارت یکتا', 'ردیف مهارت-آگهی', 'میانگین مهارت هر آگهی', 'بیشترین مهارت در یک آگهی',
          'سهم آگهی‌های دارای مهارت'])
show_table('6.6', 'خلاصه جدول مهارت‌ها', skill_stats, sample='جدول مهارت-آگهی', period=False,
           index_name='شاخص')

fig, ax = plt.subplots(figsize=(9.5, 6.6))
top20 = top_skills.head(20).iloc[::-1]
ax.barh(range(len(top20)), top20['تعداد آگهی'], color=TEAL, height=0.72)
ax.set_yticks(range(len(top20)))
ax.set_yticklabels([fa(i) for i in top20.index], fontsize=8)
ax.set_xlabel(fa('تعداد آگهی'))
for position, value in enumerate(top20['تعداد آگهی']):
    ax.text(value + 900, position, f'{value:,}', va='center', fontsize=7.5)
ax.set_xlim(0, top20['تعداد آگهی'].max() * 1.16)
heading(ax, '۶.۳', 'بیست مهارت نرم‌افزاری پرتکرار', period=PERIOD_LABEL)
complete('6.3')
```

**تفسیر جدول‌های ۶.۵ و ۶.۶ و نمودار ۶.۳:** ۱۹۳٬۷۴۲ آگهی (۵۳٫۸٪) فهرست مهارت دارند و در مجموع ۴۹۱ مهارت متفاوت در آن‌ها خواسته شده است؛ به‌طور میانگین هر آگهی ۳ مهارت و در بیشترین حالت ۶۹ مهارت.

فهرست پرتکرار، تصویری از ماهیت بازار کار ایران می‌دهد: چهار مهارت نخست همه از مجموعه آفیس مایکروسافت هستند (اکسل ۳۸٫۴٪، ورد ۳۳٫۸٪، پاورپوینت ۱۲٫۸٪ و اوتلوک ۱۰٫۹٪ کل آگهی‌ها). این یعنی «سواد اداری دیجیتال» پرتقاضاترین مهارت بازار است، نه برنامه‌نویسی. پس از آن، دسته دوم متعلق به نرم‌افزارهای حسابداری و اداری ایرانی است (سپیدار، راهکاران، همکاران سیستم، به‌همراه Access و Project) که یک خوشه بومی و کمتر دیده‌شده در تحلیل‌های جهانی است. دسته سوم ابزارهای طراحی و مهندسی (فتوشاپ، ایلاستریتور، اتوکد، سالیدورکس، کتیا) است و مهارت‌های برنامه‌نویسی و داده (SQL Server، HTML/CSS، Git، JavaScript، Python) در رده‌های بعدی قرار می‌گیرند.

نکته مدیریتی: میانه حقوق آگهی‌های دارای فهرست مهارت ۱۳٫۵ میلیون است، در برابر ۱۱ میلیون برای آگهی‌های بدون مهارت. این تفاوت نشان می‌دهد «اعلام مهارت» خود نشانه‌ای از ساخت‌یافتگی و رسمی‌بودن نقش است، نه صرفاً فهرست فنی؛ در بخش ۲۲ صرفه هر مهارت با کنترل دسته شغلی سنجیده می‌شود.

## ۶.۴. استخراج تحصیلات

</div>

```python
DEGREE_RANK = {'کاردانی': 2, 'کارشناسی': 3, 'کارشناسی ارشد': 4, 'دکترا': 5, 'دکتری': 5}
RANK_DEGREE = {rank: degree for degree, rank in DEGREE_RANK.items() if degree != 'دکتری'}

academic_rows = []
for ad_id, value in zip(ads.ad_id, ads.Jobpost_AcademicFields):
    seen = set()
    for item in load_json_list(value):
        if not isinstance(item, dict):
            continue
        field = str(item.get('FieldType') or '').strip()
        degree = str(item.get('DegreeLevel') or '').strip()
        if (field, degree) in seen or not (field or degree):
            continue
        seen.add((field, degree))
        academic_rows.append((ad_id, field, degree, DEGREE_RANK.get(degree, np.nan)))

academic_long = pd.DataFrame(academic_rows, columns=['ad_id', 'field', 'degree', 'degree_rank'])
ads['max_degree_rank'] = ads.ad_id.map(academic_long.groupby('ad_id').degree_rank.max()).fillna(0)
ads['max_degree'] = ads.max_degree_rank.map(RANK_DEGREE).fillna('اعلام نشده')
ads['field_main'] = ads.ad_id.map(academic_long.sort_values('degree_rank', ascending=False)
                                  .drop_duplicates('ad_id').set_index('ad_id').field)
ads['has_academic'] = ads.max_degree_rank.gt(0)

degree_table = (ads.groupby('max_degree')
                .agg(**{'تعداد آگهی': ('ad_id', 'size'), 'میانه حقوق': ('salary_mid', 'median')}))
degree_table['سهم'] = degree_table['تعداد آگهی'] / len(ads) * 100
degree_table['آگهی با حقوق'] = ads.groupby('max_degree').salary_disclosed.sum()
degree_table = degree_table.reindex(['اعلام نشده', 'کاردانی', 'کارشناسی', 'کارشناسی ارشد', 'دکترا'])
show_table('6.7', 'توزیع آگهی‌ها بر پایه بالاترین مقطع درخواستی', degree_table,
           formats={'تعداد آگهی': 'int', 'سهم': 'pct', 'میانه حقوق': 'money', 'آگهی با حقوق': 'int'},
           sample=f'{len(ads):,} آگهی', index_name='مقطع درخواستی')

top_fields = (academic_long.groupby('field').agg(**{'تعداد آگهی': ('ad_id', 'nunique')})
              .sort_values('تعداد آگهی', ascending=False))
top_fields['سهم'] = top_fields['تعداد آگهی'] / len(ads) * 100
show_table('6.8', 'ده رشته تحصیلی پرتکرار', top_fields.head(10),
           formats={'تعداد آگهی': 'int', 'سهم': 'pct'},
           sample=f'{int(ads.has_academic.sum()):,} آگهی دارای رشته', index_name='رشته تحصیلی')

show_metrics('تحصیلات و مهارت', [
    ('آگهی دارای مقطع', f'{ads.has_academic.mean() * 100:.1f}٪'),
    ('دکترا درخواست‌شده', f'{int(ads.max_degree.eq("دکترا").sum()):,} آگهی'),
    ('رشته یکتا', f'{academic_long.field.nunique():,}'),
    ('مهارت یکتا', f'{skills_long.skill.nunique():,}'),
])
```

**تفسیر جدول‌های ۶.۷ و ۶.۸:** فقط ۴۷٫۱٪ آگهی‌ها مقطع تحصیلی را مشخص می‌کنند و در بیش از نیمی از آگهی‌ها این الزام اعلام نشده است. در میان آگهی‌های اعلام‌کننده، کارشناسی با ۳۸٫۹٪ از کل آگهی‌ها مسلط است؛ کاردانی ۴٫۳٪، کارشناسی ارشد ۳٫۵٪ و دکترا ۰٫۴٪. یعنی بازار کار ایران عملاً «بازار کارشناسی» است و تحصیلات تکمیلی شرط اقلیت کوچکی از آگهی‌هاست.

الگوی دستمزدی روشن است: میانه حقوق آگهی‌های کارشناسی ۱۳٫۵ میلیون و آگهی‌های ارشد و بالاتر ۱۷٫۵ میلیون تومان است؛ نزدیک به ۳۰٪ اختلاف. اما هم‌زمان نرخ افشای حقوق با بالا رفتن مقطع کاهش می‌یابد (۳۸٫۵٪ در آگهی‌های بدون اعلام مقطع در برابر ۲۴٫۵٪ در آگهی‌های دکترا)؛ یعنی هرچه رده تخصصی‌تر می‌شود، افشای عمومی حقوق کمتر می‌شود. این نکته در بخش ۱۲ به‌طور دقیق‌تر با کنترل سابقه و دسته شغلی بررسی می‌شود.

در مورد رشته‌ها، چهار رشته نخست — مدیریت و بازرگانی، مالی و حسابداری، کامپیوتر و فناوری اطلاعات، مهندسی صنایع — به‌تنهایی بخش بزرگی از آگهی‌ها را پوشش می‌دهند؛ ترکیبی که با ساختار بخش‌های شغلی داده هم‌خوان است.

## ۶.۵. استخراج مزایا

</div>

```python
benefit_rows = []
for ad_id, value in zip(ads.ad_id, ads.Jobpost_BenefitFa):
    if not isinstance(value, str) or not value.strip():
        continue
    for token in sorted({token.strip() for token in value.split(',')}):
        if token:
            benefit_rows.append((ad_id, token))
benefits_long = pd.DataFrame(benefit_rows, columns=['ad_id', 'benefit'])
ads['benefit_count'] = ads.ad_id.map(benefits_long.groupby('ad_id').size()).fillna(0).astype(int)
ads['has_benefit_list'] = ads.Jobpost_BenefitFa.notna() & ads.Jobpost_BenefitFa.ne('')

top_benefits = (benefits_long.groupby('benefit')
                .agg(**{'تعداد آگهی': ('ad_id', 'nunique')})
                .sort_values('تعداد آگهی', ascending=False))
top_benefits['سهم آگهی‌های دارای مزایا'] = top_benefits['تعداد آگهی'] / int(ads.has_benefit_list.sum()) * 100
top_benefits['سهم کل آگهی‌ها'] = top_benefits['تعداد آگهی'] / len(ads) * 100
show_table('6.9', 'پانزده مزیت پرتکرار', top_benefits.head(15),
           formats={'تعداد آگهی': 'int', 'سهم آگهی‌های دارای مزایا': 'pct', 'سهم کل آگهی‌ها': 'pct'},
           sample=f'{int(ads.has_benefit_list.sum()):,} آگهی دارای فهرست مزایا', index_name='مزیت')

fig, ax = plt.subplots(figsize=(9.5, 5.4))
top15 = top_benefits.head(15).iloc[::-1]
ax.barh(range(len(top15)), top15['سهم کل آگهی‌ها'], color=GREEN, height=0.7)
ax.set_yticks(range(len(top15)))
ax.set_yticklabels([fa(name) for name in top15.index], fontsize=8.5)
ax.set_xlabel(fa('درصد از کل آگهی‌های دوره'))
for position, value in enumerate(top15['سهم کل آگهی‌ها']):
    ax.text(value + 0.4, position, f'{value:.0f}٪', va='center', fontsize=7.5)
ax.set_xlim(0, top15['سهم کل آگهی‌ها'].max() * 1.15)
heading(ax, '۶.۴', 'سهم پانزده مزیت پرتکرار از کل آگهی‌ها')
complete('6.4')
```

**تفسیر جدول ۶.۹ و نمودار ۶.۴:** سه‌چهارم آگهی‌ها (۲۷۰٬۱۰۵ آگهی) فهرست مزایا دارند و میانگین تعداد مزایا در هر آگهی حدود ۴ مورد است. رتبه‌بندی مزایا، «زبان رقابت» کارفرمایان ایرانی است: بیمه درمان تکمیلی (۴۶٫۷٪ کل آگهی‌ها)، پاداش (۴۲٫۵٪)، بسته‌ها و هدایای مناسبتی (۳۶٫۸٪)، وام (۳۰٫۶٪) و ناهار (۲۳٪).

نکته کاربردی برای مدیر: چهار مزیت نخست عملاً به «استاندارد انتظار» تبدیل شده‌اند و حذف آن‌ها آگهی را از رقابت خارج می‌کند؛ اما مزیت‌هایی مثل ساعت کاری منعطف (۹٫۸٪) یا سهام تشویقی زیر ۲٪ از آگهی‌ها، هنوز کمیاب‌اند و بیشترین توان تمایز را دارند. به بیان دیگر، در فهرست مزایا باید چهار قلم استاندارد را داشت و یک یا دو قلم کم‌یاب را برای متمایز شدن افزود.

## ۶.۶. ویژگی‌های کارفرما

</div>

```python
SIZE_ORDER = {'زیر 10 نفر': 1, '11 تا 50 نفر': 2, '51 تا 200 نفر': 3, '201 تا 500 نفر': 4,
              '501 تا 1000 نفر': 5, '1001 تا 5000 نفر': 6, 'بیش از 5000 نفر': 7}
AGE_ORDER = {'0 - 4 years': 1, '5 - 9 years': 2, '10 - 14 years': 3, '15 - 19 years': 4,
             'more than 20 years': 5}

ads['company_size_rank'] = ads.Company_SizeFa.map(SIZE_ORDER)
ads['company_age_rank'] = ads.Company_AgeFromEstablishmentYear.map(AGE_ORDER)
ads['ownership'] = ads.Company_OwnershipFa.fillna('نامشخص')
ads['activity_type'] = ads.Company_ActivityTypeFa.fillna('نامشخص')
ads['company_industry_main'] = (ads.Company_IndustryFa.fillna('نامشخص')
                                .str.split(',').str[0].str.strip().replace('', 'نامشخص'))
ads['company_industry_count'] = ads.Company_IndustryFa.fillna('').map(
    lambda value: len([part for part in value.split(',') if part.strip()]))
ads['company_city'] = ads.Company_CityFa.fillna('نامشخص')

company_size_table = (ads.groupby('Company_SizeFa')
                      .agg(**{'تعداد آگهی': ('ad_id', 'size'),
                              'میانه حقوق': ('salary_mid', 'median'),
                              'نرخ افشای حقوق': ('salary_disclosed', lambda s: s.mean() * 100),
                              'نرخ دورکاری': ('Jobpost_IsRemote', lambda s: s.mean() * 100),
                              'میانه مزایا': ('benefit_count', 'median')})
                      .sort_values('تعداد آگهی', ascending=False))
company_size_table['سهم'] = company_size_table['تعداد آگهی'] / len(ads) * 100
show_table('6.10', 'آگهی‌ها و حقوق بر پایه اندازه شرکت', company_size_table,
           formats={'تعداد آگهی': 'int', 'سهم': 'pct', 'میانه حقوق': 'money', 'نرخ افشای حقوق': 'pct',
                    'نرخ دورکاری': 'pct', 'میانه مزایا': 'hours'},
           sample=f'{len(ads):,} آگهی', index_name='اندازه شرکت',
           note='میانه حقوق فقط روی آگهی‌های دارای حقوق اعلام‌شده محاسبه شده است.')
```

**تفسیر جدول ۶.۱۰:** بدنه بازار کار ایران در شرکت‌های کوچک و متوسط است: ۲۶٪ آگهی‌ها از شرکت‌های ۱۱ تا ۵۰ نفره و ۲۴٫۹٪ از شرکت‌های ۵۱ تا ۲۰۰ نفره می‌آید؛ شرکت‌های بالای هزار نفر فقط ۱۳٫۳٪ آگهی‌ها را منتشر می‌کنند.

سه الگوی مدیریتی در این جدول دیده می‌شود. اول، **شفافیت حقوق با اندازه شرکت نسبت معکوس دارد**: شرکت‌های کوچک و متوسط نرخ افشای ۴۱ تا ۴۳ درصد دارند، در حالی که شرکت‌های ۱۰۰۱ تا ۵۰۰۰ نفره فقط ۱۶٫۶٪ آگهی‌هایشان حقوق را اعلام می‌کنند. دوم، **حقوق با اندازه شرکت افزایش می‌یابد** اما نه بی‌وقفه: میانه در شرکت‌های زیر ۱۰ نفر و ۱۱ تا ۵۰ نفر ۱۱ میلیون است و در شرکت‌های بزرگ‌تر به ۱۳٫۵ میلیون می‌رسد. سوم، **دورکاری در دو سر طیف بیشتر است**: شرکت‌های زیر ۱۰ نفر (۱۱٫۶٪) و شرکت‌های بالای ۵۰۰۰ نفر (۱۱٫۳٪) بیشترین نرخ دورکاری را دارند، در حالی که شرکت‌های ۵۱ تا ۲۰۰ نفره با ۳٫۷٪ کمترین را دارند.

تفسیر کسب‌وکاری: شرکت‌های متوسط بزرگ (۵۰۱ تا ۵۰۰۰ نفر) بدترین ترکیب را دارند؛ هم کمترین شفافیت و هم کمترین انعطاف دورکاری و کمترین شمار مزایا. این دقیقاً همان جایی است که یک آگهی شفاف و منعطف بیشترین مزیت رقابتی را می‌سازد.

## ۶.۷. ساخت شاخص درآمد ساعتی

</div>

```python
ads['monthly_hours'] = ads.weekly_hours * 52 / 12
ads['hourly_income'] = ads.salary_mid * 1000 / ads.monthly_hours   # هزار تومان در ساعت

hourly_ready = ads.loc[ads.salary_disclosed & ads.hourly_income.notna()].copy()
hourly_table = (hourly_ready.groupby('Jobpost_WorkTypeFa')
                .agg(**{'تعداد آگهی': ('ad_id', 'size'),
                        'میانه حقوق ماهانه': ('salary_mid', 'median'),
                        'میانه ساعت هفتگی': ('weekly_hours', 'median'),
                        'میانه درآمد ساعتی': ('hourly_income', 'median'),
                        'چارک اول ساعتی': ('hourly_income', lambda s: s.quantile(.25)),
                        'چارک سوم ساعتی': ('hourly_income', lambda s: s.quantile(.75))})
                .sort_values('تعداد آگهی', ascending=False))
full_time_hourly = hourly_table.loc['تمام وقت', 'میانه درآمد ساعتی']
hourly_table['شاخص نسبت به تمام وقت'] = hourly_table['میانه درآمد ساعتی'] / full_time_hourly * 100
show_table('6.11', 'درآمد ساعتی بر پایه نوع همکاری', hourly_table,
           formats={'تعداد آگهی': 'int', 'میانه حقوق ماهانه': 'money', 'میانه ساعت هفتگی': 'hours',
                    'میانه درآمد ساعتی': 'hourly', 'چارک اول ساعتی': 'hourly',
                    'چارک سوم ساعتی': 'hourly', 'شاخص نسبت به تمام وقت': 'pct'},
           sample=f'{len(hourly_ready):,} آگهی دارای حقوق و ساعت',
           index_name='نوع همکاری',
           note='درآمد ساعتی به هزار تومان در ساعت؛ شاخص تمام وقت برابر ۱۰۰٪ است.')

fig, axes = plt.subplots(1, 2, figsize=(13.5, 4.6), gridspec_kw={'width_ratios': [1.1, 1]})
ax = axes[0]
plot_sample = hourly_ready.sample(min(6000, len(hourly_ready)), random_state=RANDOM_STATE)
for work_type, color in zip(['تمام وقت', 'پاره وقت', 'قراردادی / پروژه ای', 'تمام وقت یا پاره وقت'],
                            [TEAL, GOLD, PURPLE, BLUE]):
    subset = plot_sample.loc[plot_sample.Jobpost_WorkTypeFa.eq(work_type)]
    ax.scatter(subset.weekly_hours, subset.salary_mid, s=9, alpha=0.35, color=color, label=fa(work_type))
ax.set_xlabel(fa('ساعت کار در هفته'))
ax.set_ylabel(fa('میانه بازه حقوق (میلیون تومان)'))
ax.set_ylim(0, 60)
ax.legend(fontsize=8, markerscale=1.6)
heading(ax, '۶.۵', 'حقوق در برابر ساعت کار هفتگی')
ax = axes[1]
hourly_groups = [hourly_ready.loc[hourly_ready.Jobpost_WorkTypeFa.eq(k), 'hourly_income']
                 for k in ['تمام وقت', 'تمام وقت یا پاره وقت', 'پاره وقت', 'قراردادی / پروژه ای']]
box = ax.boxplot(hourly_groups, orientation='horizontal', patch_artist=True, widths=0.6,
                 showfliers=False, medianprops=dict(color=DARK, linewidth=1.4))
for patch, color in zip(box['boxes'], [TEAL, BLUE, GOLD, PURPLE]):
    patch.set_facecolor(color)
    patch.set_alpha(0.75)
ax.set_yticklabels([fa('تمام وقت'), fa('ترکیبی'), fa('پاره وقت'), fa('قراردادی/پروژه‌ای')], fontsize=8.5)
ax.set_xlabel(fa('درآمد ساعتی (هزار تومان)'))
heading(ax, '۶.۶', 'توزیع درآمد ساعتی به تفکیک نوع همکاری')
complete('6.5-6.6')

show_metrics('درآمد ساعتی', [
    ('آگهی دارای حقوق و ساعت', f'{len(hourly_ready):,}'),
    ('میانه درآمد ساعتی کل', f'{hourly_ready.hourly_income.median():,.1f} هزار تومان'),
    ('میانه درآمد ساعتی تمام وقت', f'{full_time_hourly:,.1f} هزار تومان'),
    ('میانه ساعت ماهانه', f'{hourly_ready.monthly_hours.median():,.0f} ساعت'),
])

robust_hours = hourly_ready.loc[hourly_ready.hours_source.eq('بازه و روز از متن')]
robust_table = (robust_hours.groupby('Jobpost_WorkTypeFa')
                .agg(**{'تعداد آگهی': ('ad_id', 'size'),
                        'میانه ساعت هفتگی': ('weekly_hours', 'median'),
                        'میانه درآمد ساعتی': ('hourly_income', 'median')}))
robust_table['نسبت به تمام وقت'] = (robust_table['میانه درآمد ساعتی'] /
                                    robust_table.loc['تمام وقت', 'میانه درآمد ساعتی'] * 100)
show_table('6.11.1', 'بررسی مقاومت: درآمد ساعتی فقط با ساعت استخراج‌شده از متن',
           robust_table,
           formats={'تعداد آگهی': 'int', 'میانه ساعت هفتگی': 'hours',
                    'میانه درآمد ساعتی': 'hourly', 'نسبت به تمام وقت': 'pct'},
           sample=f'{len(robust_hours):,} آگهی با ساعت صریح',
           note='اگر نتیجه با جدول ۶.۱۱ هم‌جهت باشد، برآورد پیش‌فرض ساعت‌ها نتیجه را نساخته است.')
```

**تفسیر جدول ۶.۱۱ و نمودارهای ۶.۵ و ۶.۶:** شاخص درآمد ساعتی برای ۱۱۶٬۲۴۴ آگهی (۳۲٫۳٪) قابل محاسبه است. میانه آن ۶۳٫۵ هزار تومان در ساعت با دامنه میان‌چارکی ۴۴٫۹ تا ۸۹٫۷ هزار تومان است.

مقایسه انواع همکاری، یافته‌ای مهم و کمی غیرشهودی می‌دهد: میانه درآمد ساعتی مشاغل پاره‌وقت ۱۱۵٫۴ هزار تومان است، در برابر ۶۲٫۳ هزار تومان برای تمام‌وقت؛ یعنی نزدیک ۸۵٪ بالاتر. مشاغل قراردادی و پروژه‌ای هم با ۸۹ هزار تومان بالای تمام‌وقت قرار می‌گیرند. ریشه این تفاوت را می‌توان در دو جا دید: اول، آگهی‌های پاره‌وقت و پروژه‌ای بیشتر در نقش‌های تخصصی و مشاوره‌ای منتشر می‌شوند که نرخ ساعتی بالاتری دارند؛ دوم، بخشی از کارفرمایان برای همان سطح حقوق ماهانه، ساعت کمتری درخواست می‌کنند (میانه حقوق ماهانه پاره‌وقت و تمام‌وقت هر دو ۱۱ میلیون است) و نتیجه، نرخ ساعتی بالاتر می‌شود.

**محدودیت:** بخشی از ساعت‌های پاره‌وقت با فرض پیش‌فرض (۲۲ ساعت) برآورد شده است؛ بنابراین رقم ۱۱۵٫۴ باید به‌عنوان جهت‌نما خوانده شود، نه اندازه دقیق. برای همین، همین مقایسه یک بار دیگر فقط روی آگهی‌هایی که ساعتشان مستقیماً از متن استخراج شده تکرار شده است (جدول ۶٫۱۱٫۱). اگر ترتیب پاره‌وقت > قراردادی > تمام‌وقت در آن هم بماند، نتیجه به فرض پیش‌فرض حساس نیست.

## ۶.۸. نرمال‌سازی عنوان شغلی

</div>

```python
TITLE_NOISE = r'\b(خانم|آقا|فوری|تمام وقت|پاره وقت|نیمه وقت|دورکاری|حضوری|آنلاین)\b'


def base_title(title):
    """عنوان شغلی پایه: حذف توضیحات پرانتزی، پسوندهای جنسیت و بخش‌های فرعی."""
    text = clean_text(title)
    text = re.sub(r'\(.*?\)', ' ', text)
    text = re.split(r'[-–—|/,،:؛]', text)[0]
    text = re.sub(TITLE_NOISE, ' ', text, flags=re.IGNORECASE)
    text = re.sub(r'\s+', ' ', text).strip(' .')
    return text or clean_text(title)


def seniority_level(title):
    text = title.lower()
    if re.search(r'کارآموز|intern|trainee', text):
        return 'کارآموز'
    if re.search(r'مدیر|manager|head of|ریاست', text):
        return 'مدیریتی'
    if re.search(r'سرپرست|سوپروایزر|supervisor|مسئول', text):
        return 'سرپرستی'
    if re.search(r'ارشد|senior|expert senior', text):
        return 'کارشناسی ارشد'
    if re.search(r'کارشناس|expert|specialist', text):
        return 'کارشناسی'
    return 'سایر'


ads['base_title'] = ads.Jobpost_RawTitle.map(base_title)
ads['seniority'] = ads.base_title.map(seniority_level)
SENIORITY_ORDER = ['کارآموز', 'کارشناسی', 'کارشناسی ارشد', 'سرپرستی', 'مدیریتی', 'سایر']

seniority_table = (ads.groupby('seniority')
                   .agg(**{'تعداد آگهی': ('ad_id', 'size'),
                           'میانه حقوق': ('salary_mid', 'median'),
                           'میانه سابقه': ('Jobpost_RequiredExperienceYears', 'median'),
                           'نرخ افشای حقوق': ('salary_disclosed', lambda s: s.mean() * 100)})
                   .reindex(SENIORITY_ORDER))
seniority_table['سهم'] = seniority_table['تعداد آگهی'] / len(ads) * 100
show_table('6.12', 'رده سازمانی استخراج‌شده از عنوان شغلی', seniority_table,
           formats={'تعداد آگهی': 'int', 'سهم': 'pct', 'میانه حقوق': 'money', 'میانه سابقه': 'hours',
                    'نرخ افشای حقوق': 'pct'},
           sample=f'{len(ads):,} آگهی', index_name='رده سازمانی')

top_titles = (ads.groupby('base_title')
              .agg(**{'تعداد آگهی': ('ad_id', 'size'), 'میانه حقوق': ('salary_mid', 'median')})
              .sort_values('تعداد آگهی', ascending=False))
top_titles['سهم'] = top_titles['تعداد آگهی'] / len(ads) * 100
show_table('6.13', 'پانزده عنوان شغلی پایه پرتکرار', top_titles.head(15),
           formats={'تعداد آگهی': 'int', 'سهم': 'pct', 'میانه حقوق': 'money'},
           sample=f'{ads.base_title.nunique():,} عنوان پایه', index_name='عنوان پایه')

show_metrics('عنوان شغلی', [
    ('عنوان خام یکتا', f'{ads.Jobpost_RawTitle.nunique():,}'),
    ('عنوان پایه یکتا', f'{ads.base_title.nunique():,}'),
    ('کاهش کاردینالیتی', f'{(1 - ads.base_title.nunique() / ads.Jobpost_RawTitle.nunique()) * 100:.1f}٪'),
    ('پرتکرارترین عنوان', f'{top_titles.index[0]} — {int(top_titles["تعداد آگهی"].iloc[0]):,} آگهی'),
])
```

**تفسیر جدول‌های ۶.۱۲ و ۶.۱۳:** استخراج رده سازمانی از عنوان، تصویری سه‌لایه می‌دهد. نزدیک ۴۴٪ آگهی‌ها هیچ نشانه رده‌ای در عنوان ندارند («سایر») و میانه حقوق ۱۱ میلیون دارند؛ ۳۶٫۸٪ با عنوان «کارشناس» شناسایی می‌شوند (میانه ۱۳٫۵ میلیون و سابقه ۲ سال)؛ و رده‌های ارشد و مدیریتی روی‌هم ۱۰٫۷٪ آگهی‌ها را می‌سازند اما میانه ۱۷٫۵ میلیون دارند، یعنی حدود ۶۰٪ بالاتر از بدنه بازار.

نکته قابل توجه، رده «کارآموز» است: با وجود سابقه صفر، میانه حقوق این آگهی‌ها ۱۳٫۵ میلیون است، هم‌سطح کارشناسی. بخشی از این عنوان‌ها احتمالاً کارآموزی‌های تخصصی با حقوق بالاست و بخشی عنوان‌گذاری اشتباه کارفرمایان؛ به همین دلیل در بخش ۱۶ به‌عنوان یک متغیر کنترل وارد می‌شود، نه یک یافته قطعی.

نرمال‌سازی عنوان، کاردینالیتی را از ۶۷٬۹۵۰ به ۴۰٬۹۲۸ کاهش می‌دهد (۳۹٫۸٪). پرتکرارترین عنوان‌ها «کارشناس فروش» (۱۳٬۳۶۰ آگهی)، «حسابدار» (۹٬۲۱۲) و «کارشناس حسابداری» (۷٬۴۳۷) هستند؛ یعنی فروش و مالی دو ستون فقرات بازار کار ایران‌اند. همین نرمال‌سازی، پایه خوشه‌بندی بخش ۱۷ است.

## ۶.۹. جدول نهایی تحلیل

</div>

```python
paid = ads.loc[ads.salary_disclosed].copy()
hourly_ready = ads.loc[ads.salary_disclosed & ads.hourly_income.notna()].copy()
remote_share = ads.Jobpost_IsRemote.mean() * 100

engineered = pd.DataFrame({
    'ستون': ['release_month, jyear, jmonth_label, jseason', 'salary_min, salary_max, salary_mid',
             'salary_span, salary_span_pct, log_salary', 'salary_band, salary_range_kind',
             'weekly_hours, days_week, start_hour, end_hour', 'night_shift, flexible_hours, has_thursday, has_friday',
             'hours_source', 'skill_count, skill_level_max', 'max_degree_rank, max_degree, field_main, has_academic',
             'benefit_count, has_benefit_list', 'company_size_rank, company_age_rank', 'ownership, activity_type',
             'company_industry_main, company_industry_count, company_city', 'monthly_hours, hourly_income',
             'base_title, seniority', 'same_province, salary_disclosed'],
    'نوع': ['زمان', 'حقوق', 'حقوق', 'حقوق', 'ساعت کاری', 'ساعت کاری', 'کیفیت ساعت', 'مهارت', 'تحصیلات',
            'مزایا', 'کارفرما', 'کارفرما', 'کارفرما', 'درآمد ساعتی', 'عنوان', 'کیفیت'],
    'نقش در بخش‌های بعد': ['۱۰، ۱۶', '۷، ۹، ۱۶', '۷، ۱۶', '۷، ۱۶', '۶، ۷، ۱۲', '۶، ۲۲', '۶، ۷',
                           '۱۶، ۱۷، ۱۹، ۲۱', '۱۲، ۱۶، ۱۷', '۲۰، ۲۲', '۱۶، ۲۰', '۱۶، ۲۰', '۱۶، ۱۸، ۲۰',
                           '۷، ۱۰، ۱۴', '۶، ۱۶، ۱۹', '۵، ۹، ۱۴'],
}).set_index('ستون')
show_table('6.14', 'ستون‌های ساخته‌شده در مهندسی ویژگی', engineered,
           sample=f'جدول ads: {len(ads):,} × {ads.shape[1]}', period=False, index_name='ستون')

show_metrics('لایه تحلیل نهایی', [
    ('ابعاد ads', f'{len(ads):,} × {ads.shape[1]}'),
    ('آگهی با حقوق', f'{len(paid):,} ({len(paid) / len(ads) * 100:.1f}٪)'),
    ('آگهی با ساعت معتبر', f'{int(ads.weekly_hours.notna().sum()):,}'),
    ('حقوق و ساعت هم‌زمان', f'{int((ads.salary_disclosed & ads.weekly_hours.notna()).sum()):,}'),
    ('نرخ دورکاری', f'{remote_share:.1f}٪'),
])
```

این ساختار، قواعد «یک‌بار محاسبه، چندبار استفاده» را رعایت می‌کند: بخش‌های ۱۰ تا ۲۳ هیچ‌کدام دوباره ساعت، مهارت یا مقطع را استخراج نمی‌کنند و فقط روی همین لایه تحلیل می‌سازند. اگر فردا تعریف درآمد ساعتی تغییر کند، فقط بخش ۶ اصلاح می‌شود و همه نتایج بعدی به‌طور خودکار به‌روز می‌شوند.

پرسش این بخش ساده است: متغیرهای کلیدی چه شکلی دارند؟ پاسخ این پرسش تعیین می‌کند که از میانه استفاده کنیم یا میانگین، و کدام مقادیر پرت‌اند.

## ۷.۱. خلاصه آماری متغیرهای عددی

</div>

```python
numeric_cols = {
    'salary_mid': 'میانه بازه حقوق', 'hourly_income': 'درآمد ساعتی',
    'Jobpost_RequiredExperienceYears': 'سابقه لازم', 'age_mid': 'میانه بازه سنی درخواستی',
    'weekly_hours': 'ساعت هفتگی', 'benefit_count': 'تعداد مزایا', 'skill_count': 'تعداد مهارت',
}
ads['age_mid'] = ads[['Jobpost_RequiredMinAge', 'Jobpost_RequiredMaxAge']].mean(axis=1)
sync_snapshots()

numeric_profile = ads[list(numeric_cols)].describe(
    percentiles=[.05, .25, .5, .75, .95]).T
numeric_profile['چولگی'] = ads[list(numeric_cols)].skew()
numeric_profile['کشیدگی'] = ads[list(numeric_cols)].kurtosis()
numeric_profile = numeric_profile.rename(columns={
    'count': 'تعداد معتبر', 'mean': 'میانگین', 'std': 'انحراف معیار', 'min': 'کمینه', 'max': 'بیشینه',
    '5%': 'صدک ۵', '25%': 'چارک اول', '50%': 'میانه', '75%': 'چارک سوم', '95%': 'صدک ۹۵'})
numeric_profile.index = [numeric_cols[name] for name in numeric_profile.index]
numeric_profile = numeric_profile[['تعداد معتبر', 'میانگین', 'کمینه', 'چارک اول',
                                   'میانه', 'چارک سوم', 'بیشینه', 'چولگی']]
show_table('7.1', 'خلاصه آماری متغیرهای عددی کلیدی', numeric_profile,
           formats={'تعداد معتبر': 'int', 'میانگین': 'money2', 'کمینه': 'money2',
                    'چارک اول': 'money2', 'میانه': 'money2', 'چارک سوم': 'money2',
                    'بیشینه': 'money2', 'چولگی': 'score'},
           sample=f'{len(ads):,} آگهی (هر ستون روی داده در دسترس خود)',
           note='واحد ستون‌های حقوقی میلیون تومان، درآمد ساعتی هزار تومان، سابقه و سن سال و ساعات هفتگی ساعت است.')

fig, axes = plt.subplots(2, 2, figsize=(13.5, 8))
ax = axes[0, 0]
salary_series = paid.salary_mid
ax.hist(salary_series, bins=np.arange(0, 72, 1.5), color=TEAL, edgecolor='white')
ax.axvline(salary_series.median(), color=GOLD, linewidth=1.8, label=fa(f'میانه {salary_series.median():.1f}'))
ax.axvline(salary_series.mean(), color=RED, linewidth=1.6, linestyle='--', label=fa(f'میانگین {salary_series.mean():.1f}'))
ax.set_xlabel(fa('میانه بازه حقوق (میلیون تومان)'))
ax.set_ylabel(fa('تعداد آگهی'))
ax.legend(fontsize=8)
heading(ax, '۷.۱', 'توزیع حقوق ماهانه (خطی)')

ax = axes[0, 1]
ax.hist(np.log(salary_series), bins=45, color=BLUE, edgecolor='white')
ax.set_xlabel(fa('لگاریتم طبیعی حقوق'))
ax.set_ylabel(fa('تعداد آگهی'))
heading(ax, '۷.۲', 'توزیع لگاریتم حقوق (نزدیک به نرمال)')

ax = axes[1, 0]
hourly_series = hourly_ready.hourly_income
ax.hist(hourly_series.clip(upper=hourly_series.quantile(.99)), bins=60, color=GOLD, edgecolor='white')
ax.axvline(hourly_series.median(), color=DARK, linewidth=1.8, label=fa(f'میانه {hourly_series.median():.0f}'))
ax.set_xlabel(fa('درآمد ساعتی (هزار تومان)، برش صدک ۹۹'))
ax.set_ylabel(fa('تعداد آگهی'))
ax.legend(fontsize=8)
heading(ax, '۷.۳', 'توزیع درآمد ساعتی')

ax = axes[1, 1]
experience = ads.Jobpost_RequiredExperienceYears.clip(upper=15)
counts = experience.value_counts().sort_index()
bars = ax.bar(counts.index, counts.values, color=GREEN, width=0.75)
ax.set_xlabel(fa('سال سابقه لازم (سقف ۱۵ سال نمایش)'))
ax.set_ylabel(fa('تعداد آگهی'))
ax.set_xticks(range(0, 16))
for position, value in zip(counts.index, counts.values):
    if position in (0, 1, 2, 3, 5, 10):
        ax.text(position, value, f'{value / 1000:.0f}k', ha='center', va='bottom', fontsize=8)
heading(ax, '۷.۴', 'توزیع سابقه کار مورد نیاز')
complete('7.1-7.4')

show_metrics('توزیع‌ها', [
    ('چولگی حقوق', f'{salary_series.skew():.2f}'),
    ('چولگی لگاریتم حقوق', f'{np.log(salary_series).skew():.2f}'),
    ('میانگین منهای میانه حقوق', f'{salary_series.mean() - salary_series.median():.2f} میلیون'),
    ('چولگی درآمد ساعتی', f'{hourly_series.skew():.2f}'),
    ('آگهی بدون نیاز به سابقه', f'{ads.Jobpost_RequiredExperienceYears.eq(0).mean() * 100:.1f}٪'),
])
```

**تفسیر جدول ۷.۱ و نمودارهای ۷.۱ تا ۷.۴:** توزیع حقوق چوله راست است: میانه ۱۱ و میانگین ۱۳٫۴ میلیون تومان، با چولگی ۲٫۶۸. فاصله ۲٫۴ میلیونی میانگین و میانه خودش بهترین شاهد است که «میانگین» در این داده گمراه‌کننده است؛ یک مدیر اگر میانگین بازار را مبنای حقوق بگیرد، برای نیمی از نقش‌ها بیش از بازار پیشنهاد می‌دهد. به همین دلیل در همه بخش‌ها میانه و چارک‌ها گزارش می‌شود.

گرفتن لگاریتم، توزیع را تقریباً نرمال می‌کند (چولگی از ۲٫۶۸ به ۰٫۴۳ می‌رسد) و همین توجیه فنی استفاده از لگاریتم حقوق به‌عنوان متغیر هدف در بخش ۱۶ است.

توزیع درآمد ساعتی نیز چوله است اما دامنه‌اش باریک‌تر: میانه ۶۳٫۵ و صدک ۹۹ حدود ۳۰۰ هزار تومان در ساعت. سابقه کار درخواستی هم چوله است: میانه ۲ سال، چارک سوم ۳ سال و ۱۵٫۸٪ آگهی‌ها بدون نیاز به سابقه‌اند؛ در مقابل فقط ۷۳۲ آگهی بیش از ۱۰ سال سابقه می‌خواهند. تصویر کلی این است که بازار کار ایران بازاری «کم‌سابقه» است: معمولاً نیروی تازه‌کار یا دو تا سه ساله می‌خواهند.

**نمودار ۷.۴** نشان می‌دهد توزیع سابقه در ۰ تا ۳ سال متمرکز است و پس از ۵ سال به‌سرعت نازک می‌شود؛ همین الگو در بخش ۸ به‌عنوان سرنخ بازده نزولی سابقه بررسی می‌شود.

## ۷.۲. شناسایی پرت‌ها

</div>

```python
def iqr_bounds(series, factor=1.5):
    q1, q3 = series.quantile(.25), series.quantile(.75)
    iqr = q3 - q1
    return q1 - factor * iqr, q3 + factor * iqr


outlier_rows = []
for series, label in [(paid.salary_mid, 'میانه بازه حقوق'), (hourly_ready.hourly_income, 'درآمد ساعتی'),
                      (ads.Jobpost_RequiredExperienceYears, 'سابقه لازم')]:
    low, high = iqr_bounds(series)
    flagged = (series < low) | (series > high)
    outlier_rows.append({'متغیر': label, 'تعداد معتبر': len(series),
                         'کران پایین': low, 'کران بالا': high,
                         'تعداد پرت': int(flagged.sum()), 'سهم پرت': flagged.mean() * 100,
                         'سهم بالای کران': (series > high).mean() * 100})
outlier_table = pd.DataFrame(outlier_rows).set_index('متغیر')
show_table('7.2', 'پرت‌ها بر پایه قاعده ۱٫۵ برابر دامنه میان‌چارکی', outlier_table,
           formats={'تعداد معتبر': 'int', 'کران پایین': 'money2', 'کران بالا': 'money2',
                    'تعداد پرت': 'int', 'سهم پرت': 'pct', 'سهم بالای کران': 'pct'},
           sample='داده در دسترس هر متغیر',
           note='پرت‌ها حذف نمی‌شوند؛ فقط در مدل‌سازی سطح حقوق وینزور می‌شوند.')
```

**تفسیر جدول ۷.۲:** قاعده ۱٫۵ برابر دامنه میان‌چارکی، ۲٬۳۸۲ آگهی (۲٪) را بالای کران ۳۰٫۳ میلیون تومان علامت می‌زند و ۱۱۵ آگهی را در کف (۲٫۵ میلیون یا کمتر). این سهم کوچک است و حذف آن‌ها ترکیب بازار را تغییر نمی‌دهد، اما حذفشان را توصیه نمی‌کنیم: آگهی ۶۰ میلیونی مدیرعامل یا ۲٫۵ میلیونی کارآموز، هر دو واقعیت بازارند. در توصیف، همه را نگه می‌داریم و در مدل‌سازی فقط صدک ۱ و ۹۹ را وینزور می‌کنیم تا یک ردیف نتواند ضریب رگرسیون را جابه‌جا کند.

## ۷.۳. سابقه کار و سن درخواستی

</div>

```python
experience_bucket = pd.cut(ads.Jobpost_RequiredExperienceYears, bins=[-1, 0, 1, 2, 3, 5, 10, 100],
                           labels=['بدون سابقه', '۱ سال', '۲ سال', '۳ سال', '۴ تا ۵', '۶ تا ۱۰', 'بیش از ۱۰'])
experience_table = (ads.groupby(experience_bucket, observed=True)
                    .agg(**{'تعداد آگهی': ('ad_id', 'size'), 'میانه حقوق': ('salary_mid', 'median'),
                            'نرخ افشای حقوق': ('salary_disclosed', lambda s: s.mean() * 100)}))
experience_table['سهم'] = experience_table['تعداد آگهی'] / len(ads) * 100
show_table('7.3', 'آگهی‌ها بر پایه سابقه کار درخواستی', experience_table,
           formats={'تعداد آگهی': 'int', 'سهم': 'pct', 'میانه حقوق': 'money', 'نرخ افشای حقوق': 'pct'},
           sample=f'{len(ads):,} آگهی', index_name='سابقه درخواستی')

age_bucket = pd.cut(ads.age_mid, bins=[17, 22, 27, 32, 37, 60],
                    labels=['زیر ۲۲', '۲۲ تا ۲۷', '۲۷ تا ۳۲', '۳۲ تا ۳۷', 'بالای ۳۷'])
age_table = (ads.groupby(age_bucket, observed=True)
             .agg(**{'تعداد آگهی': ('ad_id', 'size'), 'میانه حقوق': ('salary_mid', 'median'),
                     'میانه سابقه لازم': ('Jobpost_RequiredExperienceYears', 'median')}))
age_table['سهم'] = age_table['تعداد آگهی'] / len(ads) * 100
show_table('7.4', 'آگهی‌ها بر پایه میانه بازه سنی درخواستی', age_table,
           formats={'تعداد آگهی': 'int', 'سهم': 'pct', 'میانه حقوق': 'money', 'میانه سابقه لازم': 'hours'},
           sample=f'{int(ads.age_mid.notna().sum()):,} آگهی دارای بازه سنی', index_name='بازه سنی')

show_metrics('سابقه و سن', [
    ('میانه سابقه درخواستی', f'{ads.Jobpost_RequiredExperienceYears.median():.0f} سال'),
    ('آگهی با بیش از ۱۰ سال سابقه', f'{int(ads.Jobpost_RequiredExperienceYears.gt(10).sum()):,}'),
    ('آگهی دارای بازه سنی', f'{ads.age_mid.notna().mean() * 100:.1f}٪'),
    ('میانه سن درخواستی', f'{ads.age_mid.median():.1f} سال'),
])
```

**تفسیر جدول‌های ۷.۳ و ۷.۴:** سابقه درخواستی خیلی زود به سقف می‌رسد: ۵۲٪ آگهی‌ها حداکثر ۲ سال سابقه می‌خواهند و فقط حدود ۴٪ بیش از ۵ سال. در سمت حقوق، میانه از حدود ۱۱ میلیون برای آگهی‌های بدون سابقه به ۱۳٫۵ میلیون در آگهی‌های دارای ۳ سال سابقه می‌رسد؛ یعنی بیشترین جهش دستمزد در دو سال اول رخ می‌دهد و بعد از آن شیب کند می‌شود. این همان الگوی «بازده نزولی سابقه» است که در بخش ۸ نمودار می‌شود.

بازه سنی درخواستی فقط در ۷۹٫۵٪ آگهی‌ها اعلام شده و در همان‌ها هم محدود است: میانه ۳۰ سال و چارک‌های ۲۸٫۵ تا ۳۴ سال. یعنی کارفرمایان ایرانی عمدتاً نیروی جوان تا میان‌سال می‌خواهند و درخواست بالای ۴۰ سال بسیار کم است. این الگو در بخش ۱۳ به‌عنوان محدودیت بازار و در بخش ۲۲ به‌عنوان یک هشدار اجرایی (پرهیز از محدودکردن مصنوعی دامنه سنی) بازمی‌گردد.

# ۸. همبستگی، هم‌خطی و روابط غیرخطی

## ۸.۱. ماتریس همبستگی پیرسون و اسپیرمن

</div>

```python
corr_map = {
    'salary_mid': 'میانه حقوق', 'hourly_income': 'درآمد ساعتی', 'monthly_hours': 'ساعت ماهانه',
    'weekly_hours': 'ساعت هفتگی', 'Jobpost_RequiredExperienceYears': 'سابقه لازم',
    'age_mid': 'میانه سن', 'skill_count': 'تعداد مهارت', 'benefit_count': 'تعداد مزایا',
    'company_size_rank': 'اندازه شرکت', 'company_age_rank': 'قدمت شرکت',
}
corr_base = hourly_ready[list(corr_map)].rename(columns=corr_map)
pearson = corr_base.corr(method='pearson')
spearman = corr_base.corr(method='spearman')

fig, axes = plt.subplots(1, 2, figsize=(15, 5.6))
for ax, matrix, title in [(axes[0], pearson, '۸.۱ | همبستگی پیرسون'),
                          (axes[1], spearman, '۸.۲ | همبستگی اسپیرمن (رتبه‌ای)')]:
    sns.heatmap(matrix, ax=ax, cmap='crest', center=0, vmin=-1, vmax=1, annot=True, fmt='.2f',
                annot_kws={'size': 7.5}, cbar_kws={'shrink': 0.8}, square=True,
                xticklabels=[fa(c) for c in matrix.columns], yticklabels=[fa(c) for c in matrix.index])
    ax.set_title(fa(f'نمودار {title}'), fontsize=10.5, pad=10)
    ax.tick_params(labelsize=8)
    plt.setp(ax.get_xticklabels(), rotation=45, ha='right')
complete('8.1-8.2')

top_pairs = (pearson.where(np.triu(np.ones(pearson.shape), 1).astype(bool)).stack()
             .rename('پیرسون').to_frame())
top_pairs['اسپیرمن'] = [spearman.loc[a, b] for a, b in top_pairs.index]
top_pairs['اختلاف'] = top_pairs['اسپیرمن'] - top_pairs['پیرسون']
top_pairs = top_pairs.reindex(top_pairs['پیرسون'].abs().sort_values(ascending=False).index).head(12)
top_pairs.index = [f'{a} — {b}' for a, b in top_pairs.index]
show_table('8.1', 'دوازده جفت متغیر با بیشترین قدرمطلق همبستگی پیرسون', top_pairs,
           formats={'پیرسون': 'ratio', 'اسپیرمن': 'ratio', 'اختلاف': 'ratio'},
           sample=f'{len(hourly_ready):,} آگهی دارای حقوق و ساعت', period=False, index_name='جفت متغیر',
           note='اختلاف مثبت قابل توجه یعنی رابطه رتبه‌ای قوی‌تر از رابطه خطی است (نشانه غیرخطی‌بودن).')
```

**تفسیر نمودارهای ۸.۱ و ۸.۲ و جدول ۸.۱:** قوی‌ترین رابطه، رابطه منفی ساعت هفتگی با درآمد ساعتی است (−۰٫۶۸) که تا حدی مکانیکی است: درآمد ساعتی از تقسیم حقوق بر ساعت ساخته می‌شود. مهم‌تر از آن، اختلاف پیرسون و اسپیرمن در جفت «حقوق–درآمد ساعتی» است (۰٫۴۴ در برابر ۰٫۶۹)؛ این اختلاف نشان می‌دهد رابطه این دو متغیر غیرخطی است و رتبه‌ها بیشتر از مقادیر خطی هم‌جهت‌اند.

سابقه کار با حقوق ۰٫۴۰ همبستگی دارد؛ هیچ متغیری به‌تنهایی بیش از ۰٫۵ از تغییرات حقوق را توضیح نمی‌دهد. این خودش یک یافته است: حقوق یک متغیر تک‌عاملی نیست و ترکیبی از شغل، مهارت، جغرافیا، اندازه شرکت و سابقه است. به همین دلیل مدل چندمتغیره بخش ۱۶ لازم است.

**تفسیر جدول ۸.۲:** عامل تورم واریانس برای همه متغیرهای عددی زیر ۱٫۵ است؛ یعنی هم‌خطی چندگانه جدی وجود ندارد. این نتیجه کمی غیرشهودی است، چون معمولاً انتظار می‌رود سابقه و سن یا اندازه شرکت و تعداد مزایا هم‌پوشانی داشته باشند؛ اما در این داده همبستگی‌ها ملایم‌اند و هر متغیر اطلاعات نسبتاً مستقلی حمل می‌کند. بنابراین ضرایب مدل ریج در بخش ۱۶ قابل تفسیرند. محدودیت: این آزمون فقط روی متغیرهای عددی انجام شده و متغیرهای دسته‌ای را در بر نمی‌گیرد.

## ۸.۲. هم‌خطی چندگانه

</div>

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor

vif_cols = ['salary_mid', 'Jobpost_RequiredExperienceYears', 'age_mid', 'skill_count', 'benefit_count',
            'company_size_rank', 'company_age_rank', 'weekly_hours']
vif_source = hourly_ready[vif_cols].dropna()
vif_matrix = np.column_stack([np.ones(len(vif_source)), vif_source.values])
vif_values = [variance_inflation_factor(vif_matrix, position) for position in range(1, vif_matrix.shape[1])]
vif_table = pd.DataFrame({'عامل تورم واریانس': vif_values},
                         index=[corr_map[name] for name in vif_cols]).sort_values('عامل تورم واریانس', ascending=False)
vif_table['وضعیت'] = np.select([vif_table['عامل تورم واریانس'].ge(10), vif_table['عامل تورم واریانس'].ge(5)],
                               ['هم‌خطی شدید', 'هم‌خطی متوسط'], default='قابل قبول')
show_table('8.2', 'عامل تورم واریانس متغیرهای عددی', vif_table,
           formats={'عامل تورم واریانس': 'score'}, sample=f'{len(vif_source):,} آگهی کامل',
           period=False, index_name='متغیر',
           note='VIF بیش از ۵ نشانه هم‌خطی متوسط و بیش از ۱۰ نشانه هم‌خطی شدید است.')
```

## ۸.۳. روابط غیرخطی: بازده نزولی سابقه و سن

</div>

```python
experience_series = ads.loc[ads.salary_disclosed].groupby('Jobpost_RequiredExperienceYears').agg(
    **{'تعداد آگهی': ('ad_id', 'size'), 'میانه حقوق': ('salary_mid', 'median')})
experience_series = experience_series.loc[experience_series['تعداد آگهی'].ge(MIN_N)]

age_series = paid.groupby(pd.cut(paid.age_mid, bins=range(18, 46, 2)), observed=True).agg(
    **{'تعداد آگهی': ('ad_id', 'size'), 'میانه حقوق': ('salary_mid', 'median')})
age_series = age_series.loc[age_series['تعداد آگهی'].ge(MIN_N)]

fig, axes = plt.subplots(1, 2, figsize=(13.5, 4.6))
ax = axes[0]
ax.plot(experience_series.index, experience_series['میانه حقوق'], marker='o', color=TEAL, linewidth=1.8)
for x, y, n in zip(experience_series.index, experience_series['میانه حقوق'], experience_series['تعداد آگهی']):
    ax.annotate(f'{y:.1f}', (x, y), textcoords='offset points', xytext=(0, 7), ha='center', fontsize=7.5)
ax.set_xlabel(fa('سال سابقه کار درخواستی'))
ax.set_ylabel(fa('میانه حقوق (میلیون تومان)'))
ax.set_xticks(list(experience_series.index))
heading(ax, '۸.۳', 'میانه حقوق بر پایه سابقه درخواستی')
ax = axes[1]
ax.plot([interval.mid for interval in age_series.index], age_series['میانه حقوق'],
        marker='o', color=GOLD, linewidth=1.8)
ax.set_xlabel(fa('میانه بازه سنی درخواستی (سال)'))
ax.set_ylabel(fa('میانه حقوق (میلیون تومان)'))
heading(ax, '۸.۴', 'میانه حقوق بر پایه سن درخواستی')
complete('8.3-8.4')

show_table('8.3', 'میانه حقوق به تفکیک سابقه درخواستی', experience_series,
           formats={'تعداد آگهی': 'int', 'میانه حقوق': 'money'},
           sample=f'{int(experience_series["تعداد آگهی"].sum()):,} آگهی با حقوق', index_name='سال سابقه',
           note=f'فقط سال‌هایی که دست‌کم {MIN_N} آگهی دارند نشان داده شده‌اند.')
```

**تفسیر نمودارهای ۸.۳ و ۸.۴ و جدول ۸.۳:** منحنی سابقه، الگوی «بازده نزولی» را به‌وضوح نشان می‌دهد. میانه حقوق از ۹ میلیون در آگهی‌های بدون سابقه به ۱۱ میلیون در ۱ تا ۲ سال، ۱۳٫۵ میلیون در ۳ سال، ۱۷٫۵ میلیون در ۴ تا ۶ سال و ۲۲٫۵ میلیون در ۷ سال و بالاتر می‌رسد. جهش اصلی در سال‌های ۱، ۳ و ۴ رخ می‌دهد و بین ۴ تا ۶ سال تقریباً تخت می‌شود؛ یعنی بعد از پنج سال، هر سال سابقه به‌تنهایی ارزش افزوده چشمگیری نمی‌سازد و ارزش بیشتر به نوع نقش و مهارت وابسته است.

منحنی سن هم صعودی اما کم‌شیب است: میانه از ۹ میلیون در بازه ۲۲ تا ۲۸ سال به ۱۱ میلیون در ۲۸ تا ۳۴ سال و ۱۳٫۵ میلیون در ۳۴ تا ۴۰ سال می‌رسد. در بازه‌های بالای ۴۰ سال میانه به ۱۷٫۵ و ۲۲٫۵ میلیون می‌رسد، اما نمونه این بازه‌ها کوچک است و نباید به‌عنوان «اوج درآمدی» قطعی خوانده شود؛ تحلیل دقیق‌تر در بخش ۱۳ با تفکیک صنعت می‌آید.

نکته مهم برای مدیریت: اگر آگهی «۳ سال سابقه» می‌خواهد، تفاوت دستمزدی با «بدون سابقه» حدود ۵۰٪ است؛ اما اگر همان نقش را با ۵ سال سابقه بخواهید، تقریباً همان سطح حقوق ۳ سال را باید بپردازید. این یعنی **شرط سابقه بیش از ۳ سال، در بیشتر مشاغل ارزش پرداختی اضافه‌ای ایجاد نمی‌کند**.

# ۹. سازوکار گمشدگی حقوق: تصادفی یا وابسته؟

سه‌چهارم آگهی‌ها حقوق را اعلام نمی‌کنند. پرسش کلیدی این است که آیا این پنهان‌کاری با ویژگی‌های قابل مشاهده (اندازه شرکت، استان، دسته شغلی، نوع همکاری) مرتبط است یا نه. اگر گمشدگی وابسته و ساختارمند باشد، دو نتیجه مهم دارد: اولاً تحلیل حقوق به یک بخش خاص از بازار سوگیری پیدا می‌کند، ثانیاً خودِ «افشا یا عدم افشا» به یک رفتار قابل مطالعه کارفرما تبدیل می‌شود.

## ۹.۱. نرخ افشا در ابعاد مختلف

</div>

```python
ads['province_group'] = np.where(ads.Jobpost_ProvinceFa.eq('تهران'), 'تهران', 'سایر استان‌ها')
top_categories = ads.Jobpost_MainJobCategory.value_counts().head(10).index
ads['category_group'] = np.where(ads.Jobpost_MainJobCategory.isin(top_categories),
                                 ads.Jobpost_MainJobCategory, 'سایر دسته‌ها')
top_industries = ads.Jobpost_IndustryFa.value_counts().head(8).index
ads['industry_group'] = np.where(ads.Jobpost_IndustryFa.isin(top_industries),
                                 ads.Jobpost_IndustryFa, 'سایر صنایع')
sync_snapshots()


def disclosure_table(column, number, title, levels=None, sort_by='نرخ افشا'):
    table = (ads.groupby(column)
             .agg(**{'تعداد آگهی': ('ad_id', 'size'),
                     'نرخ افشا': ('salary_disclosed', lambda s: s.mean() * 100),
                     'میانه حقوق': ('salary_mid', 'median')}))
    if levels is not None:
        table = table.reindex(levels)
    else:
        table = table.sort_values(sort_by, ascending=False)
    show_table(number, title, table,
               formats={'تعداد آگهی': 'int', 'نرخ افشا': 'pct', 'میانه حقوق': 'money'},
               sample=f'{len(ads):,} آگهی', index_name='سطح')
    return table


size_disclosure = disclosure_table('Company_SizeFa', '9.1', 'نرخ افشای حقوق بر پایه اندازه شرکت',
                                   levels=list(SIZE_ORDER))
category_disclosure = disclosure_table('category_group', '9.2', 'نرخ افشای حقوق در ده دسته شغلی پرتکرار')
province_disclosure = disclosure_table('Jobpost_ProvinceFa', '9.3',
                                       'نرخ افشای حقوق در ده استان پرتکرار',
                                       levels=ads.Jobpost_ProvinceFa.value_counts().head(10).index)

other_dims = ['Jobpost_WorkTypeFa', 'seniority', 'Jobpost_IsRemote', 'Jobpost_IsInternship',
              'Jobpost_RequiredMilitaryServiceCard', 'ownership', 'activity_type', 'industry_group']
other_rows = []
for column in other_dims:
    table = (ads.groupby(column)
             .agg(**{'تعداد آگهی': ('ad_id', 'size'),
                     'نرخ افشا': ('salary_disclosed', lambda s: s.mean() * 100)}))
    for level, row in table.iterrows():
        other_rows.append({'بُعد': column, 'سطح': level, 'تعداد آگهی': int(row['تعداد آگهی']),
                           'نرخ افشا': row['نرخ افشا']})
other_table = pd.DataFrame(other_rows)
other_pivot = other_table.pivot(index='سطح', columns='بُعد', values='نرخ افشا')
other_pivot = other_pivot.rename(columns={'Jobpost_WorkTypeFa': 'نوع همکاری', 'seniority': 'رده سازمانی',
                                          'Jobpost_IsRemote': 'دورکاری', 'Jobpost_IsInternship': 'کارآموزی',
                                          'Jobpost_RequiredMilitaryServiceCard': 'الزام پایان خدمت',
                                          'ownership': 'مالکیت', 'activity_type': 'دامنه فعالیت',
                                          'industry_group': 'صنعت'})
show_table('9.4', 'نرخ افشای حقوق بر پایه ویژگی‌های دیگر (درصد)', other_pivot,
           formats={column: 'pct' for column in other_pivot.columns},
           sample=f'{len(ads):,} آگهی', period=False, index_name='سطح')
```

**تفسیر جدول‌های ۹.۱ تا ۹.۴:** گمشدگی حقوق کاملاً ساختارمند است. نرخ افشا با اندازه شرکت رابطه معکوس دارد: شرکت‌های ۱۱ تا ۵۰ نفره ۴۳٫۴٪ آگهی‌هایشان را با حقوق منتشر می‌کنند و شرکت‌های ۱۰۰۱ تا ۵۰۰۰ نفره فقط ۱۶٫۶٪. این تفاوت ۲۶٫۸ واحد درصدی، بزرگ‌ترین شکاف در همه ابعاد است: شرکت کوچک برای جذب در بازار نام‌شناخته، شفافیت را به‌عنوان ابزار رقابت به کار می‌برد، در حالی که شرکت بزرگ به برند و مسیر شغلی تکیه می‌کند.

دسته شغلی هم الگوی روشنی دارد: آگهی‌های کارگر ساده (۴۷٫۸٪)، فروشنده و ویزیتور (۴۴٫۶٪) و دیجیتال مارکتینگ (۴۲٫۳٪) شفاف‌ترین‌اند، در حالی که DevOps و پشتیبانی فنی (۱۸٫۲٪) و برنامه‌نویسی (۲۴٫۹٪) کمترین شفافیت را دارند. یعنی هرچه نقش دانش‌محورتر و حقوق بالاتر، افشای عمومی کمتر.

دورکاری بزرگ‌ترین تفاوت را در یک ویژگی دوجمله‌ای می‌سازد: ۵۲٫۵٪ آگهی‌های دورکاری حقوق را اعلام می‌کنند، در برابر ۳۱٫۶٪ آگهی‌های حضوری. آگهی دورکاری در رقابت با بازار جهانی است و شفافیت حقوق برایش ضروری‌تر می‌شود. استان‌ها تفاوت بزرگی نشان نمی‌دهند (۳۰٫۷٪ در البرز تا ۳۹٫۵٪ در آذربایجان شرقی)؛ بنابراین «جغرافیا» به‌تنهایی توضیح‌دهنده نیست.

## ۹.۲. آزمون معناداری: خی‌دو و اندازه اثر کرامر

</div>

```python
from scipy.stats import chi2_contingency

dimension_map = {'اندازه شرکت': 'Company_SizeFa', 'استان (تهران/سایر)': 'province_group',
                 'دسته شغلی': 'category_group', 'صنعت': 'industry_group',
                 'نوع همکاری': 'Jobpost_WorkTypeFa', 'رده سازمانی': 'seniority',
                 'دورکاری': 'Jobpost_IsRemote', 'کارآموزی': 'Jobpost_IsInternship',
                 'الزام پایان خدمت': 'Jobpost_RequiredMilitaryServiceCard', 'مالکیت': 'ownership',
                 'دامنه فعالیت': 'activity_type'}
chi_rows = []
for label, column in dimension_map.items():
    table = pd.crosstab(ads[column], ads.salary_disclosed)
    chi2, p_value, dof, _ = chi2_contingency(table)
    total = table.values.sum()
    cramers_v = math.sqrt(chi2 / (total * min(table.shape[0] - 1, table.shape[1] - 1)))
    chi_rows.append({'بُعد': label, 'تعداد سطوح': table.shape[0], 'آماره خی‌دو': chi2,
                     'درجه آزادی': dof, 'p-value': p_value, 'کرامر V': cramers_v})
chi_table = pd.DataFrame(chi_rows).set_index('بُعد').sort_values('کرامر V', ascending=False)
chi_table['قدرت رابطه'] = np.select([chi_table['کرامر V'].ge(0.3), chi_table['کرامر V'].ge(0.1)],
                                    ['قوی', 'متوسط'], default='ضعیف')
show_table('9.5', 'آزمون استقلال خی‌دو بین افشای حقوق و ویژگی‌های آگهی', chi_table,
           formats={'تعداد سطوح': 'int', 'آماره خی‌دو': 'int', 'درجه آزادی': 'int',
                    'p-value': 'pvalue', 'کرامر V': 'ratio'},
           sample=f'{len(ads):,} آگهی', period=False, index_name='بُعد',
           note='کرامر V زیر ۰٫۱ رابطه ضعیف، ۰٫۱ تا ۰٫۳ متوسط و بالای ۰٫۳ قوی تلقی می‌شود.')
```

**تفسیر جدول ۹.۵:** همه آزمون‌های خی‌دو در سطح ۰٫۰۰۱ معنادارند؛ یعنی فرض «گمشدگی تصادفی کامل (MCAR)» رد می‌شود. اما معناداری آماری با اندازه اثر یکی نیست: بزرگ‌ترین کرامر V متعلق به اندازه شرکت (۰٫۱۸۸) و دسته شغلی (۰٫۱۴۷) است که در محدوده «ضعیف تا متوسط» قرار می‌گیرد؛ دورکاری (۰٫۱۱۲) و کارآموزی (۰٫۰۹۴) در پی آن‌اند و بقیه ابعاد زیر ۰٫۱ هستند. یعنی وابستگی واقعی است، اما هیچ‌کدام از این ویژگی‌ها به‌تنهایی نرخ افشا را تعیین نمی‌کند.

نکته روش‌شناختی: با نمونه ۳۶۰ هزارتایی، حتی تفاوت‌های کوچک هم معنادار می‌شوند؛ به همین دلیل گزارش اندازه اثر در کنار p-value ضروری است.

## ۹.۳. مدل رگرسیون لجستیک برای احتمال افشای حقوق

</div>

```python
import statsmodels.api as sm
from sklearn.metrics import roc_auc_score

numeric_block = pd.DataFrame({
    'اندازه شرکت (رتبه)': ads.company_size_rank,
    'قدمت شرکت (رتبه)': ads.company_age_rank,
    'سابقه لازم': ads.Jobpost_RequiredExperienceYears,
    'تعداد مهارت': ads.skill_count,
    'تعداد مزایا': ads.benefit_count,
    'تعداد صنایع شرکت': ads.company_industry_count,
    'دورکاری': ads.Jobpost_IsRemote.astype(int),
    'کارآموزی': ads.Jobpost_IsInternship.astype(int),
    'اولویت نیروی بومی': ads.Jobpost_PriorityWithLocalCandidate.astype(int),
    'الزام پایان خدمت': ads.Jobpost_RequiredMilitaryServiceCard.astype(int),
    'دارای رشته تحصیلی': ads.has_academic.astype(int),
    'تهران': ads.province_group.eq('تهران').astype(int),
})
categorical_block = pd.get_dummies(ads[['Jobpost_WorkTypeFa', 'seniority', 'ownership', 'activity_type']],
                                   columns=['Jobpost_WorkTypeFa', 'seniority', 'ownership', 'activity_type'],
                                   drop_first=True, dtype=float)
logit_design = sm.add_constant(pd.concat([numeric_block, categorical_block], axis=1), has_constant='add')
logit_target = ads.salary_disclosed.astype(float)
logit_model = sm.Logit(logit_target, logit_design).fit(disp=0, maxiter=200)

coefficients = logit_model.params.drop('const')
p_values = logit_model.pvalues.drop('const')
odds = np.exp(coefficients)
strength = coefficients.abs().sort_values(ascending=False)
top_coefficients = pd.DataFrame({
    'ضریب': coefficients.loc[strength.index],
    'نسبت شانس': odds.loc[strength.index],
    'p-value': p_values.loc[strength.index],
}).head(15)
top_coefficients['اثر بر احتمال افشا'] = np.where(top_coefficients['ضریب'] > 0, 'افزایشی', 'کاهشی')
show_table('9.6', 'پانزده متغیر اثرگذار بر احتمال افشای حقوق', top_coefficients,
           formats={'ضریب': 'score', 'نسبت شانس': 'score', 'p-value': 'pvalue'},
           sample=f'{len(ads):,} آگهی', period=False, index_name='متغیر',
           note='نسبت شانس بزرگ‌تر از ۱ یعنی متغیر احتمال افشا را بالا می‌برد و کمتر از ۱ یعنی پایین می‌آورد.')

predicted = logit_model.predict(logit_design)
model_summary = pd.DataFrame({'مقدار': [
    f'{logit_model.prsquared:.3f}', f'{logit_model.llf:,.0f}', f'{len(logit_design.columns) - 1}',
    f'{roc_auc_score(logit_target, predicted):.3f}',
    f'{(predicted.round().eq(logit_target).mean() * 100):.1f}٪',
    f'{max(logit_target.mean(), 1 - logit_target.mean()) * 100:.1f}٪']},
    index=['شبه R² مک‌فادن', 'لگاریتم درست‌نمایی', 'تعداد متغیر', 'سطح زیر منحنی ROC',
           'دقت طبقه‌بندی مدل', 'دقت مدل پایه (اکثریت)'])
show_table('9.7', 'کارایی مدل رگرسیون لجستیک', model_summary, sample=f'{len(ads):,} آگهی', period=False,
           index_name='شاخص')

show_metrics('افشای حقوق', [
    ('نرخ افشای کل', f'{ads.salary_disclosed.mean() * 100:.1f}٪'),
    ('بیشترین نرخ بر پایه اندازه شرکت', f'{size_disclosure["نرخ افشا"].idxmax()} — {size_disclosure["نرخ افشا"].max():.1f}٪'),
    ('کمترین نرخ بر پایه اندازه شرکت', f'{size_disclosure["نرخ افشا"].idxmin()} — {size_disclosure["نرخ افشا"].min():.1f}٪'),
    ('قوی‌ترین بُعد (کرامر V)', f'{chi_table.index[0]} — {chi_table["کرامر V"].iloc[0]:.3f}'),
    ('سطح زیر منحنی ROC', f'{roc_auc_score(logit_target, predicted):.3f}'),
])
```

**تفسیر جدول‌های ۹.۶ و ۹.۷ و پاسخ نهایی:** مدل رگرسیون لجستیک نشان می‌دهد شانس افشای حقوق برای آگهی‌های کارگر ساده ۲٫۷۶ برابر، آگهی‌های دورکاری ۲٫۳۶ برابر و فروشنده/ویزیتور ۲٫۳۳ برابر سطح پایه است؛ در مقابل، آگهی‌های پاره‌وقت شانس کمتری (۰٫۶۵) برای افشا دارند. همه این ضرایب از نظر آماری معنادارند، اما کارایی مدل محدود است: شبه‌R² مک‌فادن ۰٫۰۵ و سطح زیر منحنی ROC ۰٫۶۵ است و دقت طبقه‌بندی (۶۸٫۱٪) فقط ۱٫۲ واحد درصد از مدل پایه (همیشه «افشا نمی‌شود») بهتر است.

**پاسخ به پرسش پروژه:** گمشدگی حقوق **تصادفی کامل (MCAR) نیست**؛ به متغیرهای قابل مشاهده — به‌ویژه اندازه شرکت، دسته شغلی و دورکاری — وابسته است. بنابراین مدل گمشدگی از نوع **MAR** (وابسته به داده‌های مشاهده‌شده) است. اما سهم توضیح‌داده‌شده کوچک است؛ یعنی بخش بزرگی از تصمیم افشا، سیاست داخلی و ویژگی‌های مشاهده‌نشدنی هر کارفرماست.

**پیامد عملی:** میانگین‌ها و میانه‌های حقوق در این داده، به سمت شرکت‌های کوچک، نقش‌های اداری/فروش/خدماتی، آگهی‌های دورکاری و قراردادی سوگیری دارند. برای نقش‌های تخصصی و شرکت‌های بزرگ، این اعداد کف بازار را نشان می‌دهند، نه سطح واقعی رقابت. در بخش ۲۲ این محدودیت در توصیه‌های حقوقی لحاظ می‌شود.

# ۱۰. روند زمانی، فصلی‌بودن و پویایی دستمزد

## ۱۰.۱. حجم آگهی در زمان

</div>

```python
monthly_trend = (ads.groupby('release_month')
                 .agg(**{'تعداد آگهی': ('ad_id', 'size'),
                         'میانه حقوق': ('salary_mid', 'median'),
                         'نرخ افشا': ('salary_disclosed', lambda s: s.mean() * 100)}))
monthly_trend['ماه شمسی'] = [jmonth(month)[2] for month in monthly_trend.index]
monthly_trend['سهم'] = monthly_trend['تعداد آگهی'] / len(ads) * 100
monthly_trend['رشد ماهانه'] = monthly_trend['تعداد آگهی'].pct_change() * 100
monthly_trend['میانگین متحرک ۳ ماهه'] = monthly_trend['تعداد آگهی'].rolling(3).mean()
show_table('10.1', 'روند ماهانه حجم آگهی و میانه حقوق', monthly_trend,
           formats={'تعداد آگهی': 'int', 'میانه حقوق': 'money', 'نرخ افشا': 'pct', 'سهم': 'pct',
                    'رشد ماهانه': 'pp', 'میانگین متحرک ۳ ماهه': 'hourly'},
           sample=f'{len(ads):,} آگهی', index_name='ماه میلادی')

fig, ax = plt.subplots(figsize=(13, 5))
x = np.arange(len(monthly_trend))
ax.bar(x, monthly_trend['تعداد آگهی'], color=LIGHT, edgecolor=TEAL, width=0.72, label=fa('تعداد آگهی'))
ax.plot(x, monthly_trend['میانگین متحرک ۳ ماهه'], color=GOLD, linewidth=2, label=fa('میانگین متحرک ۳ ماهه'))
ax.set_xticks(x)
ax.set_xticklabels([fa(label) for label in monthly_trend['ماه شمسی']], rotation=60, ha='right', fontsize=8)
ax.set_ylabel(fa('تعداد آگهی'))
peak_month = monthly_trend['تعداد آگهی'].idxmax()
peak_position = int(monthly_trend.index.get_loc(peak_month))
ax.annotate(fa(f'اوج: {monthly_trend.loc[peak_month, "ماه شمسی"]}'),
            xy=(peak_position, monthly_trend['تعداد آگهی'].max()),
            xytext=(peak_position - 5.5, monthly_trend['تعداد آگهی'].max() * 1.03),
            fontsize=8.5, arrowprops=dict(arrowstyle='->', color=DARK))
ax.legend(fontsize=8.5)
heading(ax, '۱۰.۱', 'روند ماهانه تعداد آگهی و میانگین متحرک')
complete('10.1')
```

**تفسیر جدول ۱۰.۱ و نمودار ۱۰.۱:** حجم آگهی‌ها از ۱۲٬۷۸۲ در فروردین ۱۴۰۱ به ۲۴٬۰۱۷ در مهر ۱۴۰۲ رسیده است؛ نزدیک ۸۸٪ رشد در ۱۹ ماه. روند یکنواخت نیست: اسفند ۱۴۰۱ با ۵٬۵۳۱ آگهی کف دوره است و فروردین ۱۴۰۲ با جهش ۲۲۹٪ نسبت به آن، نشان می‌دهد این افت بیشتر فصلی و موقتی بوده است. میانگین متحرک سه‌ماهه هم صعود بلندمدت را تأیید می‌کند.

ریزش بهمن و اسفند و جهش فروردین، یک الگوی آشنای بازار کار ایران است: کارفرمایان پیش از تعطیلات نوروز استخدام را متوقف می‌کنند و بلافاصله پس از آن، آگهی‌های انباشته منتشر می‌شوند.

## ۱۰.۲. فصلی‌بودن با تقویم شمسی

</div>

```python
seasonal = (ads.loc[ads.jmonth_num.le(11)]
            .groupby(['jmonth_num', 'jmonth_label', 'jyear'])
            .agg(**{'تعداد آگهی': ('ad_id', 'size')})
            .reset_index())
seasonal_pivot = seasonal.pivot_table(index='jmonth_num', columns='jyear', values='تعداد آگهی', fill_value=0)
seasonal_pivot.index = [J_MONTHS[index - 1] for index in seasonal_pivot.index]
seasonal_pivot.columns = [f'سال {fa_num(year)}' for year in seasonal_pivot.columns]
seasonal_mean = seasonal_pivot.mean(axis=1)
seasonal_table = pd.DataFrame({'میانگین دو سال': seasonal_mean})
seasonal_table['شاخص فصلی'] = seasonal_table['میانگین دو سال'] / seasonal_table['میانگین دو سال'].mean() * 100
seasonal_table['کمینه تا بیشینه سال‌ها'] = [f'{seasonal_pivot.loc[month].min():,.0f} تا {seasonal_pivot.loc[month].max():,.0f}'
                                             for month in seasonal_table.index]
show_table('10.2', 'الگوی فصلی انتشار آگهی بر پایه ماه شمسی', seasonal_table,
           formats={'میانگین دو سال': 'int', 'شاخص فصلی': 'pct'},
           sample='میانگین ماه‌های کامل دو سال', period=False, index_name='ماه شمسی',
           note='اسفند در شاخص فصلی نیامده است، چون اسفند ۱۴۰۲ در داده کامل نیست و مقایسه دو سال ممکن نبود.')

fig, ax = plt.subplots(figsize=(12, 4.8))
positions = np.arange(len(seasonal_pivot))
width = 0.38
for offset, (year, color) in enumerate(zip(seasonal_pivot.columns, [TEAL, GOLD])):
    ax.bar(positions + (offset - 0.5) * width, seasonal_pivot[year], width=width,
           color=color, label=fa(year), alpha=0.9)
ax.plot(positions, seasonal_mean.values, color=DARK, linewidth=1.6, marker='o', markersize=3.5,
        label=fa('میانگین دو سال'))
ax.set_xticks(positions)
ax.set_xticklabels([fa(month) for month in seasonal_pivot.index], fontsize=8.5)
ax.set_ylabel(fa('تعداد آگهی'))
ax.legend(fontsize=8.5)
heading(ax, '۱۰.۲', 'الگوی فصلی انتشار آگهی در دو سال')
complete('10.2')

season_table = (ads.groupby(['jyear', 'jseason'])
                .agg(**{'تعداد آگهی': ('ad_id', 'size')})
                .reset_index())
season_table['سهم'] = season_table['تعداد آگهی'] / season_table.groupby('jyear')['تعداد آگهی'].transform('sum') * 100
season_pivot = season_table.pivot(index='jseason', columns='jyear', values='تعداد آگهی').reindex(SEASONS)
season_share = season_table.pivot(index='jseason', columns='jyear', values='سهم').reindex(SEASONS)
season_pivot.columns = [f'سال {fa_num(year)}' for year in season_pivot.columns]
season_share.columns = [f'سال {fa_num(year)} (٪)' for year in season_share.columns]
show_table('10.3', 'تعداد و سهم آگهی در فصل‌های سال شمسی', season_pivot.join(season_share),
           formats={column: 'int' for column in season_pivot.columns} |
                   {column: 'pct' for column in season_share.columns},
           sample=f'{len(ads):,} آگهی', period=False, index_name='فصل',
           note='سهم فصل‌ها در هر سال جدا محاسبه شده است؛ اسفند ۱۴۰۲ ناقص است و سهم زمستان ۱۴۰۲ را کمی کم‌نمایی می‌کند.')
```

**تفسیر جدول‌های ۱۰.۲ و ۱۰.۳ و نمودار ۱۰.۲:** شاخص فصلی، اوج روشنی در **دی** (۱۲۳٪) و سپس **مهر و آبان** (۱۱۵ و ۱۱۲٪) نشان می‌دهد؛ کمترین ماه‌ها **بهمن** (۷۹٪)، **خرداد** (۸۶٪) و **شهریور** (۸۸٪) هستند. به بیان دیگر، بازار کار در نیمه دوم سال و پیش از پایان سال شمسی پرجنب‌وجوش‌تر است و در آستانه نوروز و تعطیلات تابستانی آرام می‌شود.

اسفند را عمداً از شاخص فصلی کنار گذاشتیم: اسفند ۱۴۰۱ فقط ۵٬۵۳۱ آگهی داشت و اسفند ۱۴۰۲ در داده کامل نیست؛ اگر آن را وارد می‌کردیم، شاخص به‌دلیل یک مشاهده، عددی گمراه‌کننده (۱۸٪) می‌شد.

**هشدار روش‌شناختی:** چون حجم کل بازار در این دو سال رشد کرده، ماه‌های پایانی هر سال به‌طور طبیعی آگهی بیشتری دارند. شاخص فصلی روند رشد را کامل حذف نمی‌کند؛ بنابراین آن را به‌عنوان «جهت الگوی فصلی» بخوانید، نه ضریب دقیق پیش‌بینی.

## ۱۰.۳. پویایی دستمزد و مقایسه با تورم

</div>

```python
salary_monthly = (paid.groupby('release_month')
                  .agg(**{'تعداد آگهی دارای حقوق': ('ad_id', 'size'),
                          'میانه حقوق': ('salary_mid', 'median'),
                          'چارک اول': ('salary_mid', lambda s: s.quantile(.25)),
                          'چارک سوم': ('salary_mid', lambda s: s.quantile(.75)),
                          'میانگین حقوق': ('salary_mid', 'mean')}))
salary_monthly['ماه شمسی'] = [jmonth(month)[2] for month in salary_monthly.index]
show_table('10.4', 'میانه و چارک‌های حقوق در ماه‌های دوره', salary_monthly,
           formats={'تعداد آگهی دارای حقوق': 'int', 'میانه حقوق': 'money', 'چارک اول': 'money',
                    'چارک سوم': 'money', 'میانگین حقوق': 'money'},
           sample=f'{len(paid):,} آگهی با حقوق اعلام‌شده', index_name='ماه میلادی')

fig, ax = plt.subplots(figsize=(13, 5))
x = np.arange(len(salary_monthly))
ax.fill_between(x, salary_monthly['چارک اول'], salary_monthly['چارک سوم'], color=LIGHT, alpha=0.9,
                label=fa('چارک اول تا سوم'))
ax.plot(x, salary_monthly['میانه حقوق'], color=TEAL, linewidth=2.2, marker='o', markersize=3,
        label=fa('میانه حقوق'))
ax.plot(x, salary_monthly['میانگین حقوق'], color=RED, linewidth=1.4, linestyle='--',
        label=fa('میانگین حقوق'))
ax.set_xticks(x)
ax.set_xticklabels([fa(label) for label in salary_monthly['ماه شمسی']], rotation=60, ha='right', fontsize=8)
ax.set_ylabel(fa('حقوق (میلیون تومان)'))
ax.legend(fontsize=8.5)
heading(ax, '۱۰.۳', 'روند میانه و میانگین حقوق ماهانه')
complete('10.3')

year_salary = (paid.groupby(['jyear', 'jmonth_num'])
               .agg(**{'تعداد': ('ad_id', 'size'), 'میانه حقوق': ('salary_mid', 'median')})
               .reset_index())
common_months = sorted(set(year_salary.loc[year_salary.jyear.eq(1401), 'jmonth_num']) &
                       set(year_salary.loc[year_salary.jyear.eq(1402), 'jmonth_num']))
comparison = (year_salary[year_salary.jmonth_num.isin(common_months)]
              .pivot(index='jmonth_num', columns='jyear', values='میانه حقوق'))
comparison.index = [J_MONTHS[index - 1] for index in comparison.index]
comparison['رشد اسمی'] = (comparison[1402] / comparison[1401] - 1) * 100
comparison['نسبت'] = comparison[1402] / comparison[1401]
comparison.columns = ['میانه ۱۴۰۱', 'میانه ۱۴۰۲', 'رشد اسمی', 'نسبت']

growth_nominal = (comparison['میانه ۱۴۰۲'].mean() / comparison['میانه ۱۴۰۱'].mean() - 1) * 100
show_table('10.5', 'مقایسه میانه حقوق ماه‌های هم‌نام دو سال', comparison,
           formats={'میانه ۱۴۰۱': 'money', 'میانه ۱۴۰۲': 'money',
                    'رشد اسمی': 'pp', 'نسبت': 'score'},
           sample=f'{len(common_months)} ماه هم‌نام', period=False, index_name='ماه شمسی',
           note='رشد اسمی حقوق بدون تعدیل تورم محاسبه شده است؛ مقایسه با تورم در متن تفسیر می‌شود.')

inflations = [30, 35, 40, 45, 50]
real_growth = pd.DataFrame({'تورم فرضی سالانه': [f'{value}٪' for value in inflations],
                            'رشد حقیقی حقوق': [growth_nominal - value for value in inflations]})
show_table('10.6', 'رشد حقیقی حقوق در سناریوهای تورم', real_growth.set_index('تورم فرضی سالانه'),
           formats={'رشد حقیقی حقوق': 'pp'}, sample=f'رشد اسمی {growth_nominal:.1f}٪',
           period=False, index_name='تورم فرضی')

show_metrics('دستمزد در زمان', [
    ('رشد اسمی میانه حقوق', f'{growth_nominal:.1f}٪ بین دو سال'),
    ('میانه حقوق ابتدای دوره', f'{comparison["میانه ۱۴۰۱"].iloc[0]:,.1f} میلیون'),
    ('میانه حقوق پایان دوره', f'{salary_monthly["میانه حقوق"].iloc[-1]:,.1f} میلیون'),
    ('میانه رشد ماهانه', f'{salary_monthly["میانه حقوق"].pct_change().median() * 100:,.1f}٪'),
])
```

**تفسیر جدول‌های ۱۰.۴ تا ۱۰.۶ و نمودار ۱۰.۳:** میانه حقوق از ۷ میلیون تومان در فروردین ۱۴۰۱ به ۱۳٫۵ میلیون در بهمن ۱۴۰۲ رسیده است؛ در مقایسه ماه‌های هم‌نام، رشد اسمی **۵۳٫۷٪** است. اگر تورم سالانه را ۴۰٪ فرض کنیم، رشد حقیقی حدود **+۱۳٫۷٪** می‌شود و اگر ۵۰٪ فرض کنیم، تنها **+۳٫۷٪** باقی می‌ماند. یعنی دستمزدهای پیشنهادی در این دوره از تورم ۴۰ درصدی جلوتر، اما از تورم ۵۰ درصدی عقب‌تر بوده‌اند.

**هشدار:** این رشد ترکیبی از سه چیز است: رشد واقعی دستمزد، تغییر ترکیب آگهی‌ها به سمت نقش‌های گران‌تر، و گردکردن میانه‌ها روی مقادیر گسسته (۹، ۱۱، ۱۳٫۵، ۱۷٫۵). عدد ۵۳٫۷٪ را نباید رشد دستمزد «همان شغل» تفسیر کرد؛ برای آن، داده پنل یا تکرار مشاهده یک کارفرما لازم است.

## ۱۰.۴. تغییر سهم دسته‌های شغلی در زمان

</div>

```python
category_year = (ads.groupby(['category_group', 'jyear'])
                 .agg(**{'تعداد آگهی': ('ad_id', 'size')})
                 .reset_index())
year_totals = category_year.groupby('jyear')['تعداد آگهی'].transform('sum')
category_year['سهم'] = category_year['تعداد آگهی'] / year_totals * 100
share_pivot = category_year.pivot(index='category_group', columns='jyear', values='سهم').fillna(0)
count_pivot = category_year.pivot(index='category_group', columns='jyear', values='تعداد آگهی').fillna(0)
share_change = share_pivot.copy()
share_change.columns = ['سهم ۱۴۰۱', 'سهم ۱۴۰۲']
share_change['تغییر واحد درصدی'] = share_change['سهم ۱۴۰۲'] - share_change['سهم ۱۴۰۱']
share_change['تعداد ۱۴۰۲'] = count_pivot[1402]
share_change = share_change.sort_values('تعداد ۱۴۰۲', ascending=False).head(12)
show_table('10.7', 'تغییر سهم دسته‌های شغلی پرتکرار بین دو سال', share_change,
           formats={'سهم ۱۴۰۱': 'pct', 'سهم ۱۴۰۲': 'pct', 'تغییر واحد درصدی': 'pp', 'تعداد ۱۴۰۲': 'int'},
           sample=f'{len(ads):,} آگهی', period=False, index_name='دسته شغلی')

fig, ax = plt.subplots(figsize=(11.5, 6.4))
plot_data = share_change.sort_values('تغییر واحد درصدی')
colors = [RED if value < 0 else TEAL for value in plot_data['تغییر واحد درصدی']]
ax.barh(range(len(plot_data)), plot_data['تغییر واحد درصدی'], color=colors, height=0.7)
ax.set_yticks(range(len(plot_data)))
ax.set_yticklabels([fa(name) for name in plot_data.index], fontsize=8.5)
ax.axvline(0, color=DARK, linewidth=0.9)
ax.set_xlabel(fa('تغییر سهم ۱۴۰۲ نسبت به ۱۴۰۱ (واحد درصدی)'))
for position, value in enumerate(plot_data['تغییر واحد درصدی']):
    offset = 0.08 if value >= 0 else -0.08
    ax.text(value + offset, position, f'{value:+.1f}', va='center',
            ha='left' if value >= 0 else 'right', fontsize=7.5)
ax.set_xlim(plot_data['تغییر واحد درصدی'].min() - 0.9, plot_data['تغییر واحد درصدی'].max() + 0.9)
heading(ax, '۱۰.۴', 'تغییر سهم دسته‌های شغلی بین دو سال', period='۱۴۰۱ در برابر ۱۴۰۲')
complete('10.4')
```

**تفسیر جدول ۱۰.۷ و نمودار ۱۰.۴:** ترکیب تقاضا در دو سال تغییر کرده است. بزرگ‌ترین افت‌ها متعلق به فروش در سطوح کارشناسی و مدیریتی (−۳٫۱ واحد درصد)، برنامه‌نویسی (−۱٫۷)، مهندسی صنایع (−۰٫۹) و DevOps (−۰٫۶) است؛ بزرگ‌ترین رشدها به دیجیتال مارکتینگ (+۱٫۶)، فروشنده و ویزیتور (+۱٫۱)، مالی و حسابداری (+۰٫۶) و کارمند اداری (+۰٫۵) تعلق دارد.

این جابه‌جایی دو پیام دارد. اول، بازار از نقش‌های «فروش ارشد» و «برنامه‌نویسی خالص» به سمت نقش‌های اجرایی و دیجیتال حرکت کرده است. دوم، رشد دیجیتال مارکتینگ با رشد سهم مهارت‌های WordPress و Google Analytics در بخش ۱۹ هم‌خوان است؛ یعنی این تغییر ترکیبی، پشتوانه مهارتی هم دارد.

**محدودیت:** کاهش سهم برنامه‌نویسی به معنی کاهش تقاضای مطلق نیست؛ تعداد آگهی‌های این دسته از ۷٬۷۹۱ در ۱۴۰۲ بیشتر از سال قبل است، اما سرعت رشدش از بازار عقب مانده و سهمش کم شده است.

# ۱۱. ترجیح جنسیت، فرصت شغلی و دستمزد

**هشدار تفسیری:** داده ترجیح کارفرما در متن آگهی را ثبت می‌کند، نه جنسیت نیروی استخدام‌شده و نه حقوق فردی. بنابراین هر تفاوتی که می‌بینیم، تفاوت **میان دسته‌های آگهی** است، نه شکاف دستمزد افراد. این تفکیک در همه تفسیرهای این بخش رعایت می‌شود.

## ۱۱.۱. سهم ترجیح جنسیت در آگهی‌ها

</div>

```python
GENDER_LABELS = {'تفاوتی ندارد': 'بدون ترجیح', 'فقط آقا': 'فقط آقا', 'ترجیحاً آقا': 'ترجیحاً آقا',
                 'فقط خانم': 'فقط خانم', 'ترجیحاً خانم': 'ترجیحاً خانم'}
ads['gender_pref'] = ads.Jobpost_PreferredGender.map(GENDER_LABELS)
sync_snapshots()
gender_order = ['بدون ترجیح', 'ترجیحاً آقا', 'ترجیحاً خانم', 'فقط آقا', 'فقط خانم']

gender_table = (ads.groupby('gender_pref')
                .agg(**{'تعداد آگهی': ('ad_id', 'size'), 'میانه حقوق': ('salary_mid', 'median'),
                        'میانه سابقه': ('Jobpost_RequiredExperienceYears', 'median'),
                        'نرخ افشا': ('salary_disclosed', lambda s: s.mean() * 100)})
                .reindex(gender_order))
gender_table['سهم'] = gender_table['تعداد آگهی'] / len(ads) * 100
show_table('11.1', 'توزیع آگهی‌ها بر پایه ترجیح جنسیت', gender_table,
           formats={'تعداد آگهی': 'int', 'سهم': 'pct', 'میانه حقوق': 'money',
                    'میانه سابقه': 'hours', 'نرخ افشا': 'pct'},
           sample=f'{len(ads):,} آگهی', index_name='ترجیح جنسیت')

gender_category = (ads.assign(is_restricted=ads.gender_pref.isin(['فقط آقا', 'فقط خانم']))
                   .groupby('category_group')
                   .agg(**{'تعداد آگهی': ('ad_id', 'size'),
                           'سهم محدودیت جنسیتی': ('is_restricted', lambda s: s.mean() * 100),
                           'سهم فقط آقا': ('gender_pref', lambda s: s.eq('فقط آقا').mean() * 100),
                           'سهم فقط خانم': ('gender_pref', lambda s: s.eq('فقط خانم').mean() * 100)})
                   .sort_values('تعداد آگهی', ascending=False))
show_table('11.2', 'محدودیت جنسیتی در ده دسته شغلی پرتکرار', gender_category,
           formats={'تعداد آگهی': 'int', 'سهم محدودیت جنسیتی': 'pct', 'سهم فقط آقا': 'pct',
                    'سهم فقط خانم': 'pct'}, sample=f'{len(ads):,} آگهی', index_name='دسته شغلی')

fig, ax = plt.subplots(figsize=(11, 5.6))
plot_data = gender_category.sort_values('سهم محدودیت جنسیتی')
positions = np.arange(len(plot_data))
ax.barh(positions, -plot_data['سهم فقط آقا'], color=BLUE, height=0.72, label=fa('فقط آقا'))
ax.barh(positions, plot_data['سهم فقط خانم'], color=GOLD, height=0.72, label=fa('فقط خانم'))
ax.set_yticks(positions)
ax.set_yticklabels([fa(name) for name in plot_data.index], fontsize=8.5)
ax.axvline(0, color=DARK, linewidth=1)
ax.set_xlabel(fa('درصد آگهی‌های دارای شرط جنسیت (آقا در سمت راست منفی)'))
ax.set_xticklabels([fa(f'{abs(int(float(label.get_text())))}٪') for label in ax.get_xticklabels()])
ax.legend(fontsize=8.5, loc='lower right')
heading(ax, '۱۱.۱', 'شرط جنسیت صریح در دسته‌های شغلی پرتکرار')
complete('11.1')
```

**تفسیر جدول‌های ۱۱.۱ و ۱۱.۲ و نمودار ۱۱.۱:** در ۵۵٫۷٪ آگهی‌ها هیچ ترجیح جنسیتی وجود ندارد؛ ۳۳٫۵٪ آگهی‌ها شرط صریح «فقط آقا» یا «فقط خانم» دارند و بقیه ترجیح ملایم‌اند. نسبت آگهی‌های «فقط آقا» به «فقط خانم» ۱٫۶۱ است؛ یعنی بازار در سطح آگهی، به سمت مردان بسته‌تر است.

اما الگوی مهم‌تر، **تفکیک شغلی** است: در دسته تکنسین فنی و تعمیرکار ۷۹٫۱٪ آگهی‌ها شرط جنسیت دارند (۷۷٫۴٪ فقط آقا) و در کارگر ساده ۶۸٪ (۵۶٫۸٪ فقط آقا)؛ در مقابل، در مسئول دفتر و کارمند اداری ۶۳٫۲٪ آگهی‌ها شرط دارند که ۵۶٫۵٪ آن «فقط خانم» است. در برنامه‌نویسی فقط ۴٫۳٪ آگهی‌ها شرط جنسیت دارند. یعنی تبعیض بیش از آنکه در «دستمزد همان نقش» باشد، در «دسترسی به نقش‌ها» است.

## ۱۱.۲. آزمون فرض دستمزد میان آگهی‌های زن‌محور و مرد‌محور

</div>

```python
from scipy.stats import ttest_ind, mannwhitneyu

male_only = paid.loc[paid.gender_pref.eq('فقط آقا'), 'salary_mid'].dropna()
female_only = paid.loc[paid.gender_pref.eq('فقط خانم'), 'salary_mid'].dropna()
no_pref = paid.loc[paid.gender_pref.eq('بدون ترجیح'), 'salary_mid'].dropna()

t_stat, t_p = ttest_ind(male_only, female_only, equal_var=False)
u_stat, u_p = mannwhitneyu(male_only, female_only)
pooled_sd = math.sqrt((male_only.var() + female_only.var()) / 2)
cohens_d = (male_only.mean() - female_only.mean()) / pooled_sd
median_gap = (male_only.median() / female_only.median() - 1) * 100

test_table = pd.DataFrame({
    'مقدار': [f'{len(male_only):,}', f'{len(female_only):,}', f'{male_only.median():,.1f}',
              f'{female_only.median():,.1f}', f'{median_gap:+,.1f}٪', f'{male_only.mean() - female_only.mean():+,.2f}',
              f'{t_stat:,.2f}', f'{t_p:.2e}', f'{u_p:.2e}', f'{cohens_d:+.3f}'],
}, index=['تعداد آگهی فقط آقا', 'تعداد آگهی فقط خانم', 'میانه حقوق فقط آقا', 'میانه حقوق فقط خانم',
          'اختلاف میانه', 'اختلاف میانگین', 'آماره t (ولش)', 'p-value آزمون t',
          'p-value مان-ویتنی', 'اندازه اثر کوهن d'])
show_table('11.3', 'آزمون تفاوت دستمزد میان آگهی‌های «فقط آقا» و «فقط خانم»', test_table,
           sample=f'{len(male_only) + len(female_only):,} آگهی دارای حقوق و شرط جنسیت', period=False,
           index_name='شاخص آزمون',
           note='فرض صفر: میانه/میانگین حقوق در دو گروه برابر است. رد فرض صفر یعنی تفاوت معنادار آماری.')

fig, ax = plt.subplots(figsize=(9.5, 4.6))
groups = [paid.loc[paid.gender_pref.eq(label), 'salary_mid'].dropna() for label in gender_order]
violin = ax.violinplot(groups, showmedians=True, widths=0.85)
for body, color in zip(violin['bodies'], [TEAL, BLUE, GOLD, '#3f6fa8', '#b07a3f']):
    body.set_facecolor(color)
    body.set_alpha(0.6)
ax.set_xticks(range(1, len(gender_order) + 1))
ax.set_xticklabels([fa(label) for label in gender_order], fontsize=8.5)
ax.set_ylabel(fa('میانه بازه حقوق (میلیون تومان)'))
ax.set_ylim(0, 45)
for position, group in enumerate(groups, start=1):
    ax.text(position, 42, fa(f'n={len(group):,}'), ha='center', fontsize=7.5)
heading(ax, '۱۱.۲', 'توزیع حقوق به تفکیک ترجیح جنسیت')
complete('11.2')
```

**تفسیر جدول‌های ۱۱.۴ و ۱۱.۵:** وقتی شکاف را درون دسته‌های شغلی حساب می‌کنیم، در بیشتر دسته‌ها حدود **+۲۲ تا +۲۳٪** باقی می‌ماند (مالی، فروش، اداری، صنایع)؛ یعنی ترکیب شغلی همه تفاوت را توضیح نمی‌دهد. دو استثنا مهم است: در برنامه‌نویسی شکاف به **+۵۹٪** می‌رسد (میانه ۱۷٫۵ در برابر ۱۱)، اما فقط ۳۹ آگهی «فقط خانم» در این دسته وجود دارد و نمونه کوچک است؛ در کارگر ساده شکاف صفر است (هر دو ۹ میلیون).

در سطح رده سازمانی، شکاف با بالا رفتن رده بزرگ‌تر می‌شود: کارشناسی **+۲۲٫۷٪**، سرپرستی **+۹۴٫۴٪** و مدیریتی **+۱۰۴٫۵٪**. این الگو نشان می‌دهد هرچه به سطوح بالاتر می‌رویم، آگهی‌های زن‌محور کم‌تر و حقوق پیشنهادی‌شان پایین‌تر می‌شود؛ همان «سقف شیشه‌ای» در سطح آگهی.

**محدودیت جدی:** همه این اعداد ترجیح کارفرما در متن آگهی‌اند، نه دستمزد واقعی افراد. نتیجه را باید به‌عنوان «الگوی بازار آگهی» خواند، نه اثبات تبعیض دستمزدی.

**تفسیر جدول ۱۱.۳ و نمودار ۱۱.۲:** میانه حقوق آگهی‌های «فقط آقا» ۱۱ میلیون و «فقط خانم» ۹ میلیون تومان است؛ اختلاف **+۲۲٫۲٪**. آزمون t ولش (t=۴۲٫۵) و آزمون رتبه‌ای مان-ویتنی، هر دو این تفاوت را در سطح ۰٫۰۰۱ معنادار می‌دانند و اندازه اثر کوهن (d=۰٫۴۱) در محدوده «متوسط» قرار می‌گیرد.

**تفسیر درست:** این تفاوت، شکاف دستمزد افراد نیست. آگهی‌های زن‌محور بیشتر در دسته‌های کم‌حقوق (اداری، فروش، خدمات) متمرکزند و آگهی‌های مرد‌محور در نقش‌های فنی و صنعتی. بخش بزرگی از این ۲۲٪، اثر ترکیب شغلی است، نه قیمت‌گذاری یکسان برای نقش یکسان. برای جداکردن این دو، باید شکاف را درون دسته‌های شغلی دید؛ بخش ۱۱.۳ همین کار را می‌کند.

## ۱۱.۳. شکاف در دسته‌های شغلی و رده‌های سازمانی

</div>

```python
gap_rows = []
for category in ads.category_group.value_counts().head(10).index:
    subset = paid.loc[paid.category_group.eq(category)]
    male = subset.loc[subset.gender_pref.eq('فقط آقا'), 'salary_mid']
    female = subset.loc[subset.gender_pref.eq('فقط خانم'), 'salary_mid']
    if len(male) < MIN_N or len(female) < MIN_N:
        gap_rows.append({'دسته شغلی': category, 'تعداد فقط آقا': len(male), 'تعداد فقط خانم': len(female),
                         'میانه فقط آقا': np.nan, 'میانه فقط خانم': np.nan, 'شکاف': np.nan})
        continue
    gap_rows.append({'دسته شغلی': category, 'تعداد فقط آقا': len(male), 'تعداد فقط خانم': len(female),
                     'میانه فقط آقا': male.median(), 'میانه فقط خانم': female.median(),
                     'شکاف': (male.median() / female.median() - 1) * 100})
gap_table = pd.DataFrame(gap_rows).set_index('دسته شغلی').sort_values('شکاف', ascending=False)
show_table('11.4', 'شکاف میانه حقوق میان آگهی‌های مرد‌محور و زن‌محور در دسته‌های پرتکرار', gap_table,
           formats={'تعداد فقط آقا': 'int', 'تعداد فقط خانم': 'int', 'میانه فقط آقا': 'money',
                    'میانه فقط خانم': 'money', 'شکاف': 'pp'},
           sample='دسته‌هایی با دست‌کم ۳۰ آگهی در هر گروه', period=False, index_name='دسته شغلی')

seniority_gap = []
for level in SENIORITY_ORDER:
    subset = paid.loc[paid.seniority.eq(level)]
    male = subset.loc[subset.gender_pref.eq('فقط آقا'), 'salary_mid']
    female = subset.loc[subset.gender_pref.eq('فقط خانم'), 'salary_mid']
    seniority_gap.append({'رده سازمانی': level, 'تعداد فقط آقا': len(male), 'تعداد فقط خانم': len(female),
                          'میانه فقط آقا': male.median() if len(male) else np.nan,
                          'میانه فقط خانم': female.median() if len(female) else np.nan,
                          'شکاف': (male.median() / female.median() - 1) * 100
                          if len(male) >= MIN_N and len(female) >= MIN_N else np.nan})
seniority_gap_table = pd.DataFrame(seniority_gap).set_index('رده سازمانی')
show_table('11.5', 'شکاف حقوق بر پایه رده سازمانی', seniority_gap_table,
           formats={'تعداد فقط آقا': 'int', 'تعداد فقط خانم': 'int', 'میانه فقط آقا': 'money',
                    'میانه فقط خانم': 'money', 'شکاف': 'pp'},
           sample='آگهی‌های دارای حقوق اعلام‌شده', period=False, index_name='رده سازمانی')

show_metrics('جنسیت', [
    ('آگهی بدون ترجیح جنسیت', f'{ads.gender_pref.eq("بدون ترجیح").mean() * 100:.1f}٪'),
    ('آگهی با شرط صریح جنسیت', f'{ads.gender_pref.isin(["فقط آقا", "فقط خانم"]).mean() * 100:.1f}٪'),
    ('نسبت فقط آقا به فقط خانم', f'{ads.gender_pref.eq("فقط آقا").sum() / ads.gender_pref.eq("فقط خانم").sum():.2f}'),
    ('اختلاف میانه فقط آقا و فقط خانم', f'{median_gap:+.1f}٪'),
    ('اندازه اثر کوهن d', f'{cohens_d:+.2f}'),
])
```

<div dir="rtl" align="right">

# ۱۲. ارزش تحصیلات در برابر سابقه کار

## ۱۲.۱. بازده مدرک تحصیلی

</div>

```python
DEGREE_ORDER = ['اعلام نشده', 'کاردانی', 'کارشناسی', 'کارشناسی ارشد', 'دکترا']
degree_paid = (paid.groupby('max_degree')
               .agg(**{'تعداد آگهی': ('ad_id', 'size'), 'میانه حقوق': ('salary_mid', 'median'),
                       'میانه درآمد ساعتی': ('hourly_income', 'median'),
                       'میانه سابقه لازم': ('Jobpost_RequiredExperienceYears', 'median')})
               .reindex(DEGREE_ORDER))
baseline = degree_paid.loc['کارشناسی', 'میانه حقوق']
degree_paid['اختلاف با کارشناسی'] = (degree_paid['میانه حقوق'] / baseline - 1) * 100
degree_paid['سهم'] = degree_paid['تعداد آگهی'] / degree_paid['تعداد آگهی'].sum() * 100
show_table('12.1', 'حقوق و درآمد ساعتی بر پایه مقطع تحصیلی درخواستی', degree_paid,
           formats={'تعداد آگهی': 'int', 'سهم': 'pct', 'میانه حقوق': 'money',
                    'میانه درآمد ساعتی': 'hourly', 'میانه سابقه لازم': 'hours',
                    'اختلاف با کارشناسی': 'pp'},
           sample=f'{len(paid):,} آگهی با حقوق اعلام‌شده', index_name='مقطع درخواستی')

low_experience = paid.loc[paid.Jobpost_RequiredExperienceYears.le(1)]
degree_low_exp = (low_experience.groupby('max_degree')
                  .agg(**{'تعداد آگهی': ('ad_id', 'size'), 'میانه حقوق': ('salary_mid', 'median')})
                  .reindex(DEGREE_ORDER))
degree_low_exp['اختلاف با کارشناسی'] = (degree_low_exp['میانه حقوق'] /
                                        degree_low_exp.loc['کارشناسی', 'میانه حقوق'] - 1) * 100
show_table('12.2', 'بازده مدرک در آگهی‌های کم‌سابقه (صفر یا یک سال)', degree_low_exp,
           formats={'تعداد آگهی': 'int', 'میانه حقوق': 'money', 'اختلاف با کارشناسی': 'pp'},
           sample=f'{len(low_experience):,} آگهی با سابقه حداکثر یک سال', index_name='مقطع درخواستی')
```

**تفسیر جدول‌های ۱۲.۱ و ۱۲.۲:** در آگهی‌های دارای حقوق، ۶۱٫۶٪ مقطع تحصیلی را اعلام نکرده‌اند. در میان بقیه، کارشناسی با ۳۰٫۹٪ غالب است و ارشد و دکترا روی‌هم فقط ۳٫۱٪. میانه حقوق با مدرک بالاتر می‌رود: ۱۱ برای کاردانی و اعلام‌نشده، ۱۳٫۵ برای کارشناسی، ۱۷٫۵ برای ارشد و دکترا؛ یعنی ارشد و بالاتر حدود **+۳۰٪** بالاتر از کارشناسی.

نکته مهم در جدول ۱۲.۲ است: در آگهی‌های کم‌سابقه (صفر یا یک سال)، میانه کارشناسی ۹ و ارشد ۱۳٫۵ میلیون است؛ یعنی **+۵۰٪**. بازده مدرک در ورود به بازار بیشتر از میانه کل است، چون بخشی از آگهی‌های ارشد برای نقش‌های تخصصی‌ترند که سابقه کم هم قبول می‌کنند.

## ۱۲.۲. مدل رگرسیون برای مقایسه مستقیم مدرک و سابقه

</div>

```python
import statsmodels.formula.api as smf

model_data = paid.dropna(subset=['salary_mid', 'max_degree', 'Jobpost_RequiredExperienceYears']).copy()
model_data['experience'] = model_data.Jobpost_RequiredExperienceYears.clip(upper=20)
model_data['log_salary'] = np.log(model_data.salary_mid)
model_data['category_model'] = model_data.category_group
model_data['degree_model'] = pd.Categorical(model_data.max_degree,
                                            categories=['کارشناسی', 'اعلام نشده', 'کاردانی', 'کارشناسی ارشد', 'دکترا'])
earnings_model = smf.ols(
    'log_salary ~ experience + I(experience ** 2) + C(degree_model) + C(category_model) + C(province_group) '
    '+ company_size_rank', data=model_data).fit()

degree_terms = [term for term in earnings_model.params.index if term.startswith('C(degree_model)')]
experience_terms = [term for term in earnings_model.params.index if term.startswith(('experience', 'I(experience'))]
selected = degree_terms + experience_terms
coefficient_table = pd.DataFrame({
    'ضریب': earnings_model.params[selected],
    'خطای معیار': earnings_model.bse[selected],
    'p-value': earnings_model.pvalues[selected],
})
coefficient_table['اثر تقریبی بر حقوق'] = (np.exp(coefficient_table['ضریب']) - 1) * 100
coefficient_table.index = [index.replace('C(degree_model)[T.', '').replace(']', '')
                           .replace('experience', 'هر سال سابقه').replace('I(experience ** 2)', 'سابقه به توان دو')
                           for index in coefficient_table.index]
show_table('12.3', 'اثر مدرک و سابقه بر لگاریتم حقوق (مدل رگرسیون خطی)', coefficient_table,
           formats={'ضریب': 'score', 'خطای معیار': 'score', 'p-value': 'pvalue',
                    'اثر تقریبی بر حقوق': 'pp'},
           sample=f'{len(model_data):,} آگهی با حقوق اعلام‌شده',
           note='گروه پایه: مدرک کارشناسی، تهران/سایر، دسته‌های شغلی و اندازه شرکت کنترل‌شده است.')

degree_step = np.exp(earnings_model.params[[term for term in degree_terms
                                            if 'کارشناسی ارشد' in term][0]]) - 1
year_step = np.exp(earnings_model.params['experience']) - 1
comparison_step = pd.DataFrame({'مقدار': [f'{degree_step * 100:,.1f}٪', f'{year_step * 100:,.1f}٪',
                                          f'{degree_step / year_step:,.1f}']},
                               index=['یک پله مدرک (کارشناسی → ارشد)', 'یک سال سابقه بیشتر',
                                      'معادل‌سازی: هر پله مدرک چند سال سابقه'])
show_table('12.4', 'مقایسه ارزش یک پله مدرک با یک سال سابقه', comparison_step,
           sample=f'{len(model_data):,} آگهی', period=False, index_name='مقایسه')
```

**تفسیر جدول‌های ۱۲.۳ و ۱۲.۴:** در مدل رگرسیونی با کنترل دسته شغلی، استان و اندازه شرکت، هر سال سابقه **+۱۱٫۱٪** و هر پله مدرک (کارشناسی → ارشد) **+۱۴٫۹٪** به حقوق اضافه می‌کند. جمله مربع سابقه تقریباً صفر است (اثر −۰٫۱٪)؛ بنابراین در این داده، منحنی بازده سابقه در محدوده ۰ تا ۲۰ سال تقریباً خطی است و «بازده نزولی» شدید که در بخش ۸ دیده شد، بیشتر نتیجه تغییر ترکیب نقش‌ها در سطوح سابقه بالا است، نه خود سابقه.

**نتیجه کاربردی:** یک پله مدرک تقریباً معادل **۱٫۳ سال سابقه** ارزش دارد. اگر کارفرما بین «کارشناسی با ۳ سال سابقه» و «ارشد با ۱٫۷ سال سابقه» مخیر باشد، از نظر دستمزد بازار تقریباً هم‌ارزند.

## ۱۲.۳. کجا تجربه جای مدرک را می‌گیرد؟

</div>

```python
industry_degree_rows = []
for industry in ads.industry_group.value_counts().head(8).index:
    subset = paid.loc[paid.industry_group.eq(industry)]
    bachelor = subset.loc[subset.max_degree.eq('کارشناسی'), 'salary_mid']
    higher = subset.loc[subset.max_degree.isin(['کارشناسی ارشد', 'دکترا']), 'salary_mid']
    junior = subset.loc[subset.Jobpost_RequiredExperienceYears.le(1), 'salary_mid']
    senior = subset.loc[subset.Jobpost_RequiredExperienceYears.between(3, 5), 'salary_mid']
    industry_degree_rows.append({
        'صنعت': industry, 'تعداد': len(subset),
        'میانه کارشناسی': bachelor.median() if len(bachelor) >= MIN_N else np.nan,
        'میانه ارشد و بالاتر': higher.median() if len(higher) >= MIN_N else np.nan,
        'بازده مدرک': (higher.median() / bachelor.median() - 1) * 100
        if len(bachelor) >= MIN_N and len(higher) >= MIN_N else np.nan,
        'میانه کم‌سابقه': junior.median() if len(junior) >= MIN_N else np.nan,
        'میانه ۳ تا ۵ سال': senior.median() if len(senior) >= MIN_N else np.nan,
        'بازده سابقه': (senior.median() / junior.median() - 1) * 100
        if len(junior) >= MIN_N and len(senior) >= MIN_N else np.nan})
industry_degree = pd.DataFrame(industry_degree_rows).set_index('صنعت').sort_values('تعداد', ascending=False)
industry_degree['نسبت بازده سابقه به مدرک'] = industry_degree['بازده سابقه'] / industry_degree['بازده مدرک']
show_table('12.5', 'بازده مدرک و سابقه در صنایع پرتکرار', industry_degree,
           formats={'تعداد': 'int', 'میانه کارشناسی': 'money', 'میانه ارشد و بالاتر': 'money',
                    'بازده مدرک': 'pp', 'میانه کم‌سابقه': 'money', 'میانه ۳ تا ۵ سال': 'money',
                    'بازده سابقه': 'pp', 'نسبت بازده سابقه به مدرک': 'score'},
           sample=f'{len(paid):,} آگهی با حقوق اعلام‌شده', index_name='صنعت',
           note='نسبت بزرگ‌تر از ۱ یعنی بازده سابقه در آن صنعت از بازده مدرک بیشتر است.')

fig, axes = plt.subplots(1, 2, figsize=(14, 5), gridspec_kw={'width_ratios': [1, 1.15]})
ax = axes[0]
degrees_plot = degree_paid.dropna(subset=['میانه حقوق'])
bars = ax.bar(range(len(degrees_plot)), degrees_plot['میانه حقوق'],
              color=[GRAY, GOLD, TEAL, BLUE, PURPLE][:len(degrees_plot)], width=0.6)
ax.set_xticks(range(len(degrees_plot)))
ax.set_xticklabels([fa(index) for index in degrees_plot.index], fontsize=8.5)
ax.set_ylabel(fa('میانه حقوق (میلیون تومان)'))
for bar, value in zip(bars, degrees_plot['میانه حقوق']):
    ax.text(bar.get_x() + bar.get_width() / 2, value, f'{value:.1f}', ha='center', va='bottom', fontsize=8.5)
heading(ax, '۱۲.۱', 'میانه حقوق بر پایه مقطع تحصیلی')
ax = axes[1]
exp_bucket_model = pd.cut(paid.Jobpost_RequiredExperienceYears, bins=[-1, 1, 3, 5, 20],
                          labels=['۰ تا ۱', '۲ تا ۳', '۴ تا ۵', 'بیش از ۵'])
degree_by_exp = (paid.assign(exp_bucket=exp_bucket_model, degree_model=paid.max_degree)
                 .loc[lambda frame: frame.degree_model.isin(['کاردانی', 'کارشناسی', 'کارشناسی ارشد', 'دکترا'])]
                 .pivot_table(index='degree_model', columns='exp_bucket', values='salary_mid',
                              aggfunc='median', observed=True)
                 .reindex(['کاردانی', 'کارشناسی', 'کارشناسی ارشد', 'دکترا']))
sns.heatmap(degree_by_exp, ax=ax, cmap='crest', annot=True, fmt='.1f', annot_kws={'size': 8},
            cbar_kws={'label': 'میلیون تومان'}, linewidths=0.6, linecolor='white')
ax.set_yticklabels([fa(index) for index in degree_by_exp.index], rotation=0, fontsize=8.5)
ax.set_xticklabels([fa(str(column)) for column in degree_by_exp.columns], fontsize=8.5)
ax.set_xlabel(fa('سابقه لازم (سال)'))
ax.set_ylabel('')
heading(ax, '۱۲.۲', 'میانه حقوق: مدرک در برابر سابقه')
complete('12.1-12.2')

show_metrics('تحصیلات', [
    ('بازده یک پله مدرک', f'{degree_step * 100:,.1f}٪'),
    ('بازده یک سال سابقه', f'{year_step * 100:,.1f}٪'),
    ('هر پله مدرک معادل چند سال سابقه', f'{degree_step / year_step:,.1f} سال'),
    ('R² مدل تحصیلات و سابقه', f'{earnings_model.rsquared:.3f}'),
])
```

**تفسیر جدول ۱۲.۵ و نمودارهای ۱۲.۱ و ۱۲.۲:** بازده مدرک و سابقه یکسان نیست و به صنعت بستگی دارد:

- در **فناوری اطلاعات** بازده مدرک و سابقه هر دو حدود +۳۰٪ است (نسبت ۱٫۰)؛ یعنی مدرک و تجربه هم‌ارزند.
- در **تولید و تجارت** بازده سابقه (+۵۰٪) از بازده مدرک (+۲۹٫۶٪) بیشتر است (نسبت ۱٫۷).
- در **اینترنت و تجارت الکترونیک** بازده سابقه به **+۹۴٪** می‌رسد و مدرک +۲۹٫۶٪؛ یعنی تجربه عملاً جای مدرک را می‌گیرد (نسبت ۳٫۲).
- در **خدمات مهندسی و خودرو** بازده مدرک صفر است و بازده سابقه مثبت؛ در این صنایع مدرک تحصیلی تقریباً قیمتی ندارد.

الگوی کلی: در صنایع دانش‌محور و بالغ، مدرک و سابقه هم‌ارزند؛ در صنایع در حال رشد سریع (اینترنت) و صنایع سنتی (تولید، خودرو، مهندسی)، تجربه ارزش بیشتری دارد.

<div dir="rtl" align="right">

# ۱۳. نقطه اوج درآمدی بر اساس سن

## ۱۳.۱. میانه حقوق در بازه‌های سنی

</div>

```python
age_bins = pd.cut(paid.age_mid, bins=[17, 20, 23, 26, 29, 32, 35, 38, 41, 60])
age_profile = (paid.groupby(age_bins, observed=True)
               .agg(**{'تعداد آگهی': ('ad_id', 'size'), 'میانه حقوق': ('salary_mid', 'median'),
                       'میانه درآمد ساعتی': ('hourly_income', 'median'),
                       'میانه سابقه لازم': ('Jobpost_RequiredExperienceYears', 'median')}))
age_profile['سهم'] = age_profile['تعداد آگهی'] / age_profile['تعداد آگهی'].sum() * 100
age_profile.index = [f'{int(interval.left) + 1} تا {int(interval.right)}' for interval in age_profile.index]
show_table('13.1', 'حقوق و درآمد ساعتی بر پایه بازه سنی درخواستی', age_profile,
           formats={'تعداد آگهی': 'int', 'سهم': 'pct', 'میانه حقوق': 'money',
                    'میانه درآمد ساعتی': 'hourly', 'میانه سابقه لازم': 'hours'},
           sample=f'{int(paid.age_mid.notna().sum()):,} آگهی با حقوق و سن', index_name='بازه سنی (سال)')

peak_band = age_profile['میانه حقوق'].idxmax()
fig, ax = plt.subplots(figsize=(11.5, 4.8))
positions = np.arange(len(age_profile))
ax.bar(positions, age_profile['میانه حقوق'], color=TEAL, width=0.62, label=fa('میانه حقوق ماهانه'))
ax.set_xticks(positions)
ax.set_xticklabels([fa(index) for index in age_profile.index], rotation=30, ha='right', fontsize=8.5)
ax.set_ylabel(fa('میانه حقوق (میلیون تومان)'))
ax2 = ax.twinx()
ax2.plot(positions, age_profile['میانه درآمد ساعتی'], color=GOLD, marker='o', linewidth=1.8,
         label=fa('میانه درآمد ساعتی'))
ax2.set_ylabel(fa('درآمد ساعتی (هزار تومان)'), color=GOLD)
ax2.grid(False)
peak_position = int(age_profile.index.get_loc(peak_band))
ax.annotate(fa(f'اوج حقوق: {peak_band} سال'), xy=(peak_position, age_profile['میانه حقوق'].max()),
            xytext=(peak_position - 2.6, age_profile['میانه حقوق'].max() + 1.6), fontsize=8.5,
            arrowprops=dict(arrowstyle='->', color=DARK))
lines = ax.get_legend_handles_labels()[0] + ax2.get_legend_handles_labels()[0]
labels = ax.get_legend_handles_labels()[1] + ax2.get_legend_handles_labels()[1]
ax.legend(lines, labels, fontsize=8.5, loc='upper left')
heading(ax, '۱۳.۱', 'میانه حقوق و درآمد ساعتی بر پایه سن درخواستی')
complete('13.1')

show_metrics('اوج درآمدی', [
    ('بازه سنی اوج حقوق', peak_band),
    ('میانه حقوق در اوج', f'{age_profile["میانه حقوق"].max():,.1f} میلیون'),
    ('میانه حقوق زیر ۲۱ سال', f'{age_profile["میانه حقوق"].iloc[0]:,.1f} میلیون'),
    ('میانه سن درخواستی', f'{paid.age_mid.median():.1f} سال'),
])
```

**تفسیر جدول ۱۳.۱ و نمودار ۱۳.۱:** میانه حقوق با سن درخواستی صعودی است اما کم‌شیب: از ۹ میلیون در بازه ۲۴ تا ۲۶ سال به ۱۱ میلیون در ۲۷ تا ۳۵ سال، ۱۳٫۵ در ۳۶ تا ۴۱ سال و ۱۷٫۵ میلیون در ۴۲ سال و بالاتر. «اوج» در بازه ۴۲ تا ۶۰ سال است، اما این بازه فقط ۱٬۸۲۰ آگهی دارد و عدد ۱۷٫۵ میلیون بیشتر بازتاب چند نقش مدیریتی است تا یک قاعده عمومی.

دو نکته غیرشهودی: بازه ۱۸ تا ۲۰ سال میانه ۱۳٫۵ میلیون دارد (بالاتر از ۲۱ تا ۲۶ سال با ۹ میلیون)؛ این آگهی‌ها عمدتاً نقش‌های خاص نیروی جوان (مثلاً فروش یا کارآموزی تخصصی) هستند و نمونه‌شان کوچک است. همچنین درآمد ساعتی در بازه‌های سنی بالا افزایش می‌یابد، چون ساعت کاری این نقش‌ها کمتر است.

**هشدار:** بازه سنی، ترجیح کارفرماست. استفاده از این جدول برای «سقف سنی» در آگهی، هم اخلاقاً مسئله‌دار است و هم با کمبود نیروی باتجربه به ضرر کارفرما تمام می‌شود.

## ۱۳.۲. تفاوت نقطه اوج در صنایع مختلف

</div>

```python
industry_age_rows = []
for industry in ads.industry_group.value_counts().head(6).index:
    subset = paid.loc[paid.industry_group.eq(industry)]
    grouped = subset.groupby(pd.cut(subset.age_mid, bins=[17, 24, 29, 34, 39, 60],
                                    labels=['۱۸ تا ۲۴', '۲۵ تا ۲۹', '۳۰ تا ۳۴', '۳۵ تا ۳۹', '۴۰ و بالاتر']),
                            observed=True).agg(**{'تعداد': ('ad_id', 'size'),
                                                  'میانه حقوق': ('salary_mid', 'median')})
    grouped = grouped.loc[grouped['تعداد'].ge(MIN_N)]
    if grouped.empty:
        continue
    industry_age_rows.append({'صنعت': industry, 'تعداد آگهی': int(grouped['تعداد'].sum()),
                              'بازه اوج': grouped['میانه حقوق'].idxmax(),
                              'میانه اوج': grouped['میانه حقوق'].max(),
                              'میانه کم‌سن‌ترین بازه': grouped['میانه حقوق'].iloc[0]})
industry_age = pd.DataFrame(industry_age_rows).set_index('صنعت')
industry_age['اختلاف اوج با کم‌سن‌ترین'] = ((industry_age['میانه اوج'] /
                                            industry_age['میانه کم‌سن‌ترین بازه'] - 1) * 100).round(1)
show_table('13.2', 'بازه سنی اوج حقوق در صنایع پرتکرار', industry_age,
           formats={'تعداد آگهی': 'int', 'میانه اوج': 'money', 'میانه کم‌سن‌ترین بازه': 'money',
                    'اختلاف اوج با کم‌سن‌ترین': 'pp'},
           sample=f'{int(paid.age_mid.notna().sum()):,} آگهی با حقوق و سن', index_name='صنعت',
           note='بازه‌های با کمتر از ۳۰ آگهی در تعیین اوج نادیده گرفته شده‌اند.')

fig, ax = plt.subplots(figsize=(12.5, 5.4))
age_bands = ['۱۸ تا ۲۴', '۲۵ تا ۲۹', '۳۰ تا ۳۴', '۳۵ تا ۳۹', '۴۰ و بالاتر']
for industry, color in zip(industry_age.index, PALETTE):
    subset = paid.loc[paid.industry_group.eq(industry)]
    grouped = subset.groupby(pd.cut(subset.age_mid, bins=[17, 24, 29, 34, 39, 60], labels=age_bands),
                             observed=True).agg(**{'تعداد': ('ad_id', 'size'),
                                                   'میانه حقوق': ('salary_mid', 'median')})
    grouped = grouped.loc[grouped['تعداد'].ge(MIN_N)]
    if grouped.empty:
        continue
    index_relative = (grouped['میانه حقوق'] / grouped['میانه حقوق'].iloc[0]) * 100
    ax.plot(range(len(grouped)), index_relative, marker='o', color=color, linewidth=1.8,
            label=fa(industry))
ax.set_xticks(range(len(age_bands)))
ax.set_xticklabels([fa(band) for band in age_bands], fontsize=8.5)
ax.set_ylabel(fa('میانه حقوق نسبت به کم‌سن‌ترین بازه هر صنعت (٪)'))
ax.axhline(100, color=GRAY, linewidth=0.9, linestyle='--')
ax.legend(fontsize=8)
heading(ax, '۱۳.۲', 'منحنی سنی حقوق در صنایع پرتکرار (شاخص‌شده)')
complete('13.2')
```

**تفسیر جدول ۱۳.۲ و نمودار ۱۳.۲:** نقطه اوج درآمدی در صنایع متفاوت است. در فناوری اطلاعات و کالاهای مصرفی، اوج در بازه **۴۰ سال و بالاتر** است (+۳۰٪ نسبت به کم‌سن‌ترین بازه)؛ در تولید، تجارت و «سایر صنایع» اوج در **۳۵ تا ۳۹ سال** است؛ و در اینترنت/تجارت الکترونیک، اوج در همان بازه **۱۸ تا ۲۴ سال** است و منحنی بعد از آن بالا نمی‌رود.

تفسیر: در صنایع دانش‌محور، تجربه و سن تا دهه چهارم زندگی ارزش افزوده می‌سازد؛ در اینترنت و کسب‌وکارهای دیجیتال، بازار برای نیروی جوان و کم‌هزینه‌تر قیمت بالاتری پیشنهاد می‌کند و مسیر رشد سنی مسطح است. برای مدیر، این یعنی «ساختار سنی تیم» باید با مدل کسب‌وکار هم‌خوان باشد، نه با یک قاعده یکسان.

# ۱۴. تحلیل جغرافیایی و تمرکز استانی

## ۱۴.۱. توزیع آگهی‌ها در استان‌ها

</div>

```python
province_stats = (ads.groupby('Jobpost_ProvinceFa')
                  .agg(**{'تعداد آگهی': ('ad_id', 'size'),
                          'میانه حقوق': ('salary_mid', 'median'),
                          'میانه درآمد ساعتی': ('hourly_income', 'median'),
                          'میانه سابقه لازم': ('Jobpost_RequiredExperienceYears', 'median'),
                          'نرخ افشای حقوق': ('salary_disclosed', lambda s: s.mean() * 100),
                          'نرخ دورکاری': ('Jobpost_IsRemote', lambda s: s.mean() * 100)}))
province_stats['سهم'] = province_stats['تعداد آگهی'] / len(ads) * 100
province_stats = province_stats.sort_values('تعداد آگهی', ascending=False)
top_provinces = province_stats.head(15)
tehran_share = province_stats.loc['تهران', 'سهم']
show_table('14.1', 'پانزده استان نخست بر پایه تعداد آگهی', top_provinces,
           formats={'تعداد آگهی': 'int', 'سهم': 'pct', 'میانه حقوق': 'money',
                    'میانه درآمد ساعتی': 'hourly', 'میانه سابقه لازم': 'hours',
                    'نرخ افشای حقوق': 'pct', 'نرخ دورکاری': 'pct'},
           sample=f'{len(ads):,} آگهی در ۳۲ استان', index_name='استان')

fig, axes = plt.subplots(1, 2, figsize=(14, 5.4), gridspec_kw={'width_ratios': [1.1, 1]})
ax = axes[0]
plot_provinces = top_provinces.head(12).iloc[::-1]
ax.barh(range(len(plot_provinces)), plot_provinces['سهم'], color=TEAL, height=0.7)
ax.set_yticks(range(len(plot_provinces)))
ax.set_yticklabels([fa(name) for name in plot_provinces.index], fontsize=8.5)
ax.set_xlabel(fa('سهم از کل آگهی‌ها (٪)'))
for position, value in enumerate(plot_provinces['سهم']):
    ax.text(value + 0.4, position, f'{value:.1f}٪', va='center', fontsize=7.5)
heading(ax, '۱۴.۱', 'سهم استان‌ها از آگهی‌ها')
ax = axes[1]
salary_plot = top_provinces.dropna(subset=['میانه حقوق']).iloc[::-1]
ax.barh(range(len(salary_plot)), salary_plot['میانه حقوق'], color=GOLD, height=0.7)
ax.axvline(paid.salary_mid.median(), color=RED, linewidth=1.4, linestyle='--',
           label=fa(f'میانه کل: {paid.salary_mid.median():.1f}'))
ax.set_yticks(range(len(salary_plot)))
ax.set_yticklabels([fa(name) for name in salary_plot.index], fontsize=8.5)
ax.set_xlabel(fa('میانه حقوق (میلیون تومان)'))
ax.legend(fontsize=8)
heading(ax, '۱۴.۲', 'میانه حقوق در استان‌های پرتکرار')
complete('14.1-14.2')

show_metrics('تمرکز جغرافیایی', [
    ('سهم تهران', f'{tehran_share:.1f}٪'),
    ('سهم پنج استان نخست', f'{province_stats["سهم"].head(5).sum():.1f}٪'),
    ('استان با بیشترین حقوق', f'{province_stats.loc[province_stats["میانه حقوق"].idxmax(), "میانه حقوق"]:,.1f} میلیون — {province_stats["میانه حقوق"].idxmax()}'),
    ('نرخ دورکاری تهران', f'{province_stats.loc["تهران", "نرخ دورکاری"]:.1f}٪'),
])
```

**تفسیر جدول ۱۴.۱ و نمودارهای ۱۴.۱ و ۱۴.۲:** تهران با ۶۴٫۶٪ سهم، بازار کار ایران را تک‌مرکزی کرده است؛ پنج استان نخست روی‌هم ۸۰٫۶٪ آگهی‌ها را دارند. اما **میانه حقوق همه استان‌های پرتکرار ۱۱ میلیون تومان است** و شکاف دستمزدی بین تهران و سایر استان‌ها در سطح میانه وجود ندارد. این یافته مهم است: تمرکز جغرافیایی در «تعداد فرصت» است، نه در «قیمت».

دورکاری تصویر را برعکس می‌کند: نرخ دورکاری در تهران فقط ۳٫۱٪ است، در حالی که مازندران ۱۵٫۶٪، فارس ۱۹٫۴٪، گیلان ۱۸٫۳٪ و آذربایجان شرقی ۲۰٫۷٪ است. یعنی آگهی‌های دورکاری بیشتر بیرون تهران متمرکزند و امکان کار از راه دور، برای نیروی غیرتهرانی نقش «جبران‌کننده فاصله» را بازی می‌کند.

> آگهی‌های با استان «خارج کشور» (حدود ۱٬۵۰۰ آگهی با میانه ۲۲٫۵ میلیون) در رتبه‌بندی حقوق استان‌ها کنار گذاشته شده‌اند، چون موقعیت برون‌مرزی‌اند و با بازار داخلی قابل مقایسه نیستند.

## ۱۴.۲. شکاف تهران و سایر استان‌ها در طول زمان

</div>

```python
province_month = (ads.assign(is_tehran=ads.Jobpost_ProvinceFa.eq('تهران'))
                  .groupby(['release_month', 'is_tehran'])
                  .agg(**{'تعداد آگهی': ('ad_id', 'size'), 'میانه حقوق': ('salary_mid', 'median')})
                  .reset_index())
province_month['ماه شمسی'] = [jmonth(month)[2] for month in province_month.release_month]
count_pivot_province = province_month.pivot(index='release_month', columns='is_tehran', values='تعداد آگهی').fillna(0)
count_pivot_province.columns = ['سایر استان‌ها', 'تهران']
count_pivot_province['سهم تهران'] = count_pivot_province['تهران'] / count_pivot_province.sum(axis=1) * 100
salary_pivot_province = province_month.pivot(index='release_month', columns='is_tehran', values='میانه حقوق')
salary_pivot_province['نسبت تهران به سایر'] = (salary_pivot_province[True] / salary_pivot_province[False] * 100).round(1)
tehran_trend = count_pivot_province[['سهم تهران']].join(salary_pivot_province[['نسبت تهران به سایر']])
tehran_trend['ماه شمسی'] = [jmonth(month)[2] for month in tehran_trend.index]
show_table('14.2', 'سهم تهران از آگهی‌ها و نسبت حقوق آن به سایر استان‌ها', tehran_trend,
           formats={'سهم تهران': 'pct', 'نسبت تهران به سایر': 'pct'},
           sample=f'{len(ads):,} آگهی', index_name='ماه میلادی',
           note='نسبت ۱۰۰٪ یعنی میانه حقوق تهران برابر میانه سایر استان‌ها است.')

fig, axes = plt.subplots(1, 2, figsize=(14, 4.8))
ax = axes[0]
x = np.arange(len(tehran_trend))
ax.plot(x, tehran_trend['سهم تهران'], color=TEAL, marker='o', markersize=3, linewidth=1.8)
ax.axhline(tehran_share, color=RED, linestyle='--', linewidth=1.2, label=fa(f'میانگین کل {tehran_share:.1f}٪'))
ax.set_xticks(x[::2])
ax.set_xticklabels([fa(label) for label in tehran_trend['ماه شمسی'].iloc[::2]], rotation=60, ha='right', fontsize=7.5)
ax.set_ylabel(fa('سهم تهران از آگهی‌ها (٪)'))
ax.legend(fontsize=8)
heading(ax, '۱۴.۳', 'روند سهم تهران')
ax = axes[1]
ax.plot(x, tehran_trend['نسبت تهران به سایر'], color=GOLD, marker='o', markersize=3, linewidth=1.8)
ax.axhline(100, color=GRAY, linestyle='--', linewidth=1.2)
ax.set_xticks(x[::2])
ax.set_xticklabels([fa(label) for label in tehran_trend['ماه شمسی'].iloc[::2]], rotation=60, ha='right', fontsize=7.5)
ax.set_ylabel(fa('نسبت میانه حقوق تهران به سایر (٪)'))
heading(ax, '۱۴.۴', 'روند نسبت حقوق تهران به سایر استان‌ها')
complete('14.3-14.4')
```

**تفسیر جدول ۱۴.۲ و نمودارهای ۱۴.۳ و ۱۴.۴:** سهم تهران در طول ۲۳ ماه از ۷۲٪ به ۶۲٪ کاهش یافته است؛ حدود ۹ واحد درصد. این کاهش تدریجی است، نه یک شکست ناگهانی، و نشان می‌دهد رشد آگهی‌ها در استان‌های دیگر کمی سریع‌تر از تهران بوده است.

نسبت میانه حقوق تهران به سایر استان‌ها تقریباً همیشه ۱۰۰٪ است و در چند ماه به ۱۲۲٪ می‌رسد؛ یعنی حتی در ماه‌های پرتقاضا، شکاف دستمزدی تهران و سایر استان‌ها در سطح میانه کوچک است. نتیجه: تهران بازار «بزرگ‌تر» است، نه لزوماً «گران‌تر».

## ۱۴.۳. استان‌های در حال رشد و در حال عقب‌نشینی

</div>

```python
province_year = (ads.groupby(['Jobpost_ProvinceFa', 'jyear'])
                 .agg(**{'تعداد آگهی': ('ad_id', 'size')})
                 .reset_index())
province_year['سهم'] = province_year['تعداد آگهی'] / province_year.groupby('jyear')['تعداد آگهی'].transform('sum') * 100
province_share = province_year.pivot(index='Jobpost_ProvinceFa', columns='jyear', values='سهم').fillna(0)
province_count = province_year.pivot(index='Jobpost_ProvinceFa', columns='jyear', values='تعداد آگهی').fillna(0)
province_share_change = province_share.copy()
province_share_change.columns = ['سهم ۱۴۰۱', 'سهم ۱۴۰۲']
province_share_change['تغییر واحد درصدی'] = province_share_change['سهم ۱۴۰۲'] - province_share_change['سهم ۱۴۰۱']
province_share_change['رشد تعداد'] = (province_count[1402] / province_count[1401].replace(0, np.nan) - 1) * 100
province_share_change['تعداد ۱۴۰۲'] = province_count[1402]
province_share_change = province_share_change.sort_values('تعداد ۱۴۰۲', ascending=False).head(14)
show_table('14.3', 'تغییر سهم استان‌ها بین دو سال', province_share_change,
           formats={'سهم ۱۴۰۱': 'pct', 'سهم ۱۴۰۲': 'pct', 'تغییر واحد درصدی': 'pp',
                    'رشد تعداد': 'pp', 'تعداد ۱۴۰۲': 'int'},
           sample=f'{len(ads):,} آگهی', period=False, index_name='استان',
           note='رشد تعداد شامل اثر رشد کل بازار است؛ تغییر سهم اثر نسبی را نشان می‌دهد.')

fig, ax = plt.subplots(figsize=(11.5, 5.6))
plot_data = province_share_change.sort_values('تغییر واحد درصدی')
colors = [RED if value < 0 else TEAL for value in plot_data['تغییر واحد درصدی']]
ax.barh(range(len(plot_data)), plot_data['تغییر واحد درصدی'], color=colors, height=0.7)
ax.set_yticks(range(len(plot_data)))
ax.set_yticklabels([fa(name) for name in plot_data.index], fontsize=8.5)
ax.axvline(0, color=DARK, linewidth=0.9)
ax.set_xlabel(fa('تغییر سهم ۱۴۰۲ نسبت به ۱۴۰۱ (واحد درصدی)'))
for position, value in enumerate(plot_data['تغییر واحد درصدی']):
    offset = 0.04 if value >= 0 else -0.04
    ax.text(value + offset, position, f'{value:+.2f}', va='center',
            ha='left' if value >= 0 else 'right', fontsize=7.5)
heading(ax, '۱۴.۵', 'تغییر سهم استانی بین دو سال', period='۱۴۰۱ در برابر ۱۴۰۲')
complete('14.5')
```

**تفسیر جدول ۱۴.۳ و نمودار ۱۴.۵:** در سطح سهم، تهران **−۲٫۷ واحد درصد** از بازار را از دست داده و فارس (+۰٫۶)، خراسان رضوی (+۰٫۵)، البرز (+۰٫۳) و یزد (+۰٫۳) بزرگ‌ترین برندگان‌اند. در سطح رشد تعداد، فارس (+۱۱۷٪)، یزد (+۱۰۵٪)، خوزستان (+۸۶٪) و خراسان رضوی (+۸۲٪) سریع‌ترین رشد را داشته‌اند؛ تهران با +۵۱٪ از میانگین بازار عقب‌تر بوده است.

این الگو با «پراکندگی تدریجی بازار کار» سازگار است: استان‌های دارای دانشگاه، صنعت و نیروی متخصص، سهم بیشتری از آگهی‌ها می‌گیرند. برای کارفرمای غیرتهرانی، این یعنی رقابت برای جذب در حال افزایش است و باید زودتر و شفاف‌تر وارد بازار شود.

## ۱۴.۴. جهش‌های ناگهانی استانی

</div>

```python
province_monthly_counts = (ads.groupby(['Jobpost_ProvinceFa', 'release_month'])
                           .agg(**{'تعداد آگهی': ('ad_id', 'size')})
                           .reset_index())
baseline = province_monthly_counts.groupby('Jobpost_ProvinceFa')['تعداد آگهی'].agg(['mean', 'std'])
province_monthly_counts = province_monthly_counts.join(baseline, on='Jobpost_ProvinceFa')
province_monthly_counts['z'] = ((province_monthly_counts['تعداد آگهی'] -
                                 province_monthly_counts['mean']) / province_monthly_counts['std'])
province_monthly_counts['ماه شمسی'] = [jmonth(month)[2] for month in province_monthly_counts.release_month]
big_enough = province_monthly_counts.loc[province_monthly_counts['تعداد آگهی'].ge(MIN_CAT)]
spikes = (big_enough.sort_values('z', ascending=False)
          .head(10)[['Jobpost_ProvinceFa', 'ماه شمسی', 'تعداد آگهی', 'mean', 'z']]
          .rename(columns={'Jobpost_ProvinceFa': 'استان', 'تعداد آگهی': 'آگهی در ماه'}))
spikes['میانگین استان'] = spikes.pop('mean')
spikes['انحراف استاندارد نسبی'] = spikes.pop('z')
show_table('14.4', 'ده جهش ناگهانی در تعداد آگهی استانی', spikes.set_index('استان'),
           formats={'آگهی در ماه': 'int', 'میانگین استان': 'hourly', 'انحراف استاندارد نسبی': 'score'},
           sample=f'استان‌های با دست‌کم {MIN_CAT} آگهی در ماه', period=False, index_name='استان',
           note='انحراف استاندارد نسبی (z) بالای ۲ نشانه ماهی با فاصله معنادار از رفتار معمول آن استان است.')

top_spike = big_enough.sort_values('z', ascending=False).iloc[0]
spike_province, spike_month = top_spike.Jobpost_ProvinceFa, top_spike.release_month
spike_month_ads = ads.loc[ads.Jobpost_ProvinceFa.eq(spike_province) & ads.release_month.eq(spike_month)]
spike_province_ads = ads.loc[ads.Jobpost_ProvinceFa.eq(spike_province)]
spike_composition = pd.DataFrame({
    'در ماه جهش': spike_month_ads.category_group.value_counts(normalize=True) * 100,
    'کل استان': spike_province_ads.category_group.value_counts(normalize=True) * 100,
}).fillna(0)
spike_composition['اختلاف واحد درصدی'] = spike_composition['در ماه جهش'] - spike_composition['کل استان']
spike_composition = spike_composition.sort_values('در ماه جهش', ascending=False).head(6)
show_table('14.5', f'ترکیب دسته‌های شغلی در ماه جهش {spike_province} ({jmonth(spike_month)[2]})',
           spike_composition,
           formats={'در ماه جهش': 'pct', 'کل استان': 'pct', 'اختلاف واحد درصدی': 'pp'},
           sample=f'{len(spike_month_ads):,} آگهی در ماه جهش و {len(spike_province_ads):,} آگهی در کل استان',
           period=False, index_name='دسته شغلی',
           note='اختلاف مثبت یعنی آن دسته در ماه جهش سهم بیشتری از همیشه داشته است.')
```

**تفسیر جدول‌های ۱۴.۴ و ۱۴.۵:** بیشترین جهش‌های استانی در **دی ۱۴۰۲** رخ داده است: زنجان (۱۳۰ آگهی در برابر میانگین ۶۴، z=۲٫۶۷) و فارس (۶۷۹ در برابر ۲۹۸، z=۲٫۵۱)؛ پس از آن گلستان در شهریور، کرمان در آبان و چند استان در مهر. ترکیب دسته‌ها در ماه جهش زنجان نشان می‌دهد سهم «فروش کارشناسی و مدیریتی» و «کارگر ساده» و «برنامه‌نویسی» بالاتر از معمول استان بوده است.

**تفسیر محتاطانه:** جهش یک ماه، معمولاً نتیجه انتشار همزمان چند آگهی از یک یا چند کارفرمای بزرگ است، نه تغییر ساختاری بازار. این جدول ابزار «پایش» است: برای استان خودتان، ماه‌های با z بالای ۲ را بررسی کنید تا بفهمید رقیب جدیدی وارد شده یا یک طرح توسعه آغاز شده است.

# ۱۵. ایران در آینه بازار کار جهانی (۲۰۲۲–۲۰۲۴)

گزارش‌های جهانی این دوره سه الگوی روشن دارند: جهش استخدام پس از کرونا و سپس تعدیل نیرو در فناوری در ۲۰۲۳، تثبیت دورکاری ترکیبی، و رشد سریع تقاضا برای مهارت‌های داده و هوش مصنوعی. در این بخش، همان الگوها را با شاخص‌های جانشین در داده جاب‌ویژن می‌سنجیم. مقایسه **کیفی** است و هدف آن یافتن شباهت و واگرایی است، نه اثبات علیت.

## ۱۵.۱. شاخص‌های جانشین در داده ایران

</div>

```python
AI_SKILLS = {'python', 'tensorflow', 'pytorch', 'scikit-learn', 'keras', 'opencv', 'machine learning',
             'deep learning', 'artificial intelligence', 'data mining', 'nlp', 'chatgpt', 'llm'}
skill_lower = skills_long.skill.str.lower()
ai_skill_names = sorted(AI_SKILLS & set(skill_lower.unique()))
ai_ads = set(skills_long.loc[skill_lower.isin(AI_SKILLS), 'ad_id'])
data_skills = {'sql server', 'tableau', 'python', 'r'}
data_ads = set(skills_long.loc[skill_lower.isin(data_skills), 'ad_id'])
tech_industry = 'فناوری اطلاعات / نرم افزار و سخت افزار'

proxy = ads.assign(has_ai=ads.ad_id.isin(ai_ads), has_data_skill=ads.ad_id.isin(data_ads),
                   is_tech=ads.Jobpost_IndustryFa.eq(tech_industry),
                   is_it_category=ads.category_group.eq('توسعه نرم افزار و برنامه نویسی'))
proxy_table = (proxy.groupby('jyear')
               .agg(**{'سهم صنعت فناوری اطلاعات': ('is_tech', lambda s: s.mean() * 100),
                       'سهم دسته برنامه‌نویسی': ('is_it_category', lambda s: s.mean() * 100),
                       'سهم مهارت داده': ('has_data_skill', lambda s: s.mean() * 100),
                       'سهم مهارت هوش مصنوعی': ('has_ai', lambda s: s.mean() * 100),
                       'نرخ دورکاری': ('Jobpost_IsRemote', lambda s: s.mean() * 100)}))
proxy_table.loc['تغییر واحد درصدی'] = proxy_table.loc[1402] - proxy_table.loc[1401]
show_table('15.1', 'شاخص‌های جانشین روندهای جهانی در داده جاب‌ویژن (درصد)', proxy_table,
           formats={column: 'pct' for column in proxy_table.columns},
           sample=f'{len(ads):,} آگهی', period=False, index_name='سال',
           note='ردیف آخر تغییر واحد درصدی ۱۴۰۲ نسبت به ۱۴۰۱ است.')

tech_tracks = {'توسعه نرم افزار و برنامه نویسی': 'برنامه‌نویسی',
               'شبکه / DevOps / پشتیبانی سخت افزاری و نرم افزاری': 'شبکه و DevOps',
               'دیجیتال مارکتینگ و سئو': 'دیجیتال مارکتینگ'}
track_rows = []
for category_name, label in tech_tracks.items():
    subset = paid.loc[paid.Jobpost_MainJobCategory.eq(category_name)]
    track_rows.append({
        'مسیر فناوری': label, 'تعداد آگهی': len(subset),
        'میانه ۱۴۰۱': subset.loc[subset.jyear.eq(1401), 'salary_mid'].median(),
        'میانه ۱۴۰۲': subset.loc[subset.jyear.eq(1402), 'salary_mid'].median(),
        'نسبت به میانه کل ۱۴۰۲': subset.loc[subset.jyear.eq(1402), 'salary_mid'].median() /
                                  paid.loc[paid.jyear.eq(1402), 'salary_mid'].median() * 100,
    })
track_salary = pd.DataFrame(track_rows).set_index('مسیر فناوری')
show_table('15.2', 'حقوق مسیرهای فناوری در برابر میانه کل بازار', track_salary,
           formats={'تعداد آگهی': 'int', 'میانه ۱۴۰۱': 'money', 'میانه ۱۴۰۲': 'money',
                    'نسبت به میانه کل ۱۴۰۲': 'pct'},
           sample=f'{len(paid):,} آگهی با حقوق', period=False, index_name='مسیر فناوری',
           note='نسبت بالای ۱۰۰٪ یعنی آن مسیر از میانه کل بازار بیشتر پرداخت می‌کند.')

fig, ax = plt.subplots(figsize=(10.5, 4.8))
indicators = ['سهم صنعت فناوری اطلاعات', 'سهم دسته برنامه‌نویسی', 'سهم مهارت داده',
              'سهم مهارت هوش مصنوعی', 'نرخ دورکاری']
for name, color in zip(indicators, PALETTE):
    values = proxy_table.loc[[1401, 1402], name]
    ax.plot([1401, 1402], values, marker='o', linewidth=2, color=color, label=fa(name))
    ax.annotate(f'{values.iloc[-1]:.1f}٪', (1402, values.iloc[-1]), xytext=(6, 0),
                textcoords='offset points', fontsize=8, color=color, va='center')
ax.set_xticks([1401, 1402])
ax.set_xticklabels([fa('۱۴۰۱'), fa('۱۴۰۲')])
ax.set_xlim(1400.7, 1402.45)
ax.set_ylabel(fa('درصد از کل آگهی‌ها'))
ax.legend(fontsize=8, loc='upper left')
heading(ax, '۱۵.۱', 'روند شاخص‌های فناوری، داده و دورکاری')
complete('15.1')
```

**تفسیر جدول‌های ۱۵.۱ و ۱۵.۲ و نمودار ۱۵.۱:** شاخص‌های جانشین، تصویری متفاوت از روایت جهانی نشان می‌دهند. سهم صنعت فناوری اطلاعات از ۹٫۲٪ به ۷٫۷٪ و سهم دسته برنامه‌نویسی از ۵٫۲٪ به ۳٫۵٪ کاهش یافته است، در حالی که در همان دوره، بازار جهانی شاهد رشد تقاضای فناوری و هوش مصنوعی بود. نرخ دورکاری تقریباً ثابت مانده (۶٫۷٪ به ۷٪).

سهم مهارت‌های داده و هوش مصنوعی در این داده بسیار کوچک است: کمتر از ۱٪ آگهی‌ها مهارت‌هایی مانند Python یا Keras را ذکر کرده‌اند و واژگانی مثل TensorFlow، PyTorch یا «یادگیری ماشین» در فهرست مهارت‌های داده وجود ندارد. این محدودیت مهم است: فهرست مهارت جاب‌ویژن از نظر فناوری‌های نوظهور کامل نیست، بنابراین عدد ما **کف** تقاضای مهارت هوش مصنوعی است، نه اندازه واقعی آن.

در سمت دستمزد، مسیرهای فناوری بالاتر از میانه بازار قرار دارند: برنامه‌نویسی با میانه ۱۷٫۵ میلیون حدود **+۵۹٪** بالاتر از میانه کل است؛ شبکه و DevOps و دیجیتال مارکتینگ در سطح میانه بازار یا کمی بالاترند.

# ۱۶. مدل‌سازی حقوق و تغییر اهمیت ویژگی‌ها در زمان

## ۱۶.۱. طراحی مدل

هدف، پیش‌بینی **لگاریتم میانه بازه حقوق** از ویژگی‌های شغلی، مهارتی، مکانی و سازمانی است. لگاریتم می‌گیریم چون توزیع حقوق چوله است (بخش ۷). مدل پایه، رگرسیون ریج با ویژگی‌های استانداردشده است تا ضرایب بین متغیرها قابل مقایسه باشند. برای هر سال یک مدل جدا آموزش می‌دهیم تا ببینیم عوامل تعیین‌کننده در طول زمان چگونه تغییر کرده‌اند.

**کنترل‌های روش‌شناختی:** پرت‌های حقوق در صدک ۱ و ۹۹ وینزور می‌شوند، متغیرهای عددی استاندارد می‌شوند، متغیرهای دسته‌ای با One-Hot (حذف اولین سطح) وارد می‌شوند و ارزیابی با اعتبارسنجی متقابل ۵ تایی انجام می‌شود.

</div>

```python
from sklearn.linear_model import Ridge
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.model_selection import KFold, cross_val_score, train_test_split
from sklearn.metrics import r2_score, mean_absolute_error
from sklearn.ensemble import HistGradientBoostingRegressor

NUMERIC_FEATURES = ['Jobpost_RequiredExperienceYears', 'weekly_hours', 'benefit_count', 'skill_count',
                    'company_size_rank', 'company_age_rank', 'company_industry_count', 'is_remote',
                    'is_internship', 'military_card', 'local_priority', 'has_academic', 'age_mid']
CATEGORICAL_FEATURES = ['category_group', 'province_group', 'Jobpost_WorkTypeFa', 'seniority',
                        'max_degree', 'ownership', 'activity_type', 'industry_group', 'gender_pref']

model_frame = paid.copy()
model_frame['is_remote'] = model_frame.Jobpost_IsRemote.astype(int)
model_frame['is_internship'] = model_frame.Jobpost_IsInternship.astype(int)
model_frame['military_card'] = model_frame.Jobpost_RequiredMilitaryServiceCard.astype(int)
model_frame['local_priority'] = model_frame.Jobpost_PriorityWithLocalCandidate.astype(int)
model_frame['has_academic'] = model_frame.has_academic.astype(int)
model_frame['age_mid'] = model_frame.age_mid.fillna(model_frame.age_mid.median())
model_frame['weekly_hours'] = model_frame.weekly_hours.fillna(model_frame.weekly_hours.median())
winsor_low, winsor_high = model_frame.salary_mid.quantile([.01, .99])
model_frame['target_salary'] = model_frame.salary_mid.clip(winsor_low, winsor_high)
model_frame['log_target'] = np.log(model_frame.target_salary)
base_features = NUMERIC_FEATURES + CATEGORICAL_FEATURES

show_metrics('طراحی مدل', [
    ('آگهی آموزش', f'{len(model_frame):,}'),
    ('ویژگی عددی', f'{len(NUMERIC_FEATURES)}'),
    ('ویژگی دسته‌ای', f'{len(CATEGORICAL_FEATURES)}'),
    ('کران وینزور', f'{winsor_low:.1f} تا {winsor_high:.1f} میلیون'),
])
```

**تفسیر طراحی مدل:** مدل روی ۱۱۹٬۰۴۸ آگهی دارای حقوق آموزش می‌بیند؛ ۱۳ ویژگی عددی و ۹ ویژگی دسته‌ای (که با One-Hot به ده‌ها ستون تبدیل می‌شوند). پرت‌ها در صدک ۱ و ۹۹ وینزور شده‌اند (۵٫۵ تا ۴۰ میلیون تومان) تا یک آگهی ۸۵ میلیونی نتواند ضرایب را جابه‌جا کند. متغیر هدف، لگاریتم حقوق است تا توزیع چوله مدیریت شود.

## ۱۶.۲. آموزش مدل سالانه و کارایی

</div>

```python
def design_matrix(frame, columns=None):
    numeric = frame[NUMERIC_FEATURES].astype(float)
    dummies = pd.get_dummies(frame[CATEGORICAL_FEATURES].astype(str), drop_first=True, dtype=float)
    design = pd.concat([numeric, dummies], axis=1)
    if columns is not None:
        design = design.reindex(columns=columns, fill_value=0.0)
    return design


year_frames = {year: model_frame.loc[model_frame.jyear.eq(year)].copy() for year in [1401, 1402]}
union_columns = sorted(set(design_matrix(year_frames[1401]).columns) |
                       set(design_matrix(year_frames[1402]).columns))

models, performance_rows = {}, []
cv = KFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)
for year, frame in year_frames.items():
    design = design_matrix(frame, union_columns)
    scaler = StandardScaler().fit(design[NUMERIC_FEATURES])
    design[NUMERIC_FEATURES] = scaler.transform(design[NUMERIC_FEATURES])
    ridge = Ridge(alpha=1.0).fit(design, frame.log_target)
    prediction = ridge.predict(design)
    cv_r2 = cross_val_score(make_pipeline(StandardScaler(), Ridge(alpha=1.0)), design,
                            frame.log_target, cv=cv, scoring='r2').mean()
    mae = mean_absolute_error(frame.target_salary, np.expm1(prediction))
    baseline_mae = mean_absolute_error(frame.target_salary,
                                       np.full(len(frame), frame.target_salary.median()))
    X_train, X_test, y_train, y_test = train_test_split(design, frame.log_target, test_size=0.2,
                                                        random_state=RANDOM_STATE)
    hgb = HistGradientBoostingRegressor(max_iter=250, learning_rate=0.07, max_depth=6,
                                        random_state=RANDOM_STATE).fit(X_train, y_train)
    hgb_r2 = r2_score(y_test, hgb.predict(X_test))
    performance_rows.append({'سال': year, 'تعداد آگهی': len(frame), 'ضریب تعیین آموزش': r2_score(frame.log_target, prediction),
                             'ضریب تعیین اعتبارسنجی': cv_r2, 'خطای مطلق مدل (میلیون)': mae,
                             'خطای مطلق مدل پایه (میلیون)': baseline_mae,
                             'ضریب تعیین مدل درختی': hgb_r2})
    models[year] = {'ridge': ridge, 'scaler': scaler, 'design': design, 'frame': frame,
                    'prediction': prediction}
performance = pd.DataFrame(performance_rows).set_index('سال')
show_table('16.1', 'کارایی مدل‌های حقوق در هر سال', performance,
           formats={'تعداد آگهی': 'int', 'ضریب تعیین آموزش': 'score', 'ضریب تعیین اعتبارسنجی': 'score',
                    'خطای مطلق مدل (میلیون)': 'money', 'خطای مطلق مدل پایه (میلیون)': 'money',
                    'ضریب تعیین مدل درختی': 'score'},
           sample=f'{len(model_frame):,} آگهی با حقوق اعلام‌شده', period=False, index_name='سال',
           note='مدل پایه، پیش‌بینی میانه برای همه آگهی‌ها است؛ مدل درختی سقف عملکرد غیرخطی را نشان می‌دهد.')
```

**تفسیر جدول ۱۶.۱:** مدل ریج در هر دو سال ضریب تعیین اعتبارسنجی **۰٫۴۴** دارد؛ یعنی حدود ۴۴٪ از تغییرات لگاریتم حقوق با ویژگی‌های مشاهده‌شده توضیح داده می‌شود. خطای مطلق مدل در ۱۴۰۱ حدود ۲٫۵ میلیون و در ۱۴۰۲ حدود ۳٫۴ میلیون تومان است، در حالی که مدل پایه (پیش‌بینی میانه برای همه) خطای ۳٫۲ و ۴٫۴ میلیون دارد؛ بنابراین مدل حدود ۲۲٪ خطا را کم می‌کند.

مدل درختی گرادیان (HistGradientBoosting) به R² حدود ۰٫۵۸ می‌رسد؛ یعنی روابط غیرخطی و تعامل‌ها حدود ۱۴ واحد درصد قدرت توضیح اضافه می‌کنند. باقی ۴۲٪، سهم عواملی است که در داده نیست: برند شرکت، شرایط مذاکره، مهارت‌های نرم، و تفاوت‌های درون‌شغلی.

## ۱۶.۳. کدام عوامل بیشترین اثر را دارند و چگونه تغییر کرده‌اند؟

</div>

```python
def numeric_effects(year):
    model = models[year]
    coefficients = pd.Series(model['ridge'].coef_, index=model['design'].columns)
    selected = coefficients[NUMERIC_FEATURES]
    table = pd.DataFrame({'ضریب استانداردشده': selected,
                          'اثر هر انحراف معیار': (np.exp(selected) - 1) * 100})
    return table.reindex(table['ضریب استانداردشده'].abs().sort_values(ascending=False).index)


numeric_1401 = numeric_effects(1401).rename(columns={'ضریب استانداردشده': 'ضریب ۱۴۰۱',
                                                     'اثر هر انحراف معیار': 'اثر ۱۴۰۱ (٪)'})
numeric_1402 = numeric_effects(1402).rename(columns={'ضریب استانداردشده': 'ضریب ۱۴۰۲',
                                                     'اثر هر انحراف معیار': 'اثر ۱۴۰۲ (٪)'})
numeric_compare = numeric_1402.join(numeric_1401, how='outer')
numeric_compare['تغییر اثر (واحد درصدی)'] = numeric_compare['اثر ۱۴۰۲ (٪)'] - numeric_compare['اثر ۱۴۰۱ (٪)']
numeric_compare = numeric_compare.reindex(numeric_compare['اثر ۱۴۰۲ (٪)'].abs().sort_values(ascending=False).index).head(10)
show_table('16.2', 'اثر استانداردشده ویژگی‌های عددی بر حقوق در دو سال', numeric_compare,
           formats={'ضریب ۱۴۰۱': 'score', 'اثر ۱۴۰۱ (٪)': 'pp', 'ضریب ۱۴۰۲': 'score',
                    'اثر ۱۴۰۲ (٪)': 'pp', 'تغییر اثر (واحد درصدی)': 'pp'},
           sample=f'{len(model_frame):,} آگهی با حقوق', period=False, index_name='ویژگی',
           note='اثر هر انحراف معیار یعنی درصد تغییر حقوق با یک انحراف معیار افزایش آن متغیر، با ثبات بقیه.')

category_effects = []
for year in [1401, 1402]:
    model = models[year]
    coefficients = pd.Series(model['ridge'].coef_, index=model['design'].columns)
    dummies = [name for name in coefficients.index if name.startswith(tuple(f'{c}_' for c in CATEGORICAL_FEATURES))]
    top = coefficients[dummies].abs().sort_values(ascending=False).head(8)
    category_effects.append((coefficients[top.index] * 100).rename(f'اثر {year} (٪)'))
category_compare = pd.concat(category_effects, axis=1)
category_compare['تغییر'] = category_compare['اثر 1402 (٪)'] - category_compare['اثر 1401 (٪)']
category_compare.index = [index.replace('_', ' | ') for index in category_compare.index]
show_table('16.3', 'اثر سطوح دسته‌ای بر حقوق در دو سال (درصد نسبت به سطح پایه)', category_compare,
           formats={'اثر 1401 (٪)': 'pp', 'اثر 1402 (٪)': 'pp', 'تغییر': 'pp'},
           sample=f'{len(model_frame):,} آگهی با حقوق', period=False, index_name='ویژگی | سطح')

fig, axes = plt.subplots(1, 2, figsize=(14, 5.4), gridspec_kw={'width_ratios': [1.1, 1]})
ax = axes[0]
plot_features = numeric_compare.head(8).iloc[::-1]
positions = np.arange(len(plot_features))
ax.barh(positions - 0.19, plot_features['اثر ۱۴۰۱ (٪)'], height=0.36, color=BLUE, label=fa('۱۴۰۱'))
ax.barh(positions + 0.19, plot_features['اثر ۱۴۰۲ (٪)'], height=0.36, color=TEAL, label=fa('۱۴۰۲'))
ax.set_yticks(positions)
ax.set_yticklabels([fa(name) for name in plot_features.index], fontsize=8)
ax.axvline(0, color=DARK, linewidth=0.9)
ax.set_xlabel(fa('اثر هر انحراف معیار بر حقوق (٪)'))
ax.legend(fontsize=8)
heading(ax, '۱۶.۱', 'تغییر اثر ویژگی‌های عددی بین دو سال')
ax = axes[1]
prediction = models[1402]['prediction']
sample_index = np.random.RandomState(RANDOM_STATE).choice(len(prediction), size=min(12000, len(prediction)),
                                                          replace=False)
ax.scatter(models[1402]['frame'].target_salary.iloc[sample_index], np.expm1(prediction[sample_index]),
           s=6, alpha=0.25, color=TEAL)
limit = [0, 60]
ax.plot(limit, limit, color=RED, linewidth=1.2, linestyle='--', label=fa('خط پیش‌بینی کامل'))
ax.set_xlim(limit)
ax.set_ylim(limit)
ax.set_xlabel(fa('حقوق واقعی (میلیون تومان)'))
ax.set_ylabel(fa('حقوق پیش‌بینی‌شده (میلیون تومان)'))
ax.legend(fontsize=8)
heading(ax, '۱۶.۲', 'واقعی در برابر پیش‌بینی، مدل ۱۴۰۲')
complete('16.1-16.2')

show_metrics('مدل حقوق', [
    ('ضریب تعیین اعتبارسنجی ۱۴۰۱', f'{performance.loc[1401, "ضریب تعیین اعتبارسنجی"]:.3f}'),
    ('ضریب تعیین اعتبارسنجی ۱۴۰۲', f'{performance.loc[1402, "ضریب تعیین اعتبارسنجی"]:.3f}'),
    ('خطای مطلق ۱۴۰۲', f'{performance.loc[1402, "خطای مطلق مدل (میلیون)"]:.2f} میلیون'),
    ('مهم‌ترین عامل ۱۴۰۲', f'{numeric_compare.index[0]}'),
])
```

**تفسیر جدول‌های ۱۶.۲ و ۱۶.۳ و نمودارهای ۱۶.۱ و ۱۶.۲:** مهم‌ترین عامل عددی در هر دو سال، **سابقه کار** است: یک انحراف معیار بیشتر سابقه، در ۱۴۰۱ حدود **+۱۶٫۱٪** و در ۱۴۰۲ حدود **+۱۲٫۵٪** حقوق را بالا می‌برد. پس از آن، تعداد مزایا (+۴٫۵ به +۲٫۹)، داشتن رشته تحصیلی (+۴٫۱ به +۲٫۸) و اندازه شرکت (+۳٫۳ به +۲٫۷) قرار دارند. توجه کنید که **همه این اثرها از ۱۴۰۱ به ۱۴۰۲ کوچک‌تر شده‌اند**؛ یعنی بازار در سال دوم، پاداش کمتری به همان ویژگی‌ها داده است.

در سمت دسته‌ای، شکاف‌ها بزرگ و در حال عمیق‌ترشدن‌اند: نسبت به نقش مرجع (برنامه‌نویسی)، آگهی‌های کارگر ساده در ۱۴۰۱ حدود **−۵۵٫۹٪** و در ۱۴۰۲ **−۶۰٪**، کارمند اداری **−۵۰٫۳٪ به −۵۶٫۱٪** و تکنسین فنی **−۳۶٫۳٪ به −۳۷٫۴٪** بوده‌اند. در مقابل، نقش‌های مدیریتی در ۱۴۰۲ حدود **+۲۸٫۲٪** بالاتر از مرجع‌اند.

**جمع‌بندی:** در سالی که تورم بالا بود، بازده ویژگی‌های فردی (سابقه، مدرک، مزایا) کمی کاهش یافت اما شکاف بین «نوع شغل» عمیق‌تر شد؛ یعنی سرنوشت دستمزد بیش از پیش به انتخاب مسیر شغلی وابسته شده است، نه به افزایش سابقه.

<div dir="rtl" align="right">

# ۱۷. خوشه‌بندی عناوین شغلی بر پایه مهارت و تحصیلات

پرسش: کدام عناوین شغلی از نظر «سبد مهارت و تحصیلات» به هم نزدیک‌اند؟ برای پاسخ، هر عنوان شغلی را به یک بردار تبدیل می‌کنیم: سهم آگهی‌های آن عنوان که هر مهارت را می‌خواهند، به‌همراه سهم آگهی‌های نیازمند مدرک کارشناسی ارشد و بالاتر. سپس عناوین را خوشه‌بندی می‌کنیم.

</div>

```python
TITLE_MIN_ADS = 50
title_counts = ads.base_title.value_counts()
eligible_titles = title_counts[title_counts.ge(TITLE_MIN_ADS)].index
top_skills_list = skills_long.skill.value_counts().head(40).index

skill_by_title = (skills_long.loc[skills_long.ad_id.isin(ads.ad_id) & skills_long.skill.isin(top_skills_list)]
                  .merge(ads[['ad_id', 'base_title']], on='ad_id')
                  .groupby(['base_title', 'skill']).ad_id.nunique().unstack(fill_value=0)
                  .reindex(columns=top_skills_list, fill_value=0))
title_ads = ads.loc[ads.base_title.isin(eligible_titles)].groupby('base_title').size()
skill_share = skill_by_title.div(title_ads, axis=0).reindex(eligible_titles).fillna(0)

degree_share = (ads.loc[ads.base_title.isin(eligible_titles)]
                .assign(higher_degree=ads.max_degree.isin(['کارشناسی ارشد', 'دکترا']))
                .groupby('base_title').higher_degree.mean().rename('سهم مدرک ارشد و بالاتر'))
experience_median = (ads.loc[ads.base_title.isin(eligible_titles)]
                     .groupby('base_title').Jobpost_RequiredExperienceYears.median().rename('میانه سابقه'))
title_features = skill_share.join(degree_share).join(experience_median)
show_metrics('داده خوشه‌بندی', [
    ('عنوان با حداقل ۵۰ آگهی', f'{len(eligible_titles):,}'),
    ('آگهی پوشش‌داده‌شده', f'{int(title_ads.sum()):,} ({title_ads.sum() / len(ads) * 100:.1f}٪)'),
    ('مهارت در بردار', f'{len(top_skills_list)}'),
    ('ابعاد بردار هر عنوان', f'{title_features.shape[1]}'),
])
```

**تفسیر داده خوشه‌بندی:** ۷۶۲ عنوان شغلی با دست‌کم ۵۰ آگهی انتخاب شدند که روی‌هم ۷۰٫۷٪ آگهی‌ها (۲۵۴٬۴۵۱ آگهی) را پوشش می‌دهند. بردار هر عنوان ۴۲ بُعد دارد: سهم ۴۰ مهارت پرتقرار، سهم مدرک ارشد و میانه سابقه. پیش از خوشه‌بندی، بردار مهارت‌ها با **TF-IDF** وزن‌دهی شده است تا مهارت‌های همه‌جا حاضر (اکسل و ورد) کل فضا را قبضه نکنند — بدون این گام، همه عنوان‌ها در یک خوشه غول‌آسا می‌افتادند.

## ۱۷.۱. انتخاب تعداد خوشه

</div>

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score, adjusted_rand_score
from sklearn.decomposition import PCA
from sklearn.feature_extraction.text import TfidfTransformer

skill_weights = TfidfTransformer().fit_transform(skill_share.values).toarray()
degree_block = StandardScaler().fit_transform(
    title_features[['سهم مدرک ارشد و بالاتر', 'میانه سابقه']].values)
X_title = np.hstack([skill_weights, degree_block])
k_scores = []
for k in range(3, 9):
    kmeans = KMeans(n_clusters=k, n_init=20, random_state=RANDOM_STATE).fit(X_title)
    k_scores.append({'k': k, 'اینرسی': kmeans.inertia_,
                     'سیلوئت': silhouette_score(X_title, kmeans.labels_, sample_size=min(2000, len(X_title)),
                                                 random_state=RANDOM_STATE)})
k_table = pd.DataFrame(k_scores).set_index('k')
show_table('17.1', 'انتخاب تعداد خوشه با معیار سیلوئت', k_table,
           formats={'اینرسی': 'hourly', 'سیلوئت': 'ratio'},
           sample=f'{len(title_features)} عنوان شغلی', period=False, index_name='k',
           note='سیلوئت نزدیک به ۱ خوشه‌بندی تمیز و نزدیک به ۰ مرزهای مبهم را نشان می‌دهد.')

best_k = int(k_table['سیلوئت'].idxmax())
title_model = KMeans(n_clusters=best_k, n_init=50, random_state=RANDOM_STATE).fit(X_title)
title_features_clustered = title_features.copy()
title_features_clustered['خوشه'] = title_model.labels_

fig, axes = plt.subplots(1, 2, figsize=(15, 5.2), gridspec_kw={'width_ratios': [1, 1.15]})
ax = axes[0]
ax.plot(k_table.index, k_table['سیلوئت'], marker='o', color=TEAL)
ax.axvline(best_k, color=GOLD, linestyle='--', linewidth=1.4)
ax.set_xlabel(fa('تعداد خوشه'))
ax.set_ylabel(fa('امتیاز سیلوئت'))
heading(ax, '۱۷.۱', 'انتخاب تعداد خوشه')
ax = axes[1]
coords = PCA(n_components=2, random_state=RANDOM_STATE).fit_transform(X_title)
for cluster in range(best_k):
    mask = title_model.labels_ == cluster
    ax.scatter(coords[mask, 0], coords[mask, 1], s=28, alpha=0.75,
               color=PALETTE[cluster % len(PALETTE)], label=fa(f'خوشه {cluster + 1}'))
biggest_titles = title_ads.reindex(title_features.index).sort_values(ascending=False).head(12).index
for name in biggest_titles:
    position = list(title_features.index).index(name)
    ax.annotate(fa(name), (coords[position, 0], coords[position, 1]), fontsize=7,
                textcoords='offset points', xytext=(4, 3))
ax.set_xlabel(fa('مؤلفه اصلی اول'))
ax.set_ylabel(fa('مؤلفه اصلی دوم'))
ax.legend(fontsize=8)
heading(ax, '۱۷.۲', 'نقشه دوبعدی عناوین شغلی')
complete('17.1-17.2')
```

**تفسیر جدول ۱۷.۱ و نمودارهای ۱۷.۱ و ۱۷.۲:** معیار سیلوئت، **k=۳** را با امتیاز ۰٫۵۱۰ انتخاب می‌کند. نقشه دوبعدی نشان می‌دهد فضا یک «توده عمومی» بزرگ و دو «جزیره تخصصی» دارد؛ نه چند خوشه هم‌اندازه. این خودش یافته است: بازار عنوان‌های شغلی ایران در سطح مهارت، تنوع خوشه‌ای بالایی ندارد و بخش بزرگی از نقش‌ها با یک سبد مهارت مشترک (آفیس + ارتباطات اداری) اداره می‌شود.

## ۱۷.۲. پروفایل خوشه‌ها

</div>

```python
profile_rows = []
for cluster in range(best_k):
    members = title_features_clustered.loc[title_features_clustered['خوشه'].eq(cluster)]
    titles = members.index
    top_skill_names = members[top_skills_list].mean().sort_values(ascending=False).head(4).index
    profile_rows.append({
        'خوشه': f'خوشه {cluster + 1}', 'تعداد عنوان': len(titles),
        'تعداد آگهی': int(title_ads.reindex(titles).sum()),
        'میانه حقوق': paid.loc[paid.base_title.isin(titles), 'salary_mid'].median(),
        'میانه سابقه': members['میانه سابقه'].median(),
        'سهم مدرک ارشد': members['سهم مدرک ارشد و بالاتر'].mean() * 100,
        'مهارت‌های شاخص': '، '.join(top_skill_names[:3]),
        'عنوان‌های نمونه': '، '.join(title_ads.reindex(titles).sort_values(ascending=False).head(3).index),
    })
cluster_profile = pd.DataFrame(profile_rows).set_index('خوشه')
cluster_profile['سهم آگهی'] = cluster_profile['تعداد آگهی'] / cluster_profile['تعداد آگهی'].sum() * 100
show_table('17.2', 'پروفایل خوشه‌های عنوان شغلی', cluster_profile,
           formats={'تعداد عنوان': 'int', 'تعداد آگهی': 'int', 'سهم آگهی': 'pct',
                    'میانه حقوق': 'money', 'میانه سابقه': 'hours', 'سهم مدرک ارشد': 'pct'},
           sample=f'{int(cluster_profile["تعداد آگهی"].sum()):,} آگهی', period=False, index_name='خوشه')

cluster_skill_matrix = title_features_clustered.groupby('خوشه')[top_skills_list].mean() * 100
cluster_skill_matrix.index = [f'خوشه {index + 1}' for index in cluster_skill_matrix.index]
top_heat_skills = cluster_skill_matrix.mean().sort_values(ascending=False).head(18).index
fig, ax = plt.subplots(figsize=(13, 4.2))
sns.heatmap(cluster_skill_matrix[top_heat_skills], ax=ax, cmap='crest', annot=True, fmt='.0f',
            annot_kws={'size': 7}, cbar_kws={'label': 'درصد آگهی‌های خوشه'}, linewidths=0.5,
            linecolor='white')
ax.set_xticklabels([fa(name) for name in top_heat_skills], rotation=55, ha='right', fontsize=8)
ax.set_yticklabels([fa(name) for name in cluster_skill_matrix.index], rotation=0, fontsize=8.5)
ax.set_xlabel('')
ax.set_ylabel('')
heading(ax, '۱۷.۳', 'شدت تقاضای مهارت در هر خوشه')
complete('17.3')
```

**تفسیر جدول ۱۷.۲ و نمودار ۱۷.۳:** سه خوشه، سه دنیای متفاوت‌اند:

- **خوشه ۱ (۶۴۲ عنوان، ۲۳۳٬۶۳۸ آگهی، ۹۱٫۸٪):** توده عمومی بازار — فروش، حسابداری، اداری، خدمات. سبد مهارتی غالب: Excel، Word، PowerPoint؛ میانه حقوق ۱۱ میلیون، سابقه ۲ سال و سهم مدرک ارشد فقط ۲٪.
- **خوشه ۲ (۱۰۷ عنوان، ۱۹٬۴۱۳ آگهی):** نقش‌های **مالی و حسابداری ارشد** — حسابدار ارشد، مدیر مالی، رئیس حسابداری. میانه حقوق ۱۷٫۵ میلیون، سابقه ۵ سال و ۱۵٫۴٪ نیازمند ارشد و بالاتر.
- **خوشه ۳ (۱۳ عنوان، ۱٬۴۰۰ آگهی):** نقش‌های **فنی/رگولاتوری** — مسئول فنی، کارشناس رگولاتوری، نمایندگی بیمه. میانه حقوق ۱۷٫۵ میلیون و ۵۳٫۵٪ نیازمند ارشد و بالاتر.

پیام مدیریتی: تفاوت اصلی خوشه‌ها در «مهارت نرم‌افزاری» نیست، در **سطح ارشدیت و مدرک** است. برای ارتقای یک نقش از خوشه ۱ به خوشه ۲، افزودن یک مهارت جدید کافی نیست؛ ترکیب سابقه و مدرک تعیین‌کننده است.

## ۱۷.۳. آیا مرز خوشه‌ها در طول دو سال تغییر کرده است؟

</div>

```python
from scipy.optimize import linear_sum_assignment


def cluster_by_year(year, columns):
    frame = ads.loc[ads.jyear.eq(year) & ads.base_title.isin(eligible_titles)]
    counts = frame.groupby('base_title').size()
    keep = counts[counts.ge(30)].index
    skills_year = (skills_long.loc[skills_long.ad_id.isin(frame.ad_id) & skills_long.skill.isin(top_skills_list)]
                   .merge(ads[['ad_id', 'base_title']], on='ad_id')
                   .groupby(['base_title', 'skill']).ad_id.nunique().unstack(fill_value=0)
                   .reindex(columns=top_skills_list, fill_value=0))
    share = skills_year.div(counts, axis=0).fillna(0).reindex(keep)
    degree = (frame.assign(higher=frame.max_degree.isin(['کارشناسی ارشد', 'دکترا']))
              .groupby('base_title')[['higher']].mean().reindex(keep))
    experience = frame.groupby('base_title').Jobpost_RequiredExperienceYears.median().reindex(keep)
    matrix = share.join(degree.rename(columns={'higher': 'سهم مدرک ارشد و بالاتر'})).join(
        experience.rename('میانه سابقه')).fillna(0)
    return StandardScaler().fit_transform(matrix.values), matrix.index


X_1401, titles_1401 = cluster_by_year(1401, title_features.columns)
X_1402, titles_1402 = cluster_by_year(1402, title_features.columns)
labels_1401 = KMeans(n_clusters=best_k, n_init=30, random_state=RANDOM_STATE).fit_predict(X_1401)
labels_1402 = KMeans(n_clusters=best_k, n_init=30, random_state=RANDOM_STATE).fit_predict(X_1402)

common_titles = titles_1401.intersection(titles_1402)
position_1401 = {title: index for index, title in enumerate(titles_1401)}
position_1402 = {title: index for index, title in enumerate(titles_1402)}
matched = pd.DataFrame({
    '1401': [labels_1401[position_1401[title]] for title in common_titles],
    '1402': [labels_1402[position_1402[title]] for title in common_titles]}, index=common_titles)
ari = adjusted_rand_score(matched['1401'], matched['1402'])

contingency = pd.crosstab(matched['1401'], matched['1402'])
row_ind, col_ind = linear_sum_assignment(-contingency.values)
mapping = {contingency.index[row]: contingency.columns[col] for row, col in zip(row_ind, col_ind)}
matched['خوشه هم‌ارز ۱۴۰۲'] = matched['1401'].map(mapping).fillna(matched['1401'])
matched['جابه‌جا شده'] = matched['1401'] != matched['خوشه هم‌ارز ۱۴۰۲']
moved = matched.loc[matched['جابه‌جا شده']]

stability = pd.DataFrame({'مقدار': [f'{len(common_titles):,}', f'{ari:.3f}', f'{len(moved):,}',
                                    f'{len(moved) / len(common_titles) * 100:.1f}٪']},
                         index=['عنوان مشترک دو سال', 'شاخص رند تعدیل‌شده', 'عنوان جابه‌جا‌شده',
                                'سهم جابه‌جایی'])
show_table('17.3', 'پایداری خوشه‌ها بین دو سال', stability,
           sample=f'{len(common_titles)} عنوان مشترک', period=False, index_name='شاخص',
           note='شاخص رند تعدیل‌شده نزدیک به ۱ یعنی خوشه‌بندی دو سال تقریباً یکسان است؛ نزدیک به ۰ یعنی تصادفی.')
show_table('17.4', 'توزیع تعداد عنوان‌ها در تطابق خوشه‌های دو سال', contingency,
           formats={column: 'int' for column in contingency.columns},
           sample=f'{len(common_titles)} عنوان مشترک', period=False, index_name='خوشه ۱۴۰۱')
show_metrics('پایداری خوشه', [
    ('تعداد خوشه انتخابی', str(best_k)),
    ('شاخص رند تعدیل‌شده', f'{ari:.3f}'),
    ('عنوان جابه‌جا‌شده', f'{len(moved):,} از {len(common_titles):,}'),
    ('میانه سیلوئت', f'{k_table["سیلوئت"].max():.3f}'),
])
```

**تفسیر جدول‌های ۱۷.۳ و ۱۷.۴:** خوشه‌بندی مستقل هر سال و تطبیق آن‌ها، شاخص رند تعدیل‌شده **۰٫۹۵۲** می‌دهد؛ یعنی مرز خوشه‌ها در دو سال تقریباً یکسان مانده و **هیچ عنوانی** بین خوشه‌های هم‌ارز جابه‌جا نشده است. جدول ۱۷٫۴ هم نشان می‌دهد ماتریس تطبیق تقریباً قطری است.

**نتیجه:** در بازه دو ساله، «مرز مشاغل» تغییر نکرده است. مهارت‌های جدید (وردپرس، گوگل آنالیتیکس) درون همان خوشه‌ها جابه‌جا شده‌اند، اما باعث ادغام یا تفکیک خوشه‌ها نشده‌اند. این با یافته بخش ۱۹ هم‌خوان است: سبد مهارت یک عنوان ثابت می‌ماند و تحول، تدریجی است.

# ۱۸. صنعت × دسته شغلی: هر صنعت چه تخصصی می‌خواهد؟

</div>

```python
industry_share = pd.crosstab(ads.industry_group, ads.category_group, normalize='index') * 100
industry_ads = ads.industry_group.value_counts()
top_industries_list = industry_ads.head(8).index
top_categories_list = ads.category_group.value_counts().head(10).index

industry_rows = []
for industry in top_industries_list:
    row = industry_share.loc[industry].sort_values(ascending=False)
    top_three = row.head(3)
    hhi = (row ** 2).sum() / 100
    industry_rows.append({
        'صنعت': industry, 'تعداد آگهی': int(industry_ads[industry]),
        'دسته نخست': f'{top_three.index[0]} ({top_three.iloc[0]:.0f}٪)',
        'دسته دوم': f'{top_three.index[1]} ({top_three.iloc[1]:.0f}٪)',
        'دسته سوم': f'{top_three.index[2]} ({top_three.iloc[2]:.0f}٪)',
        'تمرکز تخصصی': hhi,
    })
industry_table = pd.DataFrame(industry_rows).set_index('صنعت')
show_table('18.1', 'سه دسته شغلی نخست و تمرکز تخصصی هر صنعت', industry_table,
           formats={'تعداد آگهی': 'int', 'تمرکز تخصصی': 'hours'},
           sample=f'{len(ads):,} آگهی', period=False, index_name='صنعت',
           note='تمرکز تخصصی در مقیاس ۰ تا ۱۰۰؛ عدد بزرگ‌تر یعنی صنعت وابسته به تخصص‌های محدودتر است.')

fig, ax = plt.subplots(figsize=(13, 5.6))
heat = industry_share.loc[top_industries_list, top_categories_list]
sns.heatmap(heat, ax=ax, cmap='crest', annot=True, fmt='.0f', annot_kws={'size': 7},
            cbar_kws={'label': 'درصد از آگهی‌های صنعت'}, linewidths=0.5, linecolor='white')
ax.set_xticklabels([fa(name) for name in top_categories_list], rotation=45, ha='right', fontsize=8)
ax.set_yticklabels([fa(name) for name in top_industries_list], rotation=0, fontsize=8.5)
ax.set_xlabel('')
ax.set_ylabel('')
heading(ax, '۱۸.۱', 'ترکیب تقاضای تخصصی در صنایع پرتکرار')
complete('18.1')
```

**تفسیر جدول ۱۸.۱ و نمودار ۱۸.۱:** هر صنعت، ترکیب تخصصی خود را دارد:

- **فناوری اطلاعات** متمرکزترین صنعت است: ۲۳٪ برنامه‌نویسی و ۱۷٪ شبکه/DevOps؛ تمرکز تخصصی ۱۶٫۴ (در مقیاس ۰ تا ۱۰۰).
- **تجارت و تولید** حول مالی و حسابداری (۲۲٪ و ۱۶٪) و فروش می‌چرخند.
- **اینترنت و تجارت الکترونیک** بیشترین سهم دیجیتال مارکتینگ (۱۳٪) را دارد.
- **خدمات مهندسی و تخصصی** کم‌تمرکزترین صنعت است (تمرکز ۲۹٫۵), چون ۵۱٪ آگهی‌هایش در دسته‌های متفرقه است.

نکته مهم برای مدیر: «صنعت» و «دسته شغلی» دو محور متفاوت‌اند. یک شرکت تولیدی می‌تواند برنامه‌نویس بخواهد (۲۳٪ از آگهی‌های صنعت فناوری اطلاعات در صنایع دیگر منتشر می‌شود). بنابراین برای بنچمارک حقوق، همیشه دسته شغلی را مبنا بگیرید، نه صنعت را.

# ۱۹. تحول مهارت‌ها و جهش عناوین شغلی

## ۱۹.۱. مهارت‌های در حال رشد و افت

</div>

```python
skill_year = (skills_long.merge(ads[['ad_id', 'jyear']], on='ad_id', how='inner')
              .groupby(['skill', 'jyear']).ad_id.nunique().unstack(fill_value=0))
ads_with_skills = ads.assign(has_skill=ads.ad_id.isin(skills_long.ad_id)).groupby('jyear').has_skill.sum()
skill_share_year = skill_year.div(ads_with_skills, axis=1) * 100
skill_share_year.columns = ['سهم ۱۴۰۱', 'سهم ۱۴۰۲']
skill_share_year['تغییر واحد درصدی'] = skill_share_year['سهم ۱۴۰۲'] - skill_share_year['سهم ۱۴۰۱']
skill_share_year['رشد نسبی'] = (skill_share_year['سهم ۱۴۰۲'] / skill_share_year['سهم ۱۴۰۱'].replace(0, np.nan) - 1) * 100
skill_share_year['تعداد ۱۴۰۲'] = skill_year.get(1402, 0)
considerable = skill_share_year.loc[(skill_share_year['سهم ۱۴۰۱'].ge(0.3)) | (skill_share_year['سهم ۱۴۰۲'].ge(0.3))]
risers = considerable.sort_values('تغییر واحد درصدی', ascending=False).head(10)
fallers = considerable.sort_values('تغییر واحد درصدی').head(10)
skill_movers = pd.concat([risers, fallers])
show_table('19.1', 'مهارت‌های با بیشترین رشد و افت سهم بین دو سال', skill_movers,
           formats={'سهم ۱۴۰۱': 'pct', 'سهم ۱۴۰۲': 'pct', 'تغییر واحد درصدی': 'pp',
                    'رشد نسبی': 'pp', 'تعداد ۱۴۰۲': 'int'},
           sample='مهارت‌هایی با سهم دست‌کم ۰٫۳٪ در یکی از دو سال', period=False, index_name='مهارت')

fig, ax = plt.subplots(figsize=(11.5, 6.2))
movers_plot = skill_movers.iloc[::-1]
positions = np.arange(len(movers_plot))
for position, (name, row) in enumerate(movers_plot.iterrows()):
    color = TEAL if row['تغییر واحد درصدی'] >= 0 else RED
    ax.plot([row['سهم ۱۴۰۱'], row['سهم ۱۴۰۲']], [position, position], color=color, linewidth=2, alpha=0.65)
    ax.scatter(row['سهم ۱۴۰۱'], position, color=GRAY, s=34, zorder=3)
    ax.scatter(row['سهم ۱۴۰۲'], position, color=color, s=34, zorder=3)
ax.set_yticks(positions)
ax.set_yticklabels([fa(name) for name in movers_plot.index], fontsize=8.5)
ax.set_xlabel(fa('سهم از آگهی‌های دارای مهارت (٪)'))
grey = plt.Line2D([], [], color=GRAY, marker='o', linestyle='', label=fa('۱۴۰۱'))
teal = plt.Line2D([], [], color=TEAL, marker='o', linestyle='', label=fa('۱۴۰۲ (رشد)'))
red = plt.Line2D([], [], color=RED, marker='o', linestyle='', label=fa('۱۴۰۲ (افت)'))
ax.legend(handles=[grey, teal, red], fontsize=8.5)
heading(ax, '۱۹.۱', 'جابه‌جایی سهم مهارت‌ها بین دو سال')
complete('19.1')
```

**تفسیر جدول ۱۹.۱ و نمودار ۱۹.۱:** سبد مهارت بازار در دو سال، تغییرات آرام اما جهت‌دار داشته است. بزرگ‌ترین رشدها: **WordPress +۱٫۴ واحد درصد**، Microsoft Word +۱٫۱، **Google Analytics +۱٫۰**، سپیدار +۰٫۸ و Excel +۰٫۷. بزرگ‌ترین افت‌ها: **Microsoft Outlook −۴٫۲**، PowerPoint −۴٫۰، Microsoft Dynamics −۱٫۱، Microsoft Project −۱٫۰، HTML/CSS −۰٫۸ و JavaScript −۰٫۷.

دو روایت از این اعداد برمی‌آید. اول، **چرخش به سمت دیجیتال مارکتینگ**: وردپرس، گوگل آنالیتیکس و Keyword Planner رشد کرده‌اند و با رشد سهم دیجیتال مارکتینگ در بخش ۱۰ هم‌خوان است. دوم، **جایگزینی ابزارهای اداری سنگین**: Outlook و PowerPoint و Project جای خود را به ابزارهای سبک‌تر و ارتباطی داده‌اند؛ بخشی از این افت ممکن است ناشی از تغییر شیوه ثبت مهارت در آگهی‌ها باشد، نه کاهش واقعی نیاز.

**هشدار:** مهارت‌های برنامه‌نویسی (JavaScript، HTML/CSS، C#) در سهم خود افت کرده‌اند، هم‌زمان با افت سهم دسته برنامه‌نویسی. این دو داده یکدیگر را تأیید می‌کنند.

## ۱۹.۲. عناوین شغلی در حال جهش

</div>

```python
title_year = (ads.groupby(['base_title', 'jyear']).size().unstack(fill_value=0))
title_year.columns = ['تعداد ۱۴۰۱', 'تعداد ۱۴۰۲']
year_totals = ads.groupby('jyear').size()
title_year['سهم ۱۴۰۱'] = title_year['تعداد ۱۴۰۱'] / year_totals[1401] * 100
title_year['سهم ۱۴۰۲'] = title_year['تعداد ۱۴۰۲'] / year_totals[1402] * 100
title_year['تغییر واحد درصدی'] = title_year['سهم ۱۴۰۲'] - title_year['سهم ۱۴۰۱']
title_year['رشد تعداد'] = (title_year['تعداد ۱۴۰۲'] / title_year['تعداد ۱۴۰۱'].replace(0, np.nan) - 1) * 100
title_year['میانه حقوق ۱۴۰۲'] = paid.loc[paid.jyear.eq(1402)].groupby('base_title').salary_mid.median()
big_titles_change = title_year.loc[(title_year['تعداد ۱۴۰۱'].ge(50)) & (title_year['تعداد ۱۴۰۲'].ge(50))]
rising_titles = big_titles_change.sort_values('رشد تعداد', ascending=False).head(10)
falling_titles = big_titles_change.sort_values('رشد تعداد').head(10)
title_movers = pd.concat([rising_titles, falling_titles])
show_table('19.2', 'عناوین شغلی با بیشترین رشد و افت تعداد', title_movers,
           formats={'تعداد ۱۴۰۱': 'int', 'تعداد ۱۴۰۲': 'int', 'سهم ۱۴۰۱': 'pct', 'سهم ۱۴۰۲': 'pct',
                    'تغییر واحد درصدی': 'pp', 'رشد تعداد': 'pp', 'میانه حقوق ۱۴۰۲': 'money'},
           sample='عنوان‌های با دست‌کم ۵۰ آگهی در هر دو سال', period=False, index_name='عنوان شغلی',
           note='شرط حضور در هر دو سال، از رشدهای انفجاری ناشی از نمونه کوچک جلوگیری می‌کند.')

fig, ax = plt.subplots(figsize=(11.5, 6.2))
title_plot = title_movers.sort_values('رشد تعداد').iloc[::-1]
colors = [RED if value < 0 else TEAL for value in title_plot['رشد تعداد']]
ax.barh(range(len(title_plot)), title_plot['رشد تعداد'], color=colors, height=0.7)
ax.set_yticks(range(len(title_plot)))
ax.set_yticklabels([fa(name) for name in title_plot.index], fontsize=8.5)
ax.axvline(0, color=DARK, linewidth=0.9)
ax.set_xlabel(fa('رشد تعداد آگهی ۱۴۰۲ نسبت به ۱۴۰۱ (٪)'))
for position, value in enumerate(title_plot['رشد تعداد']):
    offset = 4 if value >= 0 else -4
    ax.text(value + offset, position, f'{value:+,.0f}٪', va='center',
            ha='left' if value >= 0 else 'right', fontsize=7.5)
heading(ax, '۱۹.۲', 'عناوین شغلی با بیشترین جهش تقاضا')
complete('19.2')
```

**تفسیر جدول ۱۹.۲ و نمودار ۱۹.۲:** با شرط حضور دست‌کم ۵۰ آگهی در هر دو سال (برای حذف رشدهای انفجاری نمونه کوچک)، سریع‌ترین رشدها متعلق به **باریستا (+۴۲۲٪)**، **آشپز (+۳۱۸٪)**، **ادمین اینستاگرام (+۳۰۹٪)**، **کارگر بسته‌بندی (+۲۹۵٪)**، **طراح سایت (+۲۶۴٪)** و **کارشناس ایمنی (+۲۵۹٪)** است. در سمت افت: **نماینده بیمه (−۷۷٪)**، **مدیر فروش منطقه (−۶۷٪)**، کارشناس بازاریابی و تبلیغات (−۴۳٪) و کارآموز فروش (−۴۰٪).

این الگو با تغییرات ساختاری بازار ایران هم‌خوان است: رشد خدمات مصرفی (کافه، رستوران، بسته‌بندی) و مشاغل شبکه‌های اجتماعی، و کاهش نقش‌های فروش بیمه‌ای و مدیریت فروش منطقه‌ای. نکته مهم: «ادمین اینستاگرام» و «ادمین شبکه‌های اجتماعی» دو عنوان جداگانه‌اند که در نرمال‌سازی ادغام نشده‌اند؛ بنابراین رشد واقعی این حوزه از عدد هر عنوان بیشتر است.

## ۱۹.۳. آیا سبد مهارت یک عنوان ثابت تغییر کرده است؟

</div>

```python
title_skill_share = (skills_long.merge(ads[['ad_id', 'base_title', 'jyear']], on='ad_id')
                     .groupby(['base_title', 'jyear', 'skill']).ad_id.nunique().unstack(fill_value=0))
title_ads_year = (ads.assign(has_skill=ads.ad_id.isin(skills_long.ad_id))
                  .groupby(['base_title', 'jyear']).has_skill.sum())
title_skill_rate = title_skill_share.div(title_ads_year, axis=0) * 100

stability_rows = []
top_titles_by_volume = ads.base_title.value_counts().head(12).index
for title in top_titles_by_volume:
    try:
        first = set(title_skill_rate.loc[(title, 1401)].dropna().loc[lambda s: s.ge(10)].index)
        second = set(title_skill_rate.loc[(title, 1402)].dropna().loc[lambda s: s.ge(10)].index)
    except KeyError:
        continue
    if not first or not second:
        continue
    jaccard = len(first & second) / len(first | second)
    joined = pd.DataFrame({'۱۴۰۱': title_skill_rate.loc[(title, 1401)],
                           '۱۴۰۲': title_skill_rate.loc[(title, 1402)]}).dropna()
    joined['تغییر'] = joined['۱۴۰۲'] - joined['۱۴۰۱']
    stability_rows.append({
        'عنوان شغلی': title, 'تعداد آگهی': int(ads.base_title.eq(title).sum()),
        'شباهت سبد مهارت': jaccard,
        'مهارت در حال رشد': joined['تغییر'].idxmax() if len(joined) else np.nan,
        'رشد مهارت (واحد درصدی)': joined['تغییر'].max() if len(joined) else np.nan,
        'مهارت در حال افت': joined['تغییر'].idxmin() if len(joined) else np.nan,
        'افت مهارت (واحد درصدی)': joined['تغییر'].min() if len(joined) else np.nan,
    })
title_stability = pd.DataFrame(stability_rows).set_index('عنوان شغلی').sort_values('تعداد آگهی', ascending=False)
show_table('19.3', 'پایداری سبد مهارت در پرتکرارترین عناوین شغلی', title_stability,
           formats={'تعداد آگهی': 'int', 'شباهت سبد مهارت': 'ratio', 'رشد مهارت (واحد درصدی)': 'pp',
                    'افت مهارت (واحد درصدی)': 'pp'},
           sample='مهارت‌های با سهم دست‌کم ۱۰٪ در هر سال', period=False, index_name='عنوان شغلی',
           note='شباهت جاکارد ۱ یعنی سبد مهارت‌های پرتقاضا کاملاً ثابت مانده و ۰ یعنی کاملاً عوض شده است.')

show_metrics('تحول مهارت و عنوان', [
    ('میانه شباهت سبد مهارت', f'{title_stability["شباهت سبد مهارت"].median():.2f}'),
    ('بیشترین رشد مهارت', f'{risers.index[0]} ({risers["تغییر واحد درصدی"].iloc[0]:+.2f} واحد)'),
    ('بیشترین افت مهارت', f'{fallers.index[0]} ({fallers["تغییر واحد درصدی"].iloc[0]:+.2f} واحد)'),
    ('سریع‌ترین رشد عنوان', f'{rising_titles.index[0]} ({rising_titles["رشد تعداد"].iloc[0]:+,.0f}٪)'),
])
```

**تفسیر جدول ۱۹.۳:** سبد مهارت پرتقرارترین عنوان‌ها پایدار است: میانه شباهت جاکارد **۰٫۹۲**. حسابدار (۰٫۸۳)، کارشناس فروش تلفنی (۰٫۶۰) و انباردار (۰٫۷۱) بیشترین تغییر را داشته‌اند، اما تغییر اصلی درون همان مهارت‌های موجود است: Excel بالا، Outlook پایین، و ورود تدریجی نرم‌افزارهای بومی (سپیدار، راهکاران، همکاران سیستم).

یعنی «عنوان شغلی» ظرفیت ثابتی است که محتوای مهارتی‌اش آرام تغییر می‌کند؛ به همین دلیل خوشه‌بندی بخش ۱۷ در دو سال پایدار ماند.

<div dir="rtl" align="right">

# ۲۰. پروفایل‌سازی و خوشه‌بندی کارفرمایان [امتیازی]

از آنجا که نام شرکت‌ها ناشناس است، کارفرما را با ویژگی‌های قابل مشاهده توصیف می‌کنیم: اندازه، قدمت، مالکیت، دامنه فعالیت، صنعت، استان، شمار مزایا و رفتار آگهی (افشای حقوق، دورکاری). داده ترکیبی است؛ بنابراین از **فاصله گاور** استفاده می‌کنیم که هم متغیر عددی و هم دسته‌ای را در یک سنجه فاصله جمع می‌کند، و سپس خوشه‌بندی سلسله‌مراتبی روی همان ماتریس فاصله انجام می‌شود.

</div>

```python
from gower import gower_matrix
from sklearn.cluster import AgglomerativeClustering
from sklearn.metrics import silhouette_score

employer_numeric = ['company_size_rank', 'company_age_rank', 'benefit_count', 'skill_count',
                    'exp_required', 'remote_flag', 'transparency_flag']
employer_categorical = ['ownership', 'activity_type', 'company_industry_main', 'Jobpost_ProvinceFa',
                        'Jobpost_MainJobCategory']
employer_source = ads.assign(exp_required=ads.Jobpost_RequiredExperienceYears,
                             remote_flag=ads.Jobpost_IsRemote.astype(int),
                             transparency_flag=ads.salary_disclosed.astype(int))
employer_source = employer_source.dropna(subset=employer_numeric + employer_categorical)
employer_features = employer_source[employer_numeric + employer_categorical].copy()
for column in employer_categorical:
    employer_features[column] = employer_features[column].astype(str).astype('category').cat.codes
employer_features = employer_features.astype(float)

sample_index = employer_features.sample(min(4000, len(employer_features)),
                                        random_state=RANDOM_STATE).index
gower_matrix_employer = gower_matrix(employer_features.loc[sample_index].values,
                                     cat_features=[False] * len(employer_numeric) +
                                                  [True] * len(employer_categorical))

selection_rows = []
for k in [3, 4, 5, 6]:
    labels = AgglomerativeClustering(n_clusters=k, metric='precomputed',
                                     linkage='average').fit_predict(gower_matrix_employer)
    selection_rows.append({'k': k, 'سیلوئت گاور': silhouette_score(gower_matrix_employer, labels,
                                                                   metric='precomputed'),
                           'کوچک‌ترین خوشه': int(pd.Series(labels).value_counts().min())})
selection = pd.DataFrame(selection_rows).set_index('k')
show_table('20.1', 'انتخاب تعداد تیپ کارفرما روی ماتریس فاصله گاور', selection,
           formats={'سیلوئت گاور': 'ratio', 'کوچک‌ترین خوشه': 'int'},
           sample=f'{len(sample_index):,} آگهی نمونه', period=False, index_name='k',
           note='سیلوئت بالاتر و خوشه‌های متعادل‌تر بهترند؛ ساختار نهایی باید قابل تفسیر هم باشد.')

best_employer_k = 4
labels_employer = AgglomerativeClustering(n_clusters=best_employer_k, metric='precomputed',
                                          linkage='average').fit_predict(gower_matrix_employer)
employer_sample = employer_source.loc[sample_index].assign(تیپ=labels_employer)

typology_rows = []
for cluster in range(best_employer_k):
    subset = employer_sample.loc[employer_sample['تیپ'].eq(cluster)]
    typology_rows.append({
        'تیپ': f'تیپ {cluster + 1}', 'تعداد آگهی نمونه': len(subset),
        'سهم نمونه': len(subset) / len(employer_sample) * 100,
        'میانه اندازه شرکت': subset.company_size_rank.median(),
        'میانه مزایا': subset.benefit_count.median(),
        'نرخ افشای حقوق': subset.salary_disclosed.mean() * 100,
        'نرخ دورکاری': subset.Jobpost_IsRemote.mean() * 100,
        'صنعت غالب': subset.company_industry_main.value_counts().idxmax(),
        'استان غالب': subset.Jobpost_ProvinceFa.value_counts().idxmax(),
    })
employer_typology = pd.DataFrame(typology_rows).set_index('تیپ').sort_values('تعداد آگهی نمونه',
                                                                             ascending=False)
show_table('20.2', 'تیپ‌شناسی کارفرمایان بر پایه ویژگی‌های شرکت و رفتار آگهی', employer_typology,
           formats={'تعداد آگهی نمونه': 'int', 'سهم نمونه': 'pct', 'میانه اندازه شرکت': 'hours',
                    'میانه مزایا': 'hours', 'نرخ افشای حقوق': 'pct',
                    'نرخ دورکاری': 'pct'},
           sample=f'{len(employer_sample):,} آگهی نمونه', period=False, index_name='تیپ کارفرما',
           note='مقیاس اندازه شرکت ۱ تا ۷ است؛ عدد بزرگ‌تر یعنی شرکت بزرگ‌تر. ستون‌های دسته‌ای، غالب هر تیپ را نشان می‌دهند.')
```

**نقشه تیپ‌ها در فضای گاور**

</div>

```python
embedding = PCA(n_components=2, random_state=RANDOM_STATE).fit_transform(gower_matrix_employer)
fig, axes = plt.subplots(1, 2, figsize=(14, 5), gridspec_kw={'width_ratios': [1.05, 1]})
ax = axes[0]
for cluster in range(best_employer_k):
    mask = labels_employer == cluster
    ax.scatter(embedding[mask, 0], embedding[mask, 1], s=16, alpha=0.6,
               color=PALETTE[cluster % len(PALETTE)], label=fa(f'تیپ {cluster + 1}'))
ax.set_xlabel(fa('مؤلفه اول فضای فاصله گاور'))
ax.set_ylabel(fa('مؤلفه دوم فضای فاصله گاور'))
ax.legend(fontsize=8.5)
heading(ax, '۲۰.۱', 'نقشه دوبعدی کارفرمایان در فضای گاور', period='نمونه ۴٬۰۰۰ آگهی')
ax = axes[1]
for position, (name, row) in enumerate(employer_typology.iterrows()):
    ax.scatter(row['میانه مزایا'], row['نرخ افشای حقوق'], s=row['سهم نمونه'] * 80,
               color=PALETTE[position % len(PALETTE)], alpha=0.85)
    ax.annotate(fa(name), (row['میانه مزایا'], row['نرخ افشای حقوق']), xytext=(7, 5),
                textcoords='offset points', fontsize=8.5)
ax.set_xlabel(fa('میانه تعداد مزایا'))
ax.set_ylabel(fa('نرخ افشای حقوق (٪)'))
heading(ax, '۲۰.۲', 'موقعیت تیپ‌های کارفرما بر پایه مزایا و شفافیت')
complete('20.1-20.2')

centroids = employer_sample.groupby('تیپ')[employer_numeric].mean()
distance_rows = []
for first in range(best_employer_k):
    for second in range(first + 1, best_employer_k):
        distance_rows.append({'جفت تیپ': f'تیپ {first + 1} — تیپ {second + 1}',
                              'فاصله مرکزها': float(np.linalg.norm(centroids.loc[first] -
                                                                  centroids.loc[second]))})
distance_rows.sort(key=lambda row: row['فاصله مرکزها'])
show_table('20.3', 'فاصله تیپ‌های کارفرما در فضای ویژگی‌ها', pd.DataFrame(distance_rows).set_index('جفت تیپ'),
           formats={'فاصله مرکزها': 'score'}, sample=f'{len(employer_sample):,} آگهی نمونه',
           period=False, index_name='جفت تیپ',
           note='کمترین فاصله یعنی این دو تیپ کارفرما رفتاری نزدیک دارند و به هم شبیه‌اند.')

show_metrics('تیپ‌شناسی کارفرما', [
    ('تعداد تیپ', str(best_employer_k)),
    ('بیشترین نرخ افشا', f'{employer_typology["نرخ افشای حقوق"].idxmax()} — {employer_typology["نرخ افشای حقوق"].max():.1f}٪'),
    ('بیشترین میانگین مزایا', f'{employer_typology["میانه مزایا"].idxmax()} — {employer_typology["میانه مزایا"].max():.1f} مزیت'),
    ('نزدیک‌ترین جفت تیپ', distance_rows[0]['جفت تیپ']),
])
```

**تفسیر جدول‌های ۲۰.۱ تا ۲۰.۳ و نمودارهای ۲۰.۱ و ۲۰.۲:** فاصله گاور روی نمونه ۴٬۰۰۰ آگهی، ساختار روشنی نشان می‌دهد: **دو تیپ بزرگ** و **دو تیپ کوچک و متمایز**.

- **تیپ ۱ (۷۵٫۶٪ نمونه):** کارفرمایان بزرگ و متوسط با ۳ مزیت، افشای حقوق ۳۲٪ و دورکاری ۵٫۵٪ — «بدنه بازار».
- **تیپ ۳ (۲۴٫۱٪):** کارفرمایان کوچک با ۱ مزیت، افشای ۳۹٪ و دورکاری ۱۰٫۴٪ — «کارفرمای کوچک منعطف».
- **تیپ ۲ (۸ آگهی):** کارفرمایان بزرگ با ۱۱ مزیت و افشای ۸۷٫۵٪ — «کارفرمای شفاف و سخاوتمند»؛ اقلیتی بسیار کوچک اما متمایز.
- **تیپ ۴ (۲ آگهی):** یک کارفرمای کوچک در البرز با ۰٫۵ مزیت — عملاً یک پرت.

این تیپ‌شناسی نشان می‌دهد «شفافیت حقوق» و «سخاوت در مزایا» در ایران یک الگوی نادر اما موجود است؛ اگر شرکت شما در تیپ ۱ است، با افشای حقوق و افزودن دو مزیت می‌توانید به‌سرعت از توده بازار متمایز شوید.

**محدودیت:** دو تیپ آخر فقط ۱۰ آگهی دارند و نباید به‌عنوان یک «بخش بازار» تعمیم داده شوند؛ آن‌ها الگوهای حاشیه‌ای‌اند که الگوریتم جدا کرده است. سیلوئت ۰٫۲۳ نشان می‌دهد مرزها متوسط و نه کاملاً تمیزند.

<div dir="rtl" align="right">

# ۲۱. شبکه هم‌رخدادی مهارت‌ها [امتیازی]

اگر دو مهارت بارها در یک آگهی با هم خواسته شوند، احتمالاً یک «بسته کاری» را می‌سازند. شبکه‌ای می‌سازیم که گره‌هایش مهارت و یال‌هایش هم‌رخدادی دو مهارت در یک آگهی است: اندازه گره با فراوانی مهارت و ضخامت یال با تعداد هم‌رخدادی. سپس جوامع مهارتی و تغییر مرکزیت را در دو سال می‌سنجیم.

</div>

```python
from itertools import combinations
from collections import Counter

TOP_NETWORK = 35
network_skills = skills_long.skill.value_counts().head(TOP_NETWORK).index
skill_ads = skills_long.loc[skills_long.skill.isin(network_skills), ['ad_id', 'skill']].drop_duplicates()
pair_counter = Counter()
for _, group in skill_ads.groupby('ad_id').skill:
    names = sorted(set(group))
    if 1 < len(names) <= 12:
        pair_counter.update(combinations(names, 2))

frequency = skill_ads.skill.value_counts()
graph = nx.Graph()
for skill, count in frequency.items():
    graph.add_node(skill, frequency=int(count))
for (first, second), weight in pair_counter.items():
    if weight >= 20:
        graph.add_edge(first, second, weight=int(weight))
graph.remove_nodes_from(list(nx.isolates(graph)))

communities = list(nx.community.greedy_modularity_communities(graph, weight='weight'))
community_of = {node: index for index, members in enumerate(communities) for node in members}
betweenness = nx.betweenness_centrality(graph, weight='weight')
weighted_degree = dict(graph.degree(weight='weight'))

network_summary = pd.DataFrame({'مقدار': [f'{graph.number_of_nodes():,}', f'{graph.number_of_edges():,}',
                                          f'{len(communities)}', f'{len(skill_ads):,}',
                                          f'{sum(pair_counter.values()):,}']},
                               index=['گره (مهارت)', 'یال (هم‌رخدادی)', 'جامعه مهارتی',
                                      'ردیف مهارت-آگهی', 'کل جفت هم‌رخداد'])
show_table('21.1', 'خلاصه شبکه مهارت‌ها', network_summary, sample='کل دوره', period=False,
           index_name='شاخص')
```

## ۲۱.۱. ساختار شبکه و جوامع مهارتی

</div>

```python
layout = nx.spring_layout(graph, weight='weight', seed=RANDOM_STATE, k=0.55, iterations=250)
sizes = np.array([graph.nodes[node]['frequency'] for node in graph.nodes()])
size_scaled = 90 + 1500 * (sizes - sizes.min()) / (sizes.max() - sizes.min())
weights = np.array([graph[u][v]['weight'] for u, v in graph.edges()])
width_scaled = 0.4 + 4.0 * (weights - weights.min()) / max(weights.max() - weights.min(), 1)

fig, ax = plt.subplots(figsize=(12.5, 9))
nx.draw_networkx_edges(graph, layout, ax=ax, width=width_scaled, alpha=0.25, edge_color='#8aa9a6')
nx.draw_networkx_nodes(graph, layout, ax=ax, node_size=size_scaled,
                       node_color=[PALETTE[community_of[node] % len(PALETTE)] for node in graph.nodes()],
                       alpha=0.9, linewidths=0.8, edgecolors='white')
nx.draw_networkx_labels(graph, layout, ax=ax, font_size=7,
                        labels={node: fa(node) for node in graph.nodes()})
ax.set_axis_off()
heading(ax, '۲۱.۱', 'شبکه هم‌رخدادی مهارت‌ها؛ اندازه گره = فراوانی، ضخامت یال = هم‌رخدادی')
complete('21.1')

community_rows = []
for index, members in enumerate(communities, start=1):
    ranked = sorted(members, key=lambda node: frequency[node], reverse=True)
    community_rows.append({'جامعه': f'جامعه {index}', 'تعداد مهارت': len(members),
                           'مهارت‌های اصلی': '، '.join(ranked[:6]),
                           'بیشترین هم‌رخدادی': max(ranked, key=lambda node: weighted_degree[node])})
show_table('21.2', 'جوامع مهارتی شناسایی‌شده', pd.DataFrame(community_rows).set_index('جامعه'),
           sample='شبکه کل دوره', period=False, index_name='جامعه')
```

**تفسیر جدول ۲۱.۱ و نمودار ۲۱.۱:** شبکه از ۳۵ مهارت پرتکرار ساخته شد و ۲۹۰ یال با دست‌کم ۲۰ هم‌رخدادی باقی ماند. اندازه گره‌ها با فراوانی مهارت و ضخامت یال‌ها با تعداد هم‌رخدادی متناسب است. اندازه شبکه (نه تعداد گره، بلکه چگالی یال‌ها) نشان می‌دهد سبد مهارت‌های ایران «متمرکز» است: تعداد محدودی مهارت، بارها با هم تکرار می‌شوند.

**تفسیر جدول ۲۱.۲:** چهار جامعه مهارتی شناسایی شد:

1. **آفیس و نرم‌افزارهای مالی/اداری ایران (۱۲ مهارت):** Excel، Word، PowerPoint، Outlook، Access به‌همراه سپیدار، راهکاران و همکاران سیستم. این جامعه، هسته بازار اداری است.
2. **توسعه و زیرساخت (۱۱ مهارت):** SQL Server، HTML/CSS، Git، JavaScript، Network+، C#.
3. **طراحی و محتوای دیجیتال (۹ مهارت):** Photoshop، Illustrator، WordPress، Premiere، After Effects، Google Analytics.
4. **مهندسی و طراحی صنعتی (۳ مهارت):** AutoCAD، SolidWorks، CATIA.

نکته مهم: **nرم‌افزارهای بومی حسابداری ایران** در همان جامعه آفیس قرار گرفته‌اند، نه در جامعه توسعه؛ یعنی از دید بازار، این مهارت‌ها «ابزار اداری» هستند نه «فناوری». این یک تفاوت مهم با بازارهای جهانی است.

## ۲۱.۲. مرکزیت مهارت‌ها و تغییر آن در دو سال

</div>

```python
centrality_table = pd.DataFrame({
    'فراوانی آگهی': [graph.nodes[node]['frequency'] for node in graph.nodes()],
    'مرکزیت بینابینی': [betweenness[node] for node in graph.nodes()],
    'درجه وزنی': [weighted_degree[node] for node in graph.nodes()],
    'جامعه': [f'جامعه {community_of[node] + 1}' for node in graph.nodes()],
}, index=list(graph.nodes())).sort_values('درجه وزنی', ascending=False)
centrality_table['رتبه'] = range(1, len(centrality_table) + 1)
show_table('21.3', 'پانزده مهارت مرکزی شبکه', centrality_table.head(15),
           formats={'فراوانی آگهی': 'int', 'مرکزیت بینابینی': 'score', 'درجه وزنی': 'int', 'رتبه': 'int'},
           sample='شبکه کل دوره', period=False, index_name='مهارت')


def annual_degree(year):
    frame = skill_ads.merge(ads.loc[ads.jyear.eq(year), ['ad_id']], on='ad_id')
    counter = Counter()
    for _, group in frame.groupby('ad_id').skill:
        names = sorted(set(group))
        if 1 < len(names) <= 12:
            counter.update(combinations(names, 2))
    degree = {skill: 0 for skill in network_skills}
    for (first, second), weight in counter.items():
        degree[first] += weight
        degree[second] += weight
    return pd.Series(degree).sort_values(ascending=False), counter


degree_1401, pairs_1401 = annual_degree(1401)
degree_1402, pairs_1402 = annual_degree(1402)
rank_compare = pd.DataFrame({'درجه ۱۴۰۱': degree_1401, 'درجه ۱۴۰۲': degree_1402})
rank_compare['رتبه ۱۴۰۱'] = rank_compare['درجه ۱۴۰۱'].rank(ascending=False).astype(int)
rank_compare['رتبه ۱۴۰۲'] = rank_compare['درجه ۱۴۰۲'].rank(ascending=False).astype(int)
rank_compare['تغییر رتبه'] = rank_compare['رتبه ۱۴۰۱'] - rank_compare['رتبه ۱۴۰۲']
rank_compare['رشد درجه'] = (rank_compare['درجه ۱۴۰۲'] / rank_compare['درجه ۱۴۰۱'].replace(0, np.nan) - 1) * 100
rank_movers = rank_compare.reindex(rank_compare['درجه ۱۴۰۲'].sort_values(ascending=False).index).head(15)
show_table('21.4', 'تغییر مرکزیت مهارت‌ها بین دو سال', rank_movers,
           formats={'درجه ۱۴۰۱': 'int', 'درجه ۱۴۰۲': 'int', 'رتبه ۱۴۰۱': 'int', 'رتبه ۱۴۰۲': 'int',
                    'تغییر رتبه': 'signed', 'رشد درجه': 'pp'},
           sample='شبکه ۳۵ مهارت پرتکرار', period=False, index_name='مهارت')

fig, axes = plt.subplots(1, 2, figsize=(15, 6.2))
for ax, year, degree, yearly_pairs, panel in [(axes[0], 1401, degree_1401, pairs_1401, '۲'),
                                              (axes[1], 1402, degree_1402, pairs_1402, '۳')]:
    year_graph = nx.Graph()
    for skill, weight in degree.head(18).items():
        year_graph.add_node(skill, frequency=weight)
    for (first, second), weight in yearly_pairs.items():
        if first in year_graph and second in year_graph and weight >= 40:
            year_graph.add_edge(first, second, weight=weight)
    positions = nx.spring_layout(year_graph, weight='weight', seed=RANDOM_STATE, k=0.6, iterations=200)
    node_sizes = np.array([year_graph.nodes[node]['frequency'] for node in year_graph.nodes()])
    nx.draw_networkx_edges(year_graph, positions, ax=ax, alpha=0.22, edge_color='#8aa9a6')
    nx.draw_networkx_nodes(year_graph, positions, ax=ax, node_size=60 + 900 * (node_sizes / node_sizes.max()),
                           node_color=[PALETTE[community_of.get(node, 0) % len(PALETTE)] for node in year_graph.nodes()],
                           alpha=0.9, linewidths=0.6, edgecolors='white')
    nx.draw_networkx_labels(year_graph, positions, ax=ax, font_size=6.5,
                            labels={node: fa(node) for node in year_graph.nodes()})
    ax.set_axis_off()
    heading(ax, f'۲۱.{panel}', f'شبکه مهارت‌ها در سال {nf(year)}')
complete('21.2-21.3')

show_metrics('شبکه مهارت', [
    ('مهارت‌های شبکه', f'{graph.number_of_nodes()}'),
    ('جوامع', f'{len(communities)}'),
    ('مرکزی‌ترین مهارت', f'{centrality_table.index[0]}'),
    ('بیشترین رشد درجه', f'{rank_compare["رشد درجه"].idxmax()} ({rank_compare["رشد درجه"].max():+.0f}٪)'),
])
```

**تفسیر جدول‌های ۲۱.۳ و ۲۱.۴ و نمودارهای ۲۱.۲ و ۲۱.۳:** مرکزی‌ترین گره‌های شبکه، مهارت‌های آفیس‌اند: Excel با درجه وزنی ۲۶۸٬۵۰۵ و Word با ۲۵۳٬۶۸۳. اما **مرکزیت بینابینی** — توانایی اتصال جوامع — داستان دیگری دارد: WordPress (۰٫۲۳)، Visio (۰٫۱۶)، Outlook (۰٫۱۴) و Illustrator (۰٫۱۱) بالاترین مرکزیت بینابینی را دارند؛ یعنی این مهارت‌ها پل بین جامعه اداری، طراحی و فناوری‌اند.

در مقایسه دو سال، رتبه‌ها تقریباً ثابت مانده‌اند (بیشترین تغییر رتبه فقط ۲ پله است)، اما درجه همه مهارت‌ها رشد کرده است؛ بخشی از این رشد، رشد کل بازار آگهی‌هاست. بزرگ‌ترین رشد نسبی درجه: Google Analytics (+۲۴۱٪)، Adobe Premiere (+۶۹٪)، سپیدار (+۶۶٪)، WordPress (+۶۳٪) و AutoCAD (+۶۰٪).

**پیام برای کارجو:** اگر می‌خواهید بین حوزه‌ها حرکت کنید، مهارت‌های «پل» (WordPress، Visio، Excel پیشرفته) ارزش ارتباطی بالایی دارند؛ اگر می‌خواهید در یک حوزه عمیق شوید، مهارت‌های مرکزی همان جامعه را تقویت کنید.

# ۲۲. بسته پیشنهادی برای مدیر کسب‌وکار

این بخش یافته‌های پراکنده بخش‌های قبل را به تصمیم‌های اجرایی تبدیل می‌کند: چقدر حقوق بدهیم، چه زمانی آگهی منتشر کنیم، چه مهارتی را در آگهی برجسته کنیم و چه چیزی آگهی ما را از رقبا متمایز می‌کند.

## ۲۲.۱. بسته جبران خدمات: میانه و دامنه رقابتی

</div>

```python
benchmark_categories = paid.category_group.value_counts().head(12).index
salary_benchmark = (paid.loc[paid.category_group.isin(benchmark_categories)]
                    .groupby('category_group')
                    .agg(**{'تعداد آگهی': ('ad_id', 'size'),
                            'چارک اول': ('salary_mid', lambda s: s.quantile(.25)),
                            'میانه': ('salary_mid', 'median'),
                            'چارک سوم': ('salary_mid', lambda s: s.quantile(.75)),
                            'میانه ساعتی': ('hourly_income', 'median'),
                            'نرخ افشا': ('salary_disclosed', lambda s: s.mean() * 100)}))
salary_benchmark['فاصله از میانه کل'] = (salary_benchmark['میانه'] / paid.salary_mid.median() - 1) * 100
salary_benchmark['نرخ افشا'] = salary_benchmark.index.map(
    ads.groupby('category_group').salary_disclosed.mean() * 100)
salary_benchmark = salary_benchmark.sort_values('میانه', ascending=False)
show_table('22.1', 'بسته جبران خدمات پیشنهادی در دسته‌های شغلی پرتکرار', salary_benchmark,
           formats={'تعداد آگهی': 'int', 'چارک اول': 'money', 'میانه': 'money', 'چارک سوم': 'money',
                    'میانه ساعتی': 'hourly', 'نرخ افشا': 'pct', 'فاصله از میانه کل': 'pp'},
           sample=f'{len(paid):,} آگهی با حقوق اعلام‌شده', index_name='دسته شغلی',
           note='دامنه رقابتی = چارک اول تا سوم؛ میانه، سطح هدف برای نیمی از بازار است.')

seniority_province = paid.assign(province_group=paid.province_group)
benchmark_levels = (seniority_province.pivot_table(index='seniority', columns='province_group',
                                                   values='salary_mid', aggfunc='median')
                    .reindex(SENIORITY_ORDER))
benchmark_levels['تعداد آگهی'] = paid.groupby('seniority').size().reindex(SENIORITY_ORDER)
benchmark_levels['سطح نسبت به کارشناسی'] = (benchmark_levels['تهران'] /
                                            benchmark_levels.loc['کارشناسی', 'تهران'] * 100)
show_table('22.2', 'میانه حقوق بر پایه رده سازمانی و محل (میلیون تومان)', benchmark_levels,
           formats={'تهران': 'money', 'سایر استان‌ها': 'money', 'تعداد آگهی': 'int',
                    'سطح نسبت به کارشناسی': 'pct'},
           sample=f'{len(paid):,} آگهی با حقوق', period=False, index_name='رده سازمانی')
```

**تفسیر جدول ۲۲.۱:** این جدول، پاسخ مستقیم به پرسش اول پروژه است: برای هر دسته شغلی، دامنه رقابتی «چارک اول تا سوم» و سطح هدف «میانه». برای نمونه، برنامه‌نویسی دامنه ۱۳٫۵ تا ۲۲٫۵ با میانه ۱۷٫۵ میلیون دارد (+۵۹٪ بالاتر از میانه کل)؛ در مقابل، مسئول دفتر و کارگر ساده دامنه ۷ تا ۱۱ با میانه ۹ میلیون دارند (−۱۸٪).

قاعده تصمیم: اگر می‌خواهید «در بازار رقابتی» باشید، پیشنهاد را نزدیک میانه بگذارید؛ اگر می‌خواهید «متمایز» شوید، چارک سوم را هدف بگیرید و در عوض در شرایط احراز سخت‌گیری کمتری کنید. اگر بودجه محدود است، چارک اول فقط برای نقش‌های کم‌رقابت یا کارآموزی مناسب است، نه برای نقش‌های تخصصی.

## ۲۲.۲. تقویم استخدام: چه زمانی آگهی منتشر کنیم؟

</div>

```python
calendar_table = pd.DataFrame({
    'شاخص فصلی': seasonal_table['شاخص فصلی'],
    'میانه حقوق': paid.groupby('jmonth_num').salary_mid.median().rename(index=lambda value: J_MONTHS[value - 1]),
    'نرخ افشا': ads.groupby('jmonth_num').salary_disclosed.mean().mul(100).rename(index=lambda value: J_MONTHS[value - 1]),
    'میانگین آگهی ماهانه': seasonal_table['میانگین دو سال'],
}).reindex(seasonal_table.index)
calendar_table['رقابت نسبی'] = np.select([calendar_table['شاخص فصلی'].ge(115),
                                          calendar_table['شاخص فصلی'].ge(95)],
                                         ['بالا', 'متوسط'], default='پایین')
show_table('22.3', 'تقویم پیشنهادی انتشار آگهی بر پایه ماه شمسی', calendar_table,
           formats={'شاخص فصلی': 'pct', 'میانه حقوق': 'money', 'نرخ افشا': 'pct',
                    'میانگین آگهی ماهانه': 'int'},
           sample='میانگین دو سال', period=False, index_name='ماه شمسی',
           note='رقابت بالا یعنی تعداد آگهی منتشرشده در آن ماه بیش از میانگین است.')

fig, ax = plt.subplots(figsize=(12, 4.8))
index_values = seasonal_table['شاخص فصلی']
colors = [RED if value >= 115 else (GOLD if value >= 95 else TEAL) for value in index_values]
bars = ax.bar(range(len(index_values)), index_values, color=colors, width=0.65)
ax.axhline(100, color=DARK, linewidth=1, linestyle='--')
ax.set_xticks(range(len(index_values)))
ax.set_xticklabels([fa(month) for month in index_values.index], fontsize=8.5)
ax.set_ylabel(fa('شاخص فصلی (میانگین = ۱۰۰)'))
for bar, value in zip(bars, index_values):
    ax.text(bar.get_x() + bar.get_width() / 2, bar.get_height(), f'{value:.0f}', ha='center',
            va='bottom', fontsize=7.5)
heading(ax, '۲۲.۱', 'تقویم فصلی رقابت برای جذب نیرو')
complete('22.1')
```

**تفسیر جدول ۲۲.۳ و نمودار ۲۲.۱:** تقویم پیشنهادی انتشار، از شاخص فصلی بخش ۱۰ ساخته شده است. برای نقش‌های غیرفوری، ماه‌های **خرداد (۸۵٫۵٪)، شهریور (۸۸٫۲٪)، اردیبهشت (۹۱٫۲٪) و بهمن (۷۸٫۹٪)** کم‌رقابت‌ترین‌اند؛ آگهی در این ماه‌ها کمتر دیده می‌شود و شانس دیده‌شدن بیشتر است. برای نقش‌های فوری یا زمانی که حجم بالای نیروی جوینده مهم است، **دی (۱۲۳٪)، مهر (۱۱۴٫۷٪) و آبان (۱۱۲٫۱٪)** بهترین زمان‌اند.

توجه: «رقابت بالا» برای کارفرما بد نیست؛ یعنی نیروی جوینده بیشتری در بازار فعال است. انتخاب به هدف شما بستگی دارد: **دیده‌شدن آگهی** (ماه کم‌رقابت) یا **دسترسی به حجم بیشتر کاندیدا** (ماه پررقابت).

## ۲۲.۳. مهارت‌هایی که حقوق را بالا می‌برند

</div>

```python
premium_rows = []
top_skill_candidates = skills_long.skill.value_counts().head(20).index
for skill in top_skill_candidates:
    skill_ids = set(skills_long.loc[skills_long.skill.eq(skill), 'ad_id'])
    with_skill = paid.loc[paid.ad_id.isin(skill_ids)]
    if len(with_skill) < 300:
        continue
    main_category = with_skill.category_group.value_counts().idxmax()
    category_paid = paid.loc[paid.category_group.eq(main_category)]
    inside = category_paid.loc[category_paid.ad_id.isin(skill_ids), 'salary_mid']
    outside = category_paid.loc[~category_paid.ad_id.isin(skill_ids), 'salary_mid']
    if len(inside) < MIN_N or len(outside) < MIN_N:
        continue
    premium_rows.append({'مهارت': skill, 'دسته مقایسه': main_category, 'تعداد دارای مهارت': len(inside),
                         'میانه با مهارت': inside.median(), 'میانه بدون مهارت': outside.median(),
                         'صرفه در دسته': (np.exp(np.log(inside).mean() - np.log(outside).mean()) - 1) * 100})
skill_premium = pd.DataFrame(premium_rows).set_index('مهارت').sort_values('صرفه در دسته', ascending=False)
show_table('22.4', 'صرفه حقوقی مهارت‌ها در دسته شغلی خودشان', skill_premium.head(12),
           formats={'تعداد دارای مهارت': 'int', 'میانه با مهارت': 'money', 'میانه بدون مهارت': 'money',
                    'صرفه در دسته': 'pp'},
           sample=f'{len(paid):,} آگهی با حقوق', index_name='مهارت',
           note='مقایسه درونی هر دسته انجام شده تا اثر ترکیب شغلی تا حد ممکن کنترل شود؛ رابطه توصیفی است، نه علی.')

fig, ax = plt.subplots(figsize=(10.5, 5))
plot_premium = skill_premium.head(10).iloc[::-1]
bars = ax.barh(range(len(plot_premium)), plot_premium['صرفه در دسته'], color=TEAL, height=0.65)
ax.set_yticks(range(len(plot_premium)))
ax.set_yticklabels([fa(name) for name in plot_premium.index], fontsize=8.5)
ax.axvline(0, color=DARK, linewidth=0.9)
ax.set_xlabel(fa('اختلاف میانه حقوق با و بدون مهارت (٪)'))
for position, value in enumerate(plot_premium['صرفه در دسته']):
    ax.text(value + 0.4, position, f'{value:+.1f}٪', va='center', fontsize=8)
heading(ax, '۲۲.۲', 'بیشترین صرفه حقوقی مهارت‌ها در دسته خودشان')
complete('22.2')
```

**تفسیر جدول ۲۲.۴ و نمودار ۲۲.۲:** صرفه حقوقی مهارت‌ها با مقایسه میانگین لگاریتم حقوق «با مهارت» و «بدون مهارت» در همان دسته شغلی محاسبه شده است. بالاترین صرفه‌ها: **AutoCAD +۱۴٫۴٪**، **Microsoft Visio +۱۳٫۲٪**، **راهکاران +۱۱٫۸٪**، **Git +۱۱٫۴٪** و **همکاران سیستم +۸٫۲٪**. WordPress در دیجیتال مارکتینگ +۶٫۱٪ است.

این جدول برای نوشتن آگهی و مذاکره حقوق مفید است: اگر نقشی با AutoCAD یا Git می‌خواهید، بازار برای همان مهارت صرفه قابل توجهی پرداخت می‌کند؛ در مقابل، Excel و Word مهارت‌های پایه‌اند و به‌تنهایی صرفه نمی‌سازند. **توجه:** این رابطه توصیفی است؛ ترکیب مهارت با نقش و صنعت می‌تواند بخشی از صرفه را توضیح دهد.
